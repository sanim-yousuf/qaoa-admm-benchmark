# When Does QAOA Help Inside ADMM? A Matched-Budget Benchmark

This notebook reproduces every result of the manuscript: QAOA as the binary-subproblem solver of the alternating direction method of multipliers (ADMM), compared with classical solvers at a matched budget of objective evaluations. It covers:
- the scoring rules and angle-transfer protocols;
- a benign family and a planted (Wishart) family with a rugged landscape, with landscape diagnostics;
- classical baselines from simulated annealing to parallel tempering, tabu search with restarts, and SDP rounding with local search;
- fixed-budget scaling and time to solution;
- the ADMM penalty, the direct solution of the reduced problem, and a mixed-binary family without that reduction;
- validation on IBM quantum hardware;
- a correctness gate and a registry of every hypothesis test.

**Running.** Every experiment is cached to CSV or JSON. With `RESUME = True` (Section 9) and the outputs of a previous run attached, completed experiments are reloaded and only missing ones are computed. `QUICK_TEST = True` runs a reduced end-to-end check. Hardware jobs are submitted only when their switch is on, their results are missing, and the `IBM_QUANTUM_TOKEN` secret is available.

## 4. Problem families, exact statistics, and the QUBO-to-Ising conversion

Normalized gap $g_{\mathrm{unif}}=(C-C^*)/(\bar C-C^*)$: 0 at the optimum, 1 for a uniform guess. QAOA uses the cost Hamiltonian divided by the root-mean-square of its Pauli coefficients. `cost_vector_big` is a chunked float64 enumeration used for $n>16$.

```python
def generate_instance(n, seed):
    rng = np.random.default_rng(seed)
    Q = rng.uniform(-1, 1, size=(n, n)); Q = (Q + Q.T) / 2
    np.fill_diagonal(Q, rng.uniform(-1, 1, size=n))
    c = rng.uniform(-1, 1, size=n)
    B = rng.uniform(-0.7, 0.7, size=(n, n))
    b = rng.uniform(-1, 1, size=n)
    return Q, c, B, b

def generate_instance_sparse(n, seed, p_link):
    rng = np.random.default_rng(seed)
    mask = rng.random((n, n)) < p_link
    mask = np.triu(mask, 1); mask = mask + mask.T
    Q = rng.uniform(-1, 1, size=(n, n)) * mask; Q = (Q + Q.T) / 2
    np.fill_diagonal(Q, rng.uniform(-1, 1, size=n))
    c = rng.uniform(-1, 1, size=n)
    B = rng.uniform(-0.7, 0.7, size=(n, n)) * (rng.random((n, n)) < p_link)
    b = rng.uniform(-1, 1, size=n)
    return Q, c, B, b

def generate_maxcut_instance(n, p_edge, seed):
    G = nx.erdos_renyi_graph(n, p_edge, seed=seed)
    Qp = np.zeros((n, n)); h = np.zeros(n)
    for (i, j) in G.edges():
        Qp[i, j] = 1.0; Qp[j, i] = 1.0; h[i] -= 1.0; h[j] -= 1.0
    return Qp, h, G

def build_qubo_subproblem(x, lam, Q, B, b, rho):
    Qp = Q + (rho / 2.0) * (B.T @ B); Qp = (Qp + Qp.T) / 2.0
    h = B.T @ lam + rho * (B.T @ (x - b))
    return Qp, h

def x_update(z, lam, c, B, b, rho):
    return (c - lam - rho * (B @ z - b)) / (1.0 + rho)

def first_zupdate(n, seed, rho=2.0, gen="dense", p_link=None):
    Q, c, B, b = generate_instance(n, seed) if gen == "dense" else generate_instance_sparse(n, seed, p_link)
    return build_qubo_subproblem(np.zeros(n), np.zeros(n), Q, B, b, rho)

def qubo_cost(z, Qp, h):
    z = np.asarray(z, dtype=float); return float(z @ Qp @ z + h @ z)

def qubo_density(Qp, tol=1e-9):
    n = Qp.shape[0]; off = Qp - np.diag(np.diag(Qp))
    return float(np.sum(np.abs(off) > tol)) / (n * (n - 1))

def all_bits(n, dtype):
    idx = np.arange(2 ** n, dtype=np.int64)
    return ((idx[:, None] >> np.arange(n)) & 1).astype(dtype)

def z_from_index(k, n):
    return ((int(k) >> np.arange(n)) & 1).astype(float)

def z_to_index(z):
    return int(sum(int(v) << i for i, v in enumerate(np.asarray(z).astype(int))))

def cost_vector(Qp, h):
    # index k has z_i = bit i of k (Qiskit little-endian order)
    n = len(h); dt = np.float64 if n <= 16 else np.float32
    bits = all_bits(n, dt)
    return np.einsum("ij,jk,ik->i", bits, Qp.astype(dt), bits) + bits @ h.astype(dt)

def cost_vector_big(Qp, h, chunk=2 ** 18):
    n = len(h); N_ = 2 ** n; out = np.empty(N_, dtype=np.float64); sh = np.arange(n)
    for a in range(0, N_, chunk):
        idx = np.arange(a, min(a + chunk, N_), dtype=np.int64)
        Z = ((idx[:, None] >> sh) & 1).astype(np.float64)
        out[a:a + len(idx)] = np.einsum("ij,ij->i", Z @ Qp, Z) + Z @ h
    return out

def exact_stats(Qp, h, keep=False):
    Cv = cost_vector(Qp, h); k = int(np.argmin(Cv)); n = len(h)
    zs = z_from_index(k, n)
    out = {"Cstar": qubo_cost(zs, Qp, h), "zstar": zs, "Cmax": float(Cv.max()), "Cmean": float(Cv.mean())}
    if keep:
        out["C"] = Cv.astype(np.float64)
    return out

def exact_stats_big(Qp, h, keep=True):
    Cv = cost_vector_big(Qp, h); k = int(np.argmin(Cv)); n = len(h)
    out = {"Cstar": float(Cv[k]), "zstar": z_from_index(k, n), "Cmax": float(Cv.max()), "Cmean": float(Cv.mean())}
    if keep:
        out["C"] = Cv
    return out

def gaps(cost, st, tol=1e-7):
    g = float(cost) - st["Cstar"]
    if g <= tol * max(1.0, abs(st["Cstar"])):
        g = 0.0
    du, dr = st["Cmean"] - st["Cstar"], st["Cmax"] - st["Cstar"]
    return {"gap": g, "gap_unif": g / du if du > tol else np.nan,
            "gap_range": g / dr if dr > tol else np.nan, "hit": float(g == 0.0)}

def qubo_to_ising_coeffs(Qp, h, defective=False):
    # z_i = (1 - s_i)/2; each off-diagonal term feeds the linear terms of both spins.
    # defective=True reproduces the historical one-sided update (Section 22 only).
    n = len(h); off = 0.0; cs = np.zeros(n); css = np.zeros((n, n))
    for i in range(n):
        off += h[i] / 2.0 + Qp[i, i] / 2.0
        cs[i] += -h[i] / 2.0 - Qp[i, i] / 2.0
    for i in range(n):
        for j in range(n):
            if i == j:
                continue
            off += Qp[i, j] / 4.0
            cs[i] += -Qp[i, j] / 4.0
            if not defective:
                cs[j] += -Qp[i, j] / 4.0
            css[i, j] += Qp[i, j] / 4.0
    return cs, css, off

def pauli_op(cs, css):
    n = len(cs); terms = []
    for i in range(n):
        if abs(cs[i]) > 1e-12:
            lab = ["I"] * n; lab[n - 1 - i] = "Z"; terms.append(("".join(lab), cs[i]))
    for i in range(n):
        for j in range(i + 1, n):
            cij = css[i, j] + css[j, i]
            if abs(cij) > 1e-12:
                lab = ["I"] * n; lab[n - 1 - i] = "Z"; lab[n - 1 - j] = "Z"; terms.append(("".join(lab), cij))
    return SparsePauliOp.from_list(terms or [("I" * n, 0.0)])

def qubo_to_ising_paulis(Qp, h):
    cs, css, off = qubo_to_ising_coeffs(Qp, h); return pauli_op(cs, css), off

def qubo_to_ising_paulis_DEFECTIVE(Qp, h):
    cs, css, off = qubo_to_ising_coeffs(Qp, h, defective=True); return pauli_op(cs, css), off

def ising_scale(H):
    c = np.real(H.coeffs); c = c[np.abs(c) > 1e-12]
    return float(np.sqrt(np.mean(c ** 2))) if len(c) else 1.0

def ising_scale_offset(Qp, h):
    cs, css, off = qubo_to_ising_coeffs(Qp, h)
    n = len(h); iu = np.triu_indices(n, 1)
    v = np.concatenate([cs, (css + css.T)[iu]]); v = v[np.abs(v) > 1e-12]
    return (float(np.sqrt(np.mean(v ** 2))) if len(v) else 1.0), off

def scaled_diagonal(Qp, h, Cv, normalize=True):
    sc, off = ising_scale_offset(Qp, h)
    return (Cv - off) / (sc if normalize else 1.0)

def qaoa_ansatz(Qp, h, reps, converter=None, normalize=True, measure=True):
    H, off = (converter or qubo_to_ising_paulis)(Qp, h)
    scale = ising_scale(H) if normalize else 1.0
    ans = QAOAAnsatz(cost_operator=H * (1.0 / scale), reps=reps).decompose(reps=2)
    if measure:
        ans.measure_all()
    return ans, scale, off, H

def qaoa_probs_fast(d_scaled, params, p, n):
    # parameter order as QAOAAnsatz: [beta_1..beta_p, gamma_1..gamma_p]
    psi = np.full(2 ** n, 2 ** (-n / 2), dtype=complex)
    for l in range(p):
        psi = psi * np.exp(-1j * params[p + l] * d_scaled)
        psi = psi.reshape([2] * n)
        cb, sb = np.cos(params[l]), -1j * np.sin(params[l])
        for ax in range(n):
            psi = cb * psi + sb * np.flip(psi, axis=ax)
        psi = psi.reshape(-1)
    return np.abs(psi) ** 2

class FastSampler:
    # exact QAOA distribution with multinomial shot sampling (verified against Qiskit in Section 5)
    def __init__(self, d_scaled, reps, seed):
        self.d, self.p = d_scaled, reps; self.n = int(round(np.log2(len(d_scaled))))
        self.rng = np.random.default_rng(seed)
    def probs(self, params):
        pr = qaoa_probs_fast(self.d, np.asarray(params, float), self.p, self.n); return pr / pr.sum()
    def sample(self, params, shots):
        ks = self.rng.choice(len(self.d), size=int(shots), p=self.probs(params))
        u, c_ = np.unique(ks, return_counts=True); return u.astype(np.int64), c_.astype(np.int64)

def counts_to_arrays(counts):
    ks = np.fromiter((int(b.replace(" ", ""), 2) for b in counts), dtype=np.int64, count=len(counts))
    cts = np.fromiter(counts.values(), dtype=np.int64, count=len(counts))
    return ks, cts

def counts_to_probs(counts, n):
    ks, cts = counts_to_arrays(counts); p = np.zeros(2 ** n); np.add.at(p, ks, cts); return p / p.sum()
print("Building blocks defined.")
```

## 5. Correctness gate

Matrix level (diagonal of $H_C$ plus offset equals the QUBO cost), circuit level (cost layer and $p=2$ ansatz against exact references), sampling level (sampled mean cost against the exact mean), the fast sampler used for $n>12$, and the chunked cost enumeration. The notebook halts on any failure.

```python
def run_correctness_gate():
    rows = []
    for fam in ("ADMM", "MaxCut"):
        for n in (4, 6, 8, 12, 16):
            Qp, h = first_zupdate(n, 1) if fam == "ADMM" else generate_maxcut_instance(n, 0.5, 1)[:2]
            H, off = qubo_to_ising_paulis(Qp, h)
            d = np.real(H.to_matrix(sparse=True).diagonal())
            err = float(np.max(np.abs(d + off - cost_vector(Qp, h))))
            rows.append({"check": "matrix diagonal = QUBO cost", "family": fam, "n": n, "value": err, "pass": err < 1e-9})
        for n in (4, 8, 12):
            Qp, h = first_zupdate(n, 2) if fam == "ADMM" else generate_maxcut_instance(n, 0.5, 2)[:2]
            ans, scale, off, H = qaoa_ansatz(Qp, h, reps=1, measure=False)
            d = (cost_vector(Qp, h) - off) / scale
            gamma = 0.73
            sv = Statevector(ans.assign_parameters([0.0, gamma])).data
            fid = abs(np.vdot(np.exp(-1j * gamma * d) / 2 ** (n / 2), sv))
            rows.append({"check": "cost layer = exp(-i gamma H)", "family": fam, "n": n, "value": 1 - fid, "pass": 1 - fid < 1e-9})
            ans2, _, _, _ = qaoa_ansatz(Qp, h, reps=2, measure=False)
            prm = np.random.default_rng(n).uniform(-1, 1, 4)
            psi = Statevector(ans2.assign_parameters(prm)).data
            probs_err = float(np.max(np.abs(np.abs(psi) ** 2 - qaoa_probs_fast(d, prm, 2, n))))
            rows.append({"check": "p=2 ansatz = fast simulator", "family": fam, "n": n, "value": probs_err, "pass": probs_err < 1e-10})
            d2 = scaled_diagonal(Qp, h, cost_vector(Qp, h).astype(np.float64))
            err2 = float(np.max(np.abs(d2 - d)))
            rows.append({"check": "scaled diagonal (no Pauli build) = Qiskit", "family": fam, "n": n, "value": err2, "pass": err2 < 1e-9})
    Qp, h = first_zupdate(8, 3); Cv = cost_vector(Qp, h)
    ans, scale, off, _ = qaoa_ansatz(Qp, h, reps=2)
    prm = np.array([0.4, -0.3, 0.8, 0.5])
    exact = float(qaoa_probs_fast((Cv - off) / scale, prm, 2, 8) @ Cv)
    counts = SamplerV2(seed=7).run([(ans, prm)], shots=40000).result()[0].data.meas.get_counts()
    ks, cts = counts_to_arrays(counts)
    emp = float((Cv[ks] * cts).sum() / cts.sum())
    se = float(np.sqrt(((Cv[ks] - emp) ** 2 * cts).sum() / cts.sum() / cts.sum()))
    rows.append({"check": "sampled mean = exact mean (bit order)", "family": "ADMM", "n": 8,
                 "value": abs(emp - exact) / se, "pass": abs(emp - exact) < 5 * se})
    ks, cts = FastSampler((Cv - off) / scale, 2, 7).sample(prm, 40000)
    emp = float((Cv[ks] * cts).sum() / cts.sum()); se = float(np.sqrt(((Cv[ks] - emp) ** 2 * cts).sum() / cts.sum() / cts.sum()))
    rows.append({"check": "fast sampler mean = exact mean", "family": "ADMM", "n": 8,
                 "value": abs(emp - exact) / se, "pass": abs(emp - exact) < 5 * se})
    Qp, h = first_zupdate(16, 4); err = float(np.max(np.abs(cost_vector_big(Qp, h, chunk=4096) - cost_vector(Qp, h))))
    rows.append({"check": "chunked enumeration = direct enumeration", "family": "ADMM", "n": 16, "value": err, "pass": err < 1e-9})
    df = pd.DataFrame(rows)
    print(df.to_string(index=False))
    assert df["pass"].all(), "Correctness gate failed."
    print("\nCorrectness gate passed.")
    return df

df_gate = run_correctness_gate(); df_gate.to_csv("S0_correctness_gate.csv", index=False)
```

## 6. Solvers under a matched objective-evaluation budget

One evaluation is one cost query; for QAOA, one shot. Every QAOA shot is logged. Primary score: the best bitstring sampled anywhere within the budget (the rule applied to all solvers). Secondary score: the best bitstring of the final read-out. Budget left by early optimizer termination is spent sampling at the final angles. `sim="fast"` uses the exact-distribution sampler of Section 5. Tabu search and local search charge $n$ evaluations per full neighborhood scan.

```python
def _aggregate(cs, cts, objective="mean", alpha=0.1):
    if objective == "cvar":
        o = np.argsort(cs); cs, cts = cs[o], cts[o]
        k = max(1, int(np.ceil(alpha * cts.sum())))
        before = np.cumsum(cts) - cts
        take = np.minimum(cts, np.maximum(0, k - before))
        return float((cs * take).sum() / k)
    return float((cs * cts).sum() / cts.sum())

class ShotLog:
    def __init__(self, Cv):
        self.C, self.used, self.best, self.best_k = Cv, 0, np.inf, None
    def add_arrays(self, ks, cts, shots):
        cs = self.C[ks]; self.used += int(shots)
        i = int(np.argmin(cs))
        if cs[i] < self.best:
            self.best, self.best_k = float(cs[i]), int(ks[i])
        return ks, cs, cts
    def add(self, counts, shots):
        ks, cts = counts_to_arrays(counts); return self.add_arrays(ks, cts, shots)

def run_qaoa(Qp, h, Cv=None, reps=2, shots=32, maxiter=24, seed=0, optimizer="COBYLA", objective="mean",
             cvar_alpha=0.1, init="random", x0=None, readout_shots=None, budget=None, spend_remaining=True,
             normalize=True, converter=None, noise_model=None, rhobeg=1.0, tol=1e-4, sim="qiskit"):
    n = len(h); Cv = cost_vector(Qp, h).astype(np.float64) if Cv is None else Cv
    if sim == "fast":
        fs = FastSampler(scaled_diagonal(Qp, h, Cv, normalize), reps, seed); npar = 2 * reps
    else:
        ans, scale, _, _ = qaoa_ansatz(Qp, h, reps, converter, normalize)
        sampler = seeded_sampler(seed, noise_model); npar = ans.num_parameters
    R = shots if readout_shots is None else readout_shots
    B = shots * maxiter if budget is None else budget
    opt_budget = B - R
    log = ShotLog(Cv); rng = np.random.default_rng(seed)
    st = {"last": np.inf, "nfev": 0}

    def sample(params, k):
        if sim == "fast":
            ks, cts = fs.sample(params, k); return log.add_arrays(ks, cts, k)
        counts = sampler.run([(ans, params)], shots=int(k)).result()[0].data.meas.get_counts()
        return log.add(counts, k)

    def obj(params):
        if log.used + shots > opt_budget:
            return st["last"]
        _, cs, cts = sample(params, shots); st["nfev"] += 1
        st["last"] = _aggregate(cs, cts, objective, cvar_alpha); return st["last"]

    if init == "given" and x0 is not None:
        xi = np.asarray(x0, float).copy()
    elif init == "linear_ramp":
        xi = np.concatenate([np.linspace(0.5, 0.5 / reps, reps), np.linspace(0.5 / reps, 0.5, reps)])
    else:
        xi = rng.uniform(-np.pi / 4, np.pi / 4, npar)
    max_evals = opt_budget // shots if shots > 0 else 0
    xbest = xi
    if max_evals >= 1:
        if optimizer == "COBYLA":
            xbest = minimize(obj, xi, method="COBYLA",
                             options={"maxiter": max(max_evals, npar + 2), "rhobeg": rhobeg, "tol": tol}).x
        elif optimizer == "Nelder-Mead":
            xbest = minimize(obj, xi, method="Nelder-Mead",
                             options={"maxfev": max(max_evals, npar + 2), "xatol": 1e-3, "fatol": 1e-3}).x
        elif optimizer == "SPSA":
            a, c0, A, al, ga = 0.2, 0.1, max(1, max_evals // 10), 0.602, 0.101
            x = xi.copy(); fbest = np.inf; k = 0
            while log.used + 2 * shots <= opt_budget:
                ak, ck = a / ((k + 1 + A) ** al), c0 / ((k + 1) ** ga)
                dlt = rng.choice([-1.0, 1.0], size=npar)
                fp, fm_ = obj(x + ck * dlt), obj(x - ck * dlt)
                x = x - ak * ((fp - fm_) / (2 * ck)) * dlt
                if 0.5 * (fp + fm_) < fbest:
                    fbest, xbest = 0.5 * (fp + fm_), x.copy()
                k += 1
        else:
            raise ValueError(optimizer)
    ro_best = ro_mean = np.nan; ro_k = None
    if R > 0:
        ks, cs, cts = sample(xbest, R); i = int(np.argmin(cs))
        ro_best, ro_k, ro_mean = float(cs[i]), int(ks[i]), float((cs * cts).sum() / cts.sum())
    used_before_topup = log.used
    if spend_remaining and B - log.used > 0:
        sample(xbest, B - log.used)
    return {"best": log.best, "z": z_from_index(log.best_k, n), "readout_best": ro_best,
            "readout_z": z_from_index(ro_k, n) if ro_k is not None else None, "readout_mean": ro_mean,
            "evals": log.used, "evals_before_topup": used_before_topup, "nfev": st["nfev"],
            "x": np.asarray(xbest, float), "final_obj": st["last"]}

def run_qaoa_fixed(Qp, h, params, Cv=None, reps=2, budget=768, seed=0, normalize=True, noise_model=None,
                   converter=None, sim="qiskit"):
    n = len(h); Cv = cost_vector(Qp, h).astype(np.float64) if Cv is None else Cv
    log = ShotLog(Cv)
    if sim == "fast":
        ks, cts = FastSampler(scaled_diagonal(Qp, h, Cv, normalize), reps, seed).sample(params, budget)
        ks, cs, cts = log.add_arrays(ks, cts, budget)
    else:
        ans, _, _, _ = qaoa_ansatz(Qp, h, reps, converter, normalize)
        counts = seeded_sampler(seed, noise_model).run([(ans, np.asarray(params))], shots=int(budget)).result()[0].data.meas.get_counts()
        ks, cs, cts = log.add(counts, budget)
    return {"best": log.best, "z": z_from_index(log.best_k, n), "readout_best": log.best,
            "readout_z": z_from_index(log.best_k, n), "readout_mean": float((cs * cts).sum() / cts.sum()),
            "evals": int(budget), "evals_before_topup": int(budget), "nfev": 0, "x": np.asarray(params), "final_obj": np.nan}

def run_qaoa_multistart(Qp, h, Cv=None, n_restarts=4, reps=2, shots=16, budget=768, mode="pooled",
                        readout_shots=16, seed=0, **kw):
    # pooled: each restart has its own read-out; best: one read-out at the restart with the lowest final estimate
    n = len(h); Cv = cost_vector(Qp, h).astype(np.float64) if Cv is None else Cv
    runs = []
    if mode == "pooled":
        per = budget // n_restarts
        for r in range(n_restarts):
            runs.append(run_qaoa(Qp, h, Cv, reps=reps, shots=shots, budget=per, readout_shots=readout_shots,
                                 seed=seed * 100 + r, spend_remaining=True, **kw))
        b = min(runs, key=lambda d: d["best"]); ro = min(runs, key=lambda d: d["readout_best"])
        return {"best": b["best"], "z": b["z"], "readout_best": ro["readout_best"], "readout_z": ro["readout_z"],
                "readout_mean": float(np.mean([d["readout_mean"] for d in runs])),
                "evals": sum(d["evals"] for d in runs), "evals_before_topup": sum(d["evals_before_topup"] for d in runs),
                "nfev": sum(d["nfev"] for d in runs), "x": b["x"], "final_obj": np.nan}
    per = ((budget - readout_shots) // n_restarts) // shots * shots
    for r in range(n_restarts):
        runs.append(run_qaoa(Qp, h, Cv, reps=reps, shots=shots, budget=per, readout_shots=0,
                             seed=seed * 100 + r, spend_remaining=False, **kw))
    sel = min(runs, key=lambda d: d["final_obj"])
    norm = kw.get("normalize", True)
    ro = run_qaoa_fixed(Qp, h, sel["x"], Cv, reps, readout_shots, seed * 100 + 98, norm)
    used = sum(d["evals"] for d in runs) + readout_shots
    extra = [run_qaoa_fixed(Qp, h, sel["x"], Cv, reps, budget - used, seed * 100 + 99, norm)] if budget - used > 0 else []
    allr = runs + [ro] + extra; b = min(allr, key=lambda d: d["best"])
    return {"best": b["best"], "z": b["z"], "readout_best": ro["best"], "readout_z": ro["z"],
            "readout_mean": ro["readout_mean"], "evals": budget, "evals_before_topup": used,
            "nfev": sum(d["nfev"] for d in runs), "x": sel["x"], "final_obj": sel["final_obj"]}

def run_sa(Qp, h, budget, seed=0, T0=2.0, Tmin=0.01):
    n = len(h); rng = np.random.default_rng(seed)
    z = rng.integers(0, 2, size=n).astype(float); cur = qubo_cost(z, Qp, h)
    bz, bc = z.copy(), cur; steps = max(int(budget), 1)
    for s in range(steps):
        T = T0 * ((Tmin / T0) ** (s / max(steps - 1, 1)))
        i = rng.integers(0, n); zn = z.copy(); zn[i] = 1.0 - zn[i]
        nc = qubo_cost(zn, Qp, h); dlt = nc - cur
        if dlt <= 0 or rng.random() < np.exp(-dlt / max(T, 1e-9)):
            z, cur = zn, nc
            if cur < bc:
                bz, bc = z.copy(), cur
    return {"best": bc, "z": bz, "evals": steps}

def run_sa_restarts(Qp, h, budget, seed=0, restarts=4, T0=2.0, Tmin=0.01):
    per = int(budget) // restarts; runs = [run_sa(Qp, h, per, seed * 100 + r, T0, Tmin) for r in range(restarts)]
    b = min(runs, key=lambda d: d["best"]); return {"best": b["best"], "z": b["z"], "evals": per * restarts}

def _flip_deltas(Qp, h, z, dQ):
    s = 1.0 - 2.0 * z
    return s * (2.0 * (Qp @ z) + h) + dQ

def run_tabu(Qp, h, budget, seed=0, tenure=None):
    n = len(h); rng = np.random.default_rng(seed); dQ = np.diag(Qp).copy()
    z = rng.integers(0, 2, size=n).astype(float); cur = qubo_cost(z, Qp, h); used = 1
    bz, bc = z.copy(), cur; tabu = np.zeros(n, dtype=int); ten = tenure or max(2, n // 4); it = 0
    while used + n <= budget:
        d = _flip_deltas(Qp, h, z, dQ); used += n; it += 1
        ok = (tabu <= it) | (cur + d < bc - 1e-12)
        if not ok.any():
            ok[:] = True
        i = int(np.argmin(np.where(ok, d, np.inf)))
        z[i] = 1.0 - z[i]; cur += d[i]; tabu[i] = it + ten
        if cur < bc - 1e-12:
            bz, bc = z.copy(), cur
    return {"best": qubo_cost(bz, Qp, h), "z": bz, "evals": used}

def run_local_search(Qp, h, budget, seed=0):
    n = len(h); rng = np.random.default_rng(seed); dQ = np.diag(Qp).copy()
    z = rng.integers(0, 2, size=n).astype(float); cur = qubo_cost(z, Qp, h); used = 1
    bz, bc = z.copy(), cur
    while used + n <= budget:
        d = _flip_deltas(Qp, h, z, dQ); used += n; i = int(np.argmin(d))
        if d[i] < -1e-12:
            z[i] = 1.0 - z[i]; cur += d[i]
            if cur < bc - 1e-12:
                bz, bc = z.copy(), cur
        elif used + 1 <= budget:
            z = rng.integers(0, 2, size=n).astype(float); cur = qubo_cost(z, Qp, h); used += 1
            if cur < bc - 1e-12:
                bz, bc = z.copy(), cur
    return {"best": qubo_cost(bz, Qp, h), "z": bz, "evals": used}

def run_random(Qp, h, budget, seed=0, Cv=None):
    n = len(h); rng = np.random.default_rng(seed)
    if Cv is not None:
        ks = rng.integers(0, 2 ** n, size=int(budget)); i = int(np.argmin(Cv[ks]))
        return {"best": float(Cv[ks[i]]), "z": z_from_index(ks[i], n), "evals": int(budget)}
    Z = rng.integers(0, 2, size=(int(budget), n)).astype(float)
    cs = np.einsum("ij,jk,ik->i", Z, Qp, Z) + Z @ h; i = int(np.argmin(cs))
    return {"best": float(cs[i]), "z": Z[i], "evals": int(budget)}

def _hyperplane_round(Ymat, trials, rng):
    Ymat = (Ymat + Ymat.T) / 2; w, V = np.linalg.eigh(Ymat); V = V @ np.diag(np.sqrt(np.clip(w, 0, None)))
    R = rng.normal(size=(V.shape[1], int(trials)))
    S = np.sign(V @ R); S[S == 0] = 1.0
    return S

def run_sdp(Qp, h, budget, seed=0):
    # general-QUBO SDP via auxiliary-spin lifting; returns best rounding and the relaxation lower bound
    n = len(h); cs, css, off = qubo_to_ising_coeffs(Qp, h); m = n + 1
    J = np.zeros((m, m)); J[0, 1:] = cs; J[1:, 0] = cs; J[1:, 1:] = css + css.T
    iu = np.triu_indices(m, 1); Jup = J[iu]
    t0 = time.perf_counter()
    Y = cp.Variable((m, m), PSD=True)
    val = cp.sum(cp.multiply(np.triu(J, 1), (1 - Y))) / 2
    prob = cp.Problem(cp.Maximize(val), [cp.diag(Y) == 1]); prob.solve(solver=cp.SCS, verbose=False)
    sdp_t = time.perf_counter() - t0
    S = _hyperplane_round(Y.value, budget, np.random.default_rng(seed))
    S = S * np.where(S[0] < 0, -1.0, 1.0)[None, :]
    Z = ((1 - S[1:]) / 2).T
    costs = np.einsum("ij,jk,ik->i", Z, Qp, Z) + Z @ h; i = int(np.argmin(costs))
    lower = off + Jup.sum() - 2 * prob.value
    return {"best": float(costs[i]), "z": Z[i], "evals": int(budget), "bound": float(lower), "sdp_time": sdp_t}

def run_gw(G, n, budget, seed=0):
    Y = cp.Variable((n, n), PSD=True)
    edges = list(G.edges())
    if not edges:
        return {"best": 0.0, "z": np.zeros(n), "evals": int(budget), "bound_cut": 0.0}
    val = cp.sum([(1 - Y[i, j]) / 2 for (i, j) in edges])
    prob = cp.Problem(cp.Maximize(val), [cp.diag(Y) == 1]); prob.solve(solver=cp.SCS, verbose=False)
    S = _hyperplane_round(Y.value, budget, np.random.default_rng(seed)); Z = ((1 - S) / 2).T
    E = np.array(edges); cuts = (Z[:, E[:, 0]] != Z[:, E[:, 1]]).sum(1); i = int(np.argmax(cuts))
    return {"best": -float(cuts[i]), "z": Z[i], "evals": int(budget), "bound_cut": float(prob.value)}

def spectral_relaxation(G, n):
    L = nx.laplacian_matrix(G, nodelist=range(n)).toarray().astype(float)
    v = np.linalg.eigh(L)[1][:, -1]; v = v / (np.max(np.abs(v)) + 1e-12)
    return np.clip((1 + v) / 2, 1e-3, 1 - 1e-3)

def run_warmstart_qaoa(Qp, h, G, Cv=None, reps=2, shots=32, maxiter=24, seed=0):
    n = len(h); Cv = cost_vector(Qp, h).astype(np.float64) if Cv is None else Cv
    cs, css, _ = qubo_to_ising_coeffs(Qp, h); sc = ising_scale(pauli_op(cs, css)); cs, css = cs / sc, css / sc
    th = 2 * np.arcsin(np.sqrt(spectral_relaxation(G, n)))
    bp, gp = ParameterVector("b", reps), ParameterVector("g", reps); qc = QuantumCircuit(n)
    for i in range(n):
        qc.ry(th[i], i)
    for l in range(reps):
        for i in range(n):
            if abs(cs[i]) > 1e-12: qc.rz(2 * gp[l] * cs[i], i)
        for i in range(n):
            for j in range(i + 1, n):
                cij = css[i, j] + css[j, i]
                if abs(cij) > 1e-12: qc.rzz(2 * gp[l] * cij, i, j)
        for i in range(n):
            qc.ry(-th[i], i); qc.rz(2 * bp[l], i); qc.ry(th[i], i)
    qc.measure_all()
    sampler = SamplerV2(seed=seed); B = shots * maxiter; opt_b = B - shots
    log = ShotLog(Cv); st = {"last": np.inf}
    def obj(p):
        if log.used + shots > opt_b: return st["last"]
        counts = sampler.run([(qc, p)], shots=shots).result()[0].data.meas.get_counts()
        _, c_, t_ = log.add(counts, shots); st["last"] = _aggregate(c_, t_); return st["last"]
    x0 = np.random.default_rng(seed).uniform(-np.pi / 4, np.pi / 4, 2 * reps)
    xb = minimize(obj, x0, method="COBYLA", options={"maxiter": max(opt_b // shots, 2 * reps + 2)}).x
    counts = sampler.run([(qc, xb)], shots=shots).result()[0].data.meas.get_counts(); ks, c_, _ = log.add(counts, shots)
    ro = float(c_.min())
    if B - log.used > 0:
        log.add(sampler.run([(qc, xb)], shots=B - log.used).result()[0].data.meas.get_counts(), B - log.used)
    return {"best": log.best, "z": z_from_index(log.best_k, n), "readout_best": ro, "evals": log.used}
print("Solvers defined.")
```

## 7. Statistics and the test registry

Unit of analysis: the instance. Paired comparisons use the Wilcoxon signed-rank test with an exact null distribution computed on the non-zero differences (midranks for ties), the matched-pairs rank-biserial effect $r$ ($r<0$: first-named method better), and bootstrap 95% intervals. Tests on identical data are registered once. Equivalence is assessed with a paired bootstrap 90% interval against a pre-specified margin (two one-sided tests at level 0.05).

```python
REGISTRY = []; _SIG = {}

def fmtP(P):
    if P is None or not np.isfinite(P): return "n/a"
    return "<1e-10" if P < 1e-10 else f"{P:.3g}"

def bootstrap_ci(x, n_boot=5000, seed=0):
    x = np.asarray(x, float); x = x[np.isfinite(x)]
    if len(x) == 0: return np.nan, np.nan, np.nan
    rng = np.random.default_rng(seed); m = x[rng.integers(0, len(x), (n_boot, len(x)))].mean(1)
    return float(x.mean()), float(np.percentile(m, 2.5)), float(np.percentile(m, 97.5))

def _rb(d):
    d = d[np.abs(d) > 1e-12]
    if len(d) == 0: return 0.0
    r = rankdata(np.abs(d)); return float((r[d > 0].sum() - r[d < 0].sum()) / r.sum())

def wilcoxon_exact(d):
    # exact two-sided signed-rank test on non-zero differences, conditional on the tie pattern
    d = np.asarray(d, float); d = d[np.abs(d) > 1e-12]; n = len(d)
    if n == 0: return 1.0
    r2 = np.rint(2 * rankdata(np.abs(d))).astype(int); T = int(r2.sum())
    dist = np.zeros(T + 1); dist[0] = 1.0
    for v in r2:
        dist[v:] = dist[v:] + dist[:-v].copy() if v > 0 else dist
    dist /= dist.sum(); w = int(r2[d > 0].sum())
    lo, hi = dist[:w + 1].sum(), dist[w:].sum()
    return float(min(1.0, 2 * min(lo, hi)))

def paired(a, b, label=None, family=None, kind="confirmatory", n_boot=2000, record=True):
    a, b = np.asarray(a, float), np.asarray(b, float); ok = np.isfinite(a) & np.isfinite(b); a, b = a[ok], b[ok]
    d = a - b; nz = int(np.sum(np.abs(d) > 1e-12))
    P = wilcoxon_exact(d)
    rng = np.random.default_rng(0); idx = rng.integers(0, len(d), (n_boot, len(d)))
    rs = np.array([_rb(d[i]) for i in idx]) if len(d) else np.array([0.0])
    md_, mlo, mhi = bootstrap_ci(d)
    out = {"label": label, "family": family, "kind": kind, "n": len(d), "n_nonzero": nz, "P": float(P),
           "r": _rb(d), "r_lo": float(np.percentile(rs, 2.5)), "r_hi": float(np.percentile(rs, 97.5)),
           "mean_diff": md_, "md_lo": mlo, "md_hi": mhi}
    if label and record:
        sig = hashlib.sha1((np.round(d, 12) + 0.0).tobytes()).hexdigest()  # + 0.0 maps -0.0 to 0.0
        if sig in _SIG:
            REGISTRY[_SIG[sig]]["also_reported_as"] = (REGISTRY[_SIG[sig]].get("also_reported_as", "") + "; " + label).strip("; ")
        else:
            _SIG[sig] = len(REGISTRY)
            REGISTRY.append({k: out[k] for k in ("label", "family", "kind", "n", "n_nonzero", "P", "r")})
    return out

def register(label, family, kind, n, n_nonzero, P, r):
    REGISTRY.append({"label": label, "family": family, "kind": kind, "n": n, "n_nonzero": n_nonzero, "P": P, "r": r})

def equivalence(a, b, margin, n_boot=5000, seed=0):
    d = np.asarray(a, float) - np.asarray(b, float); d = d[np.isfinite(d)]
    rng = np.random.default_rng(seed); m = d[rng.integers(0, len(d), (n_boot, len(d)))].mean(1)
    lo, hi = np.percentile(m, 5), np.percentile(m, 95)
    return {"mean_diff": float(d.mean()), "ci90_lo": float(lo), "ci90_hi": float(hi), "margin": margin,
            "equivalent": bool(lo > -margin and hi < margin)}

def holm(ps):
    ps = np.asarray(ps, float); o = np.argsort(ps); m = len(ps); adj = np.empty(m); run = 0.0
    for rank, i in enumerate(o):
        run = max(run, (m - rank) * ps[i]); adj[i] = min(run, 1.0)
    return adj

def show_tests(rows, holm_within=True):
    df = pd.DataFrame(rows)
    if holm_within and len(df):
        df["P_holm"] = holm(df["P"].values)
    for _, t in df.iterrows():
        ph = f"  P_holm={fmtP(t['P_holm'])}" if "P_holm" in df else ""
        print(f"  {t['label']:<60s} r={t['r']:+.3f} [{t['r_lo']:+.2f},{t['r_hi']:+.2f}]  "
              f"P={fmtP(t['P'])}{ph}  non-tied={t['n_nonzero']}/{t['n']}")
    return df

def spearman_ci(x, y, n_boot=2000, seed=0):
    x, y = np.asarray(x, float), np.asarray(y, float); ok = np.isfinite(x) & np.isfinite(y); x, y = x[ok], y[ok]
    if np.std(y) == 0 or np.std(x) == 0: return 0.0, 1.0, np.nan, np.nan
    rho, P = spearmanr(x, y); rng = np.random.default_rng(seed); bs = []
    for _ in range(n_boot):
        i = rng.integers(0, len(x), len(x))
        if np.std(x[i]) > 0 and np.std(y[i]) > 0: bs.append(spearmanr(x[i], y[i])[0])
    return float(rho), float(P), float(np.percentile(bs, 2.5)), float(np.percentile(bs, 97.5))

def tvd(p, q): return 0.5 * float(np.abs(p - q).sum())
def hellinger_fidelity(p, q): return float(np.sum(np.sqrt(p * q)) ** 2)

_d = np.array([0.31, -0.12, 0.23, 0.54, 0.45, 0.26, -0.05, 0.67, 0.18, 0.79, 0.2, 0.9])
assert abs(wilcoxon_exact(_d) - wilcoxon(_d, method="exact").pvalue) < 1e-12
print("Statistics utilities ready.")
```

## 8. ADMM with convergence and problem-level metrics

Per iteration: best-sample and read-out gaps, primal residual, evaluations, the iterate index, and its feasible objective $F(z)$. Per run: final residual, best feasible objective over the iterates, its normalized gap to the optimum of the mixed-binary problem (by enumeration of $z$), and whether that optimum was found. `exact_trajectory` and `score_on_trajectory` score every solver on identical subproblems.

```python
def mbo_feasible_values(Q, c, B, b):
    n = len(b); Z = all_bits(n, np.float64); X = b[None, :] - Z @ B.T
    return 0.5 * (X ** 2).sum(1) - X @ c + np.einsum("ij,jk,ik->i", Z, Q, Z)

def mbo_feasible_values_big(Q, c, B, b, chunk=2 ** 17):
    n = len(b); N_ = 2 ** n; out = np.empty(N_); sh = np.arange(n)
    for a in range(0, N_, chunk):
        idx = np.arange(a, min(a + chunk, N_), dtype=np.int64)
        Z = ((idx[:, None] >> sh) & 1).astype(np.float64); X = b[None, :] - Z @ B.T
        out[a:a + len(idx)] = 0.5 * (X ** 2).sum(1) - X @ c + np.einsum("ij,ij->i", Z @ Q, Z)
    return out

def mbo_value(x, z, c, Q): return float(0.5 * x @ x - c @ x + z @ Q @ z)

def _instance(n, seed, gen="dense", p_link=None):
    return generate_instance(n, seed) if gen == "dense" else generate_instance_sparse(n, seed, p_link)

def run_admm(n, seed, solver, K=12, rho=2.0, gen="dense", p_link=None, big=False):
    Q, c, B, b = _instance(n, seed, gen, p_link); stats = exact_stats_big if big else (lambda Qp, h: exact_stats(Qp, h, keep=True))
    x = np.zeros(n); z = np.zeros(n); lam = np.zeros(n); ctx = {}; rows = []; Zs = []
    for k in range(K):
        Qp, h = build_qubo_subproblem(x, lam, Q, B, b, rho); st = stats(Qp, h)
        t0 = time.perf_counter(); out = solver(Qp, h, st, seed * 1000 + k, ctx); wall = time.perf_counter() - t0
        g = gaps(out["best"], st)
        ro = out.get("readout_best", out["best"]); gr = gaps(ro if np.isfinite(ro) else out["best"], st)
        z = np.asarray(out["z"], float); x = x_update(z, lam, c, B, b, rho); res = x + B @ z - b; lam = lam + rho * res
        rows.append({"iter": k, **g, "gap_ro": gr["gap"], "gap_ro_unif": gr["gap_unif"], "hit_ro": gr["hit"],
                     "resid": float(np.linalg.norm(res)), "evals": out.get("evals", 0),
                     "evals_before_topup": out.get("evals_before_topup", out.get("evals", 0)),
                     "nfev": out.get("nfev", 0), "wall": wall, "bound": out.get("bound", np.nan),
                     "Cstar": st["Cstar"], "Cmean": st["Cmean"], "zidx": z_to_index(z)})
        Zs.append(z.copy())
    Fall = mbo_feasible_values_big(Q, c, B, b) if big else mbo_feasible_values(Q, c, B, b)
    Fstar, Fmean = float(Fall.min()), float(Fall.mean())
    Ff = Fall[[z_to_index(zz) for zz in Zs]]
    df = pd.DataFrame(rows); df["F_iter"] = Ff
    tolF = 1e-7 * max(1, abs(Fstar))
    first_opt = int(np.argmax(Ff - Fstar <= tolF)) if np.any(Ff - Fstar <= tolF) else -1
    summ = {"resid_final": rows[-1]["resid"], "F_final": mbo_value(x, z, c, Q), "Fstar": Fstar,
            "Ffeas_best": float(Ff.min()), "Ffeas_last": float(Ff[-1]),
            "mbo_gap_best_unif": float((Ff.min() - Fstar) / max(Fmean - Fstar, 1e-12)),
            "mbo_gap_last_unif": float((Ff[-1] - Fstar) / max(Fmean - Fstar, 1e-12)),
            "mbo_opt_found": float(Ff.min() - Fstar <= tolF), "first_opt_iter": first_opt}
    return df, summ

def exact_trajectory(n, seed, K=12, rho=2.0, gen="dense", p_link=None):
    Q, c, B, b = _instance(n, seed, gen, p_link); x = np.zeros(n); lam = np.zeros(n); out = []
    for _ in range(K):
        Qp, h = build_qubo_subproblem(x, lam, Q, B, b, rho); st = exact_stats(Qp, h, keep=True)
        out.append((Qp, h, st)); z = st["zstar"]; x = x_update(z, lam, c, B, b, rho); lam = lam + rho * (x + B @ z - b)
    return out

def score_on_trajectory(traj, solver, seed):
    ctx = {}; rows = []
    for k, (Qp, h, st) in enumerate(traj):
        out = solver(Qp, h, st, seed * 1000 + k, ctx); g = gaps(out["best"], st)
        ro = out.get("readout_best", out["best"]); gr = gaps(ro if np.isfinite(ro) else out["best"], st)
        rows.append({"iter": k, **g, "gap_ro_unif": gr["gap_unif"], "evals": out.get("evals", 0)})
    return pd.DataFrame(rows)

def mk_exact():        return lambda Qp, h, st, s, ctx: {"best": st["Cstar"], "z": st["zstar"], "evals": 0}
def mk_sa(B, **k):     return lambda Qp, h, st, s, ctx: run_sa(Qp, h, B, s, **k)
def mk_sa_restarts(B, **k): return lambda Qp, h, st, s, ctx: run_sa_restarts(Qp, h, B, s, **k)
def mk_tabu(B):        return lambda Qp, h, st, s, ctx: run_tabu(Qp, h, B, s)
def mk_ls(B):          return lambda Qp, h, st, s, ctx: run_local_search(Qp, h, B, s)
def mk_random(B):      return lambda Qp, h, st, s, ctx: run_random(Qp, h, B, s, st["C"])
def mk_sdp(B):         return lambda Qp, h, st, s, ctx: run_sdp(Qp, h, B, s)
def mk_qaoa(**k):      return lambda Qp, h, st, s, ctx: run_qaoa(Qp, h, st["C"], seed=s, **k)
def mk_fixed(params, reps, B, **k):
    return lambda Qp, h, st, s, ctx: run_qaoa_fixed(Qp, h, params, st["C"], reps, B, s, **k)
def mk_multistart(**k): return lambda Qp, h, st, s, ctx: run_qaoa_multistart(Qp, h, st["C"], seed=s, **k)
def mk_qaoa_warm(**k):
    # angles at iteration k start from the optimized angles of iteration k-1
    def f(Qp, h, st, s, ctx):
        x0 = ctx.get("x")
        out = run_qaoa(Qp, h, st["C"], seed=s, init="given" if x0 is not None else "random", x0=x0, **k)
        ctx["x"] = out["x"]; return out
    return f

def run_arms_embedded(arms, seeds, n, K, tag, gen="dense", p_link=None, rho=2.0, big=False):
    it_rows, sm_rows = [], []; t0 = time.time()
    for j, s in enumerate(seeds):
        for name, fac in arms.items():
            df, summ = run_admm(n, s, fac, K, rho=rho, gen=gen, p_link=p_link, big=big)
            df["seed"], df["arm"] = s, name; it_rows.append(df)
            agg = df[["gap", "gap_unif", "gap_range", "hit", "gap_ro", "gap_ro_unif", "hit_ro"]].mean().to_dict()
            agg.update({"evals": df["evals"].sum(), "evals_before_topup": df["evals_before_topup"].sum(),
                        "wall": df["wall"].sum(), "seed": s, "arm": name, **summ})
            sm_rows.append(agg)
        if (j + 1) % max(1, len(seeds) // 5) == 0:
            print(f"  [{tag}] seed {j + 1}/{len(seeds)}  ({time.time() - t0:.0f}s)")
    return {"iters": pd.concat(it_rows, ignore_index=True), "seeds": pd.DataFrame(sm_rows)}

ARM_RENAME = {"QAOA p=2 raw scale (v5)": "QAOA p=2 unnormalized", "COBYLA raw scale (v5)": "COBYLA unnormalized",
              "QAOA p=2, no top-up (v5)": "QAOA p=2, no top-up", "COBYLA 16x48": "COBYLA 16-shot",
              "SS CVaR alpha* (held-out)": "SS CVaR selected level", "MS + CVaR alpha*": "MS + CVaR selected level",
              "corrected, raw": "corrected, unnormalized", "defective, raw": "defective, unnormalized"}

def cached(name, fn):
    keys_file = f"{name}__keys.json"
    if RESUME and os.path.exists(keys_file):
        keys = json.load(open(keys_file))
        if all(os.path.exists(f"{name}_{k}.csv") for k in keys):
            print(f"[resume] loaded {name}")
            out = {k: pd.read_csv(f"{name}_{k}.csv") for k in keys}
            for v in out.values():
                if "arm" in v: v["arm"] = v["arm"].replace(ARM_RENAME)
            return out
    out = fn()
    for k, v in out.items(): v.to_csv(f"{name}_{k}.csv", index=False)
    json.dump(list(out.keys()), open(keys_file, "w")); return out
print("ADMM machinery ready.")
```

## 10. Transferred angles

Angles are trained noiselessly on held-out instances (seeds from 5000) by minimizing the mean exact expected $g_{\mathrm{unif}}$. ADMM training set: iterations 1, 3 and 5 of the exact trajectories of 12 instances; MaxCut: 12 graphs with edge probability 0.5. At test time the whole budget is spent sampling at the fixed angles.

```python
def _prep(Qp, h, st):
    H, off = qubo_to_ising_paulis(Qp, h)
    return {"d": (st["C"] - off) / ising_scale(H), "C": st["C"], "Cstar": st["Cstar"], "Cmean": st["Cmean"], "n": len(h)}

def training_set(family, n=N):
    out = []
    for s in range(TRAIN_OFFSET, TRAIN_OFFSET + N_TRAIN):
        if family == "admm":
            for (Qp, h, st) in exact_trajectory(n, s, K=6)[::2]:
                out.append(_prep(Qp, h, st))
        else:
            Qp, h, _ = generate_maxcut_instance(n, 0.5, s); out.append(_prep(Qp, h, exact_stats(Qp, h, keep=True)))
    return out

def transfer_objective(params, p, train):
    return float(np.mean([(qaoa_probs_fast(t["d"], params, p, t["n"]) @ t["C"] - t["Cstar"]) / (t["Cmean"] - t["Cstar"])
                          for t in train]))

def train_angles(family, p, n=None):
    train = training_set(family, N if n is None else n); rng = np.random.default_rng(p); best = (np.inf, None)
    starts = [np.concatenate([np.linspace(0.5, 0.5 / p, p), np.linspace(0.5 / p, 0.5, p)])]
    starts += [rng.uniform(-np.pi / 4, np.pi / 4, 2 * p) for _ in range(TRAIN_RESTARTS - 1)]
    for x0 in starts:
        r = minimize(transfer_objective, x0, args=(p, train), method="Nelder-Mead",
                     options={"maxfev": TRAIN_MAXFEV, "xatol": 1e-4, "fatol": 1e-6})
        if r.fun < best[0]: best = (r.fun, r.x)
    return best[1], best[0], len(train)

def _load_or_train():
    if RESUME and os.path.exists("S10a_transferred_angles.json"):
        print("[resume] loaded transferred angles")
        return {(k.split("|")[0], int(k.split("|")[1])): np.array(v) for k, v in json.load(open("S10a_transferred_angles.json")).items()}
    ang, rows = {}, []
    for fam in ("admm", "maxcut"):
        for p in DEPTHS:
            t0 = time.time(); x, f, m = train_angles(fam, p); ang[(fam, p)] = x
            rows.append({"family": fam, "p": p, "train_instances": m, "train_Egap_unif": f, "seconds": time.time() - t0})
            print(f"  {fam:6s} p={p}: training mean expected g_unif = {f:.3f}")
    json.dump({f"{k[0]}|{k[1]}": v.tolist() for k, v in ang.items()}, open("S10a_transferred_angles.json", "w"))
    pd.DataFrame(rows).to_csv("S10a_transfer_training.csv", index=False)
    return ang

TRANSFER = _load_or_train()
```

## 11. Main comparison, ADMM-embedded

$n=12$, $K=12$, 768 evaluations per subproblem. Every QAOA arm is scored by best sample (primary) and final read-out (secondary).

```python
VIAB = []
def viab(setting, group, comparator, t):
    VIAB.append({"setting": setting, "group": group, "comparator": comparator, "r": t["r"], "r_lo": t["r_lo"],
                 "r_hi": t["r_hi"], "P": t["P"], "n": t["n"], "n_nonzero": t["n_nonzero"],
                 "mean_diff": t["mean_diff"], "md_lo": t["md_lo"], "md_hi": t["md_hi"]})

def col(d, arm, c, key="seed"):
    return d[d.arm == arm].sort_values(key)[c].values

def arm_table(d, order, cols):
    rows = []
    for a in order:
        s = d[d.arm == a]; row = {"arm": a}
        for c_ in cols:
            m, lo, hi = bootstrap_ci(s[c_]); row[c_], row[c_ + "_lo"], row[c_ + "_hi"] = m, lo, hi
        rows.append(row)
    return pd.DataFrame(rows)

def print_table(t, cols, fmt="{:.3f}"):
    for _, r in t.iterrows():
        print(f"  {r['arm']:<32s}" + "".join(f" {c_}={fmt.format(r[c_])} [{fmt.format(r[c_ + '_lo'])},{fmt.format(r[c_ + '_hi'])}]" for c_ in cols))

arms_A = {"Exact": mk_exact(), "SA": mk_sa(BUDGET), "Random": mk_random(BUDGET), "SDP": mk_sdp(BUDGET)}
for p in DEPTHS:
    arms_A[f"QAOA p={p}"] = mk_qaoa(reps=p, shots=SHOTS, maxiter=MAXITER)
arms_A["QAOA p=2 unnormalized"] = mk_qaoa(reps=2, shots=SHOTS, maxiter=MAXITER, normalize=False)
arms_A["QAOA p=2 warm-start"] = mk_qaoa_warm(reps=2, shots=SHOTS, maxiter=MAXITER)
arms_A["QAOA p=2 transferred"] = mk_fixed(TRANSFER[("admm", 2)], 2, BUDGET)
arms_A["QAOA p=3 transferred"] = mk_fixed(TRANSFER[("admm", 3)], 3, BUDGET)
arms_A["QAOA multi-start"] = mk_multistart(reps=2, shots=MS_SHOTS, budget=BUDGET, mode="pooled", readout_shots=MS_SHOTS)
arms_A["QAOA budget control"] = mk_qaoa(reps=2, shots=MS_SHOTS, maxiter=12, readout_shots=MS_SHOTS)
ORDER_A = list(arms_A); QAOA_A = [a for a in ORDER_A if a.startswith("QAOA")]
A = cached("S10_main", lambda: run_arms_embedded(arms_A, list(range(N_MAIN)), N, K_MAIN, "S10"))
dA = A["seeds"]
```

```python
tabA = arm_table(dA, ORDER_A, ["gap", "gap_unif", "hit", "gap_ro_unif", "resid_final", "mbo_gap_best_unif", "mbo_opt_found", "evals"])
tabA.to_csv("S10_main_table.csv", index=False)
print("Per-arm means [95% bootstrap CI]")
print_table(tabA, ["gap_unif", "hit", "gap_ro_unif"])
print("\nADMM-level outcome")
print_table(tabA, ["resid_final", "mbo_gap_best_unif", "mbo_opt_found"], "{:.3g}")
print("\nEvaluations per instance:")
for a in ORDER_A:
    s = dA[dA.arm == a]; print(f"  {a:<32s} total={s.evals.mean():8.1f}  before top-up={s.evals_before_topup.mean():8.1f}")

print("\n[S10 vs SA]")
tA_sa = show_tests([paired(col(dA, a, "gap_unif"), col(dA, "SA", "gap_unif"), f"S10 {a} vs SA", "S10 vs SA")
                    for a in ORDER_A if a not in ("Exact", "SA")])
print("\n[S10 QAOA vs Random, best-sample]")
tA_rnd = show_tests([paired(col(dA, a, "gap_unif"), col(dA, "Random", "gap_unif"), f"S10 {a} vs Random", "S10 QAOA vs Random") for a in QAOA_A])
for _, t in tA_rnd.iterrows():
    viab("ADMM embedded, n=12: " + t["label"].split("S10 ")[1].split(" vs")[0], "ADMM embedded", "Random", t)
print("\n[S10 QAOA vs Random, read-out]")
show_tests([paired(col(dA, a, "gap_ro_unif"), col(dA, "Random", "gap_unif"), f"S10 {a} (read-out) vs Random",
                   "S10 read-out vs Random", "exploratory") for a in ["QAOA p=2", "QAOA p=2 unnormalized", "QAOA multi-start"]])
best_q = min(QAOA_A, key=lambda a: dA[dA.arm == a].gap_unif.mean())
for comp in ("SA", "SDP"):
    viab(f"ADMM embedded, n=12: best QAOA protocol ({best_q})", "ADMM embedded", comp,
         paired(col(dA, best_q, "gap_unif"), col(dA, comp, "gap_unif"), record=False))
try:
    fs, fp = friedmanchisquare(*[col(dA, f"QAOA p={p}", "gap_unif") for p in DEPTHS])
    register("S10 Friedman depth p=1,2,3", "S10 depth", "confirmatory", N_MAIN, N_MAIN, float(fp), np.nan)
    print(f"\nFriedman test across depths: chi2={fs:.3f}, P={fmtP(fp)}")
except ValueError as e:
    print("Friedman skipped:", e)
print("\n[S10 ADMM level: best feasible gap vs SA]")
show_tests([paired(col(dA, a, "mbo_gap_best_unif"), col(dA, "SA", "mbo_gap_best_unif"), f"S10 ADMM-level {a} vs SA",
                   "S10 ADMM-level vs SA", "exploratory") for a in ["QAOA p=2", best_q, "Random"]])
sdp_iter = A["iters"][A["iters"].arm == "SDP"]
ig = ((sdp_iter.Cstar - sdp_iter.bound) / (sdp_iter.Cmean - sdp_iter.Cstar)).clip(lower=0)
print(f"\nSDP integrality gap (normalized): median {ig.median():.3f}; tight (<1e-3) on {(ig < 1e-3).mean():.2f}; "
      f"rounding exact on {sdp_iter.hit.mean():.2%} of subproblems")
```

## 13. Shared-trajectory comparison

Every solver is scored on the identical subproblems of the exact-solver trajectory, without feedback.

```python
arms_T = {k: arms_A[k] for k in ["SA", "Random", "SDP", "QAOA p=2", "QAOA p=2 unnormalized", "QAOA p=2 warm-start",
                                  "QAOA p=2 transferred", "QAOA multi-start"]}
def _run_traj():
    rows = []; t0 = time.time()
    for s in range(N_TRAJ):
        traj = exact_trajectory(N, s, K_MAIN)
        for a, f in arms_T.items():
            df = score_on_trajectory(traj, f, s)
            rows.append({"seed": s, "arm": a, **df[["gap", "gap_unif", "hit", "gap_ro_unif"]].mean().to_dict()})
        if (s + 1) % max(1, N_TRAJ // 4) == 0: print(f"  [S11] seed {s + 1}/{N_TRAJ} ({time.time() - t0:.0f}s)")
    return {"seeds": pd.DataFrame(rows)}
T = cached("S11_shared_trajectory", _run_traj); dT = T["seeds"]
print_table(arm_table(dT, list(arms_T), ["gap_unif", "hit"]), ["gap_unif", "hit"])
print("\n[S11 QAOA vs Random]")
tT = show_tests([paired(col(dT, a, "gap_unif"), col(dT, "Random", "gap_unif"), f"S11 {a} vs Random", "S11 vs Random")
                 for a in arms_T if a.startswith("QAOA")])
for _, t in tT.iterrows():
    viab("ADMM shared trajectory: " + t["label"].split("S11 ")[1].split(" vs")[0], "ADMM shared trajectory", "Random", t)
```

## 14. Mechanism: restart diversity versus read-out size

Standalone first subproblem, 768 evaluations. Equal read-out sizes isolate restart diversity; read-out 64 versus 32 isolates read-out size; pooled versus best-restart read-out isolates pooling. Decomposition tests use read-out scoring.

```python
MECH = {
    "SS, read-out 32": lambda Qp, h, st, s: run_qaoa(Qp, h, st["C"], reps=2, shots=32, budget=BUDGET, readout_shots=32, seed=s),
    "SS, read-out 64": lambda Qp, h, st, s: run_qaoa(Qp, h, st["C"], reps=2, shots=32, budget=BUDGET, readout_shots=64, seed=s),
    "MS pooled, 4 x 16": lambda Qp, h, st, s: run_qaoa_multistart(Qp, h, st["C"], reps=2, shots=16, budget=BUDGET, mode="pooled", readout_shots=16, seed=s),
    "MS best restart, read-out 32": lambda Qp, h, st, s: run_qaoa_multistart(Qp, h, st["C"], reps=2, shots=16, budget=BUDGET, mode="best", readout_shots=32, seed=s),
    "MS best restart, read-out 64": lambda Qp, h, st, s: run_qaoa_multistart(Qp, h, st["C"], reps=2, shots=16, budget=BUDGET, mode="best", readout_shots=64, seed=s),
    "Budget control, 192": lambda Qp, h, st, s: run_qaoa(Qp, h, st["C"], reps=2, shots=16, maxiter=12, readout_shots=16, seed=s),
    "Random": lambda Qp, h, st, s: run_random(Qp, h, BUDGET, s, st["C"]),
    "SA": lambda Qp, h, st, s: run_sa(Qp, h, BUDGET, s),
}
def run_standalone(arms, seeds, make, tag):
    rows = []; t0 = time.time()
    for j, s in enumerate(seeds):
        Qp, h, extra = make(s); st = exact_stats(Qp, h, keep=True)
        for a, f in arms.items():
            t1 = time.perf_counter(); out = f(Qp, h, st, s) if extra is None else f(Qp, h, st, s, extra); wall = time.perf_counter() - t1
            g = gaps(out["best"], st); ro = out.get("readout_best", out["best"])
            gr = gaps(ro if np.isfinite(ro) else out["best"], st)
            row = {"seed": s, "arm": a, **g, "gap_ro": gr["gap"], "gap_ro_unif": gr["gap_unif"], "hit_ro": gr["hit"],
                   "evals": out.get("evals", np.nan), "evals_before_topup": out.get("evals_before_topup", np.nan),
                   "nfev": out.get("nfev", np.nan), "Cstar": st["Cstar"], "density": qubo_density(Qp), "wall": wall}
            if "bound" in out: row["bound"] = out["bound"]; row["Cmean"] = st["Cmean"]
            if st["Cstar"] < 0: row["approx"] = out["best"] / st["Cstar"]; row["approx_ro"] = ro / st["Cstar"]
            rows.append(row)
        if (j + 1) % max(1, len(seeds) // 4) == 0: print(f"  [{tag}] seed {j + 1}/{len(seeds)} ({time.time() - t0:.0f}s)")
    return pd.DataFrame(rows)
admm_first = lambda s: (*first_zupdate(N, s), None)
M = cached("S12_mechanism", lambda: {"seeds": run_standalone(MECH, range(N_MECH), admm_first, "S12")}); dM = M["seeds"]
print_table(arm_table(dM, list(MECH), ["gap_unif", "gap_ro_unif"]), ["gap_unif", "gap_ro_unif"])
print("\n[S12 decomposition]")
show_tests([
    paired(col(dM, "MS best restart, read-out 32", "gap_ro_unif"), col(dM, "SS, read-out 32", "gap_ro_unif"), "S12 restart diversity at read-out 32 (MS best vs SS)", "S12 mechanism"),
    paired(col(dM, "MS best restart, read-out 64", "gap_ro_unif"), col(dM, "SS, read-out 64", "gap_ro_unif"), "S12 restart diversity at read-out 64 (MS best vs SS)", "S12 mechanism"),
    paired(col(dM, "SS, read-out 64", "gap_ro_unif"), col(dM, "SS, read-out 32", "gap_ro_unif"), "S12 read-out size (SS 64 vs SS 32)", "S12 mechanism"),
    paired(col(dM, "MS pooled, 4 x 16", "gap_ro_unif"), col(dM, "MS best restart, read-out 64", "gap_ro_unif"), "S12 pooling (MS pooled 64 vs MS best 64)", "S12 mechanism"),
    paired(col(dM, "MS pooled, 4 x 16", "gap_unif"), col(dM, "SS, read-out 32", "gap_unif"), "S12 multi-start vs single-start, best-sample", "S12 mechanism"),
    paired(col(dM, "Budget control, 192", "gap_unif"), col(dM, "SS, read-out 32", "gap_unif"), "S12 budget control vs full budget, best-sample", "S12 mechanism"),
])
```

## 15. Classical-optimizer ablation

Standalone first subproblem, $p=2$, 768 evaluations.

```python
ABL = {
    "COBYLA": lambda Qp, h, st, s: run_qaoa(Qp, h, st["C"], reps=2, shots=SHOTS, maxiter=MAXITER, seed=s),
    "Nelder-Mead": lambda Qp, h, st, s: run_qaoa(Qp, h, st["C"], reps=2, shots=SHOTS, maxiter=MAXITER, seed=s, optimizer="Nelder-Mead"),
    "SPSA": lambda Qp, h, st, s: run_qaoa(Qp, h, st["C"], reps=2, shots=SHOTS, maxiter=MAXITER, seed=s, optimizer="SPSA"),
    "COBYLA 16-shot": lambda Qp, h, st, s: run_qaoa(Qp, h, st["C"], reps=2, shots=16, maxiter=48, seed=s),
    "COBYLA unnormalized": lambda Qp, h, st, s: run_qaoa(Qp, h, st["C"], reps=2, shots=SHOTS, maxiter=MAXITER, seed=s, normalize=False),
    "Transferred p=2": lambda Qp, h, st, s: run_qaoa_fixed(Qp, h, TRANSFER[("admm", 2)], st["C"], 2, BUDGET, s),
    "SA": lambda Qp, h, st, s: run_sa(Qp, h, BUDGET, s),
    "Random": lambda Qp, h, st, s: run_random(Qp, h, BUDGET, s, st["C"]),
}
Bb = cached("S13_ablation", lambda: {"seeds": run_standalone(ABL, range(N_ABL), admm_first, "S13")}); dB = Bb["seeds"]
print_table(arm_table(dB, list(ABL), ["gap_unif", "gap_ro_unif", "nfev", "evals_before_topup"]), ["gap_unif", "gap_ro_unif", "nfev"], "{:.2f}")
print("\n[S13 vs Random]")
tB = show_tests([paired(col(dB, a, "gap_unif"), col(dB, "Random", "gap_unif"), f"S13 {a} vs Random", "S13 vs Random")
                 for a in list(ABL)[:6]])
for _, t in tB.iterrows():
    viab("ADMM standalone, n=12: " + t["label"].split("S13 ")[1].split(" vs")[0], "ADMM standalone", "Random", t)
print("\n[S13 best QAOA protocol vs SA]")
bq = min(list(ABL)[:6], key=lambda a: dB[dB.arm == a].gap_unif.mean())
show_tests([paired(col(dB, bq, "gap_unif"), col(dB, "SA", "gap_unif"), f"S13 {bq} vs SA", "S13 vs SA")])
```

## 16. Size scaling at a fixed budget fraction, and SDP extension

ADMM-embedded, $K=8$, budget 18.75% of the search space (25% at $n=8$, the whole space at $n=6$; $n=6$ is excluded from trend tests). The fixed absolute budget is studied in Section 26.

```python
def _run_size():
    its, sms = [], []
    for n_ in SIZES:
        budget = max(int(round(2 ** n_ * 0.1875)), 16); mi = max(budget // 16, 4); budget = 16 * mi
        arms = {"QAOA p=2": mk_qaoa(reps=2, shots=16, maxiter=mi), "QAOA p=2 transferred": mk_fixed(TRANSFER[("admm", 2)], 2, budget),
                "SA": mk_sa(budget), "Random": mk_random(budget), "SDP": mk_sdp(budget)}
        out = run_arms_embedded(arms, list(range(N_SCAL)), n_, K_SUB, f"S14 n={n_}")
        for d in (out["iters"], out["seeds"]): d["n"], d["frac"] = n_, budget / 2 ** n_
        its.append(out["iters"]); sms.append(out["seeds"])
    return {"iters": pd.concat(its, ignore_index=True), "seeds": pd.concat(sms, ignore_index=True)}
D = cached("S14_size", _run_size); dD = D["seeds"]
print(dD.groupby(["n", "arm"])["gap_unif"].mean().unstack().round(3))
print("\n[S14 size trends, n >= 8]")
for a in ["QAOA p=2", "QAOA p=2 transferred", "SA", "Random", "SDP"]:
    s = dD[(dD.arm == a) & dD.n.isin(SIZES_TREND)]
    r_, P_, lo, hi = spearman_ci(s.n, s.gap_unif)
    register(f"S14 size trend {a}", "S14 size trends", "exploratory", len(s), len(s), P_, r_)
    print(f"  {a:<22s} r_s={r_:+.3f} [{lo:+.2f},{hi:+.2f}] P={fmtP(P_)}")
for n_ in SIZES_TREND:
    s = dD[dD.n == n_]
    viab(f"ADMM embedded, n={n_}", "ADMM size", "Random",
         paired(col(s, "QAOA p=2", "gap_unif"), col(s, "Random", "gap_unif"), f"S14 n={n_} QAOA vs Random", "S14 per-size", "exploratory"))
```

```python
def _run_sdpx():
    rows = []
    for n_ in SIZES_SDPX:
        for s in range(N_SDPX):
            Qp, h = first_zupdate(n_, s); st = exact_stats(Qp, h)
            r = run_sdp(Qp, h, min(int(round(2 ** n_ * 0.1875)), 1024), s); g = gaps(r["best"], st)
            ig_ = max(st["Cstar"] - r["bound"], 0) / (st["Cmean"] - st["Cstar"])
            rows.append({"n": n_, "seed": s, **g, "integrality_gap_unif": ig_, "sdp_time": r["sdp_time"]})
        print(f"  [S14b] n={n_} done")
    return {"seeds": pd.DataFrame(rows)}
DX = cached("S14b_sdp_extended", _run_sdpx); dX = DX["seeds"]
print(dX.groupby("n").agg(gap_unif=("gap_unif", "mean"), exact=("hit", "mean"), integrality=("integrality_gap_unif", "median"),
                          sdp_seconds=("sdp_time", "mean")).round(4))
r_, P_, lo, hi = spearman_ci(dX.n, dX.gap_unif)
register("S14b SDP extended size trend", "S14 size trends", "exploratory", len(dX), len(dX), P_, r_)
print(f"SDP gap trend: r_s={r_:+.3f} [{lo:+.2f},{hi:+.2f}] P={fmtP(P_)}")
```

## 17. Depolarizing noise

ADMM-embedded, $K=5$, 12 instances; SA and random search are noise-free references on the same instances.

```python
def build_noise_model(p1):
    nm = NoiseModel()
    nm.add_all_qubit_quantum_error(depolarizing_error(p1, 1), ["rx", "ry", "rz", "h", "x", "sx"])
    nm.add_all_qubit_quantum_error(depolarizing_error(min(2 * p1, 1.0), 2), ["cx", "cz", "rzz"])
    return nm
def _run_noise():
    sms = []
    ref = run_arms_embedded({"SA": mk_sa(BUDGET), "Random": mk_random(BUDGET)}, list(range(N_NOISE)), N, K_NOISE, "S15 refs")["seeds"]
    ref["noise"] = np.nan; sms.append(ref)
    for p1 in NOISE_LEVELS:
        nm = None if p1 == 0 else build_noise_model(p1)
        arms = {"QAOA p=2": mk_qaoa(reps=2, shots=SHOTS, maxiter=MAXITER, noise_model=nm),
                "QAOA p=2 transferred": mk_fixed(TRANSFER[("admm", 2)], 2, BUDGET, noise_model=nm)}
        out = run_arms_embedded(arms, list(range(N_NOISE)), N, K_NOISE, f"S15 noise={p1}")["seeds"]; out["noise"] = p1; sms.append(out)
    return {"seeds": pd.concat(sms, ignore_index=True)}
E = cached("S15_noise", _run_noise); dE = E["seeds"]
print(dE[dE.noise.notna()].groupby(["noise", "arm"])["gap_unif"].mean().unstack().round(3))
print("Noise-free references:", dE[dE.noise.isna()].groupby("arm")["gap_unif"].mean().round(3).to_dict())
for a in ["QAOA p=2", "QAOA p=2 transferred"]:
    # the same instances are run at every noise level: one Spearman correlation per instance, then an exact signed-rank test across instances
    s = dE[(dE.arm == a) & dE.noise.notna()]
    rs = np.array([spearmanr(g.noise, g.gap_unif)[0] for _, g in s.groupby("seed")]); rs = np.nan_to_num(rs)
    P_ = wilcoxon_exact(rs)
    register(f"S15 noise trend {a}", "S15 noise", "exploratory", len(rs), int(np.sum(np.abs(rs) > 1e-12)), P_, float(np.median(rs)))
    print(f"  noise trend {a:<22s} per-instance r_s: median {np.median(rs):+.3f}, positive on {int((rs > 0).sum())}/{len(rs)} instances, P={fmtP(P_)}")
rnd_ref = dE[dE.arm == "Random"].sort_values("seed").gap_unif.values
for p1 in NOISE_LEVELS:
    s = dE[(dE.noise == p1) & (dE.arm == "QAOA p=2")].sort_values("seed")
    viab(f"ADMM embedded, noise {p1:g}", "ADMM noise", "Random",
         paired(s.gap_unif.values, rnd_ref, f"S15 noise={p1:g} QAOA vs Random", "S15 per-level", "exploratory"))
```

## 18. Budget sweep with evaluation accounting

ADMM-embedded, $K=8$. The no-top-up arm leaves budget unspent after early optimizer termination.

```python
def _run_budget():
    sms = []
    for fr in BUDGET_FRACS:
        mi = max(int(round(SPACE * fr)) // 32, 6); budget = 32 * mi
        arms = {"QAOA p=2": mk_qaoa(reps=2, shots=32, maxiter=mi),
                "QAOA p=2, no top-up": mk_qaoa(reps=2, shots=32, maxiter=mi, spend_remaining=False),
                "QAOA p=2 transferred": mk_fixed(TRANSFER[("admm", 2)], 2, budget),
                "SA": mk_sa(budget), "Random": mk_random(budget)}
        out = run_arms_embedded(arms, list(range(N_SWEEP)), N, K_SUB, f"S16 {fr:.0%}")["seeds"]
        out["frac"], out["budget_per_iter"] = fr, budget; sms.append(out)
    return {"seeds": pd.concat(sms, ignore_index=True)}
F = cached("S16_budget", _run_budget); dF = F["seeds"]
acc = dF.groupby(["frac", "arm"]).agg(gap_unif=("gap_unif", "mean"), gap_ro_unif=("gap_ro_unif", "mean"),
                                      evals=("evals", "mean"), before_topup=("evals_before_topup", "mean")).round(3)
print(acc); acc.to_csv("S16_budget_accounting.csv")
for fr in BUDGET_FRACS:
    s = dF[dF.frac == fr]
    t1 = paired(col(s, "QAOA p=2", "gap_unif"), col(s, "Random", "gap_unif"), f"S16 {fr:.2%} QAOA vs Random (best-sample)", "S16 per-fraction", "exploratory")
    t2 = paired(col(s, "QAOA p=2, no top-up", "gap_ro_unif"), col(s, "Random", "gap_unif"), f"S16 {fr:.2%} QAOA read-out, no top-up vs Random", "S16 per-fraction, read-out", "exploratory")
    viab(f"ADMM embedded, budget {fr:.1%}", "ADMM budget", "Random", t1)
    print(f"  {fr:6.2%}: best-sample r={t1['r']:+.3f} P={fmtP(t1['P'])} | read-out r={t2['r']:+.3f} P={fmtP(t2['P'])}")
```

## 19. MaxCut with Goemans-Williamson

Standalone. Two disjoint seed ranges at edge probability 0.5, then sweeps over edge probability and size.

```python
def mc_arms(budget, shots=SHOTS, maxiter=MAXITER, full=True):
    arms = {"SA": lambda Qp, h, st, s, G: run_sa(Qp, h, budget, s),
            "Random": lambda Qp, h, st, s, G: run_random(Qp, h, budget, s, st["C"]),
            "GW": lambda Qp, h, st, s, G: run_gw(G, len(h), budget, s),
            "QAOA p=2": lambda Qp, h, st, s, G: run_qaoa(Qp, h, st["C"], reps=2, shots=shots, maxiter=maxiter, seed=s),
            "QAOA p=2 transferred": lambda Qp, h, st, s, G: run_qaoa_fixed(Qp, h, TRANSFER[("maxcut", 2)], st["C"], 2, budget, s)}
    if full:
        for p in (1, 3):
            arms[f"QAOA p={p}"] = (lambda pp: lambda Qp, h, st, s, G: run_qaoa(Qp, h, st["C"], reps=pp, shots=shots, maxiter=maxiter, seed=s))(p)
        arms["QAOA multi-start"] = lambda Qp, h, st, s, G: run_qaoa_multistart(Qp, h, st["C"], reps=2, shots=MS_SHOTS, budget=budget, mode="pooled", readout_shots=MS_SHOTS, seed=s)
    return arms
MC_ORDER = ["SA", "Random", "GW", "QAOA p=1", "QAOA p=2", "QAOA p=3", "QAOA multi-start", "QAOA p=2 transferred"]
mc_make = lambda pe, n_=N: (lambda s: generate_maxcut_instance(n_, pe, s))
G_ = cached("S17_maxcut_main", lambda: {
    "orig": run_standalone(mc_arms(BUDGET), range(N_MC), mc_make(0.5), "S17 original"),
    "repl": run_standalone(mc_arms(BUDGET), range(REPL_OFFSET, REPL_OFFSET + N_MC), mc_make(0.5), "S17 replication")})
for key, d in G_.items():
    nm_ = {"orig": "original", "repl": "replication"}[key]
    print(f"\n[S17 {nm_}] approximation ratio and g_unif")
    print_table(arm_table(d, MC_ORDER, ["approx", "gap_unif", "hit"]), ["approx", "gap_unif", "hit"])
    show_tests([paired(col(d, a, "gap_unif"), col(d, "SA", "gap_unif"), f"S17 {nm_} {a} vs SA", f"S17 {nm_} vs SA")
                for a in MC_ORDER if a != "SA"])
    tq = show_tests([paired(col(d, a, "gap_unif"), col(d, "Random", "gap_unif"), f"S17 {nm_} {a} vs Random", f"S17 {nm_} vs Random")
                     for a in MC_ORDER if a.startswith("QAOA")])
    if key == "orig":
        for _, t in tq.iterrows():
            viab("MaxCut n=12: " + t["label"].split("original ")[1].split(" vs")[0], "MaxCut", "Random", t)
        bq = min([a for a in MC_ORDER if a.startswith("QAOA")], key=lambda a: d[d.arm == a].gap_unif.mean())
        viab(f"MaxCut n=12: best QAOA ({bq})", "MaxCut", "GW", paired(col(d, bq, "gap_unif"), col(d, "GW", "gap_unif"), record=False))
```

```python
def _run_mc_sweeps():
    dens = []
    for pe in DENSITY_SWEEP:
        d = run_standalone(mc_arms(BUDGET, full=False), range(N_DENS), mc_make(pe), f"S17 density {pe}"); d["p_edge"] = pe; dens.append(d)
    size = []
    for n_ in SIZES:
        budget = max(int(round(2 ** n_ * 0.1875)), 16); mi = max(budget // 8, 6); budget = 8 * mi
        d = run_standalone(mc_arms(budget, shots=8, maxiter=mi, full=False), range(N_SCAL), mc_make(0.5, n_), f"S17 size {n_}")
        d["n"] = n_; size.append(d)
    return {"density": pd.concat(dens, ignore_index=True), "size": pd.concat(size, ignore_index=True)}
GS = cached("S17_maxcut_sweeps", _run_mc_sweeps); dGd, dGs = GS["density"], GS["size"]
print("Mean approximation ratio by edge probability:"); print(dGd.groupby(["p_edge", "arm"])["approx"].mean().unstack().round(4))
for pe in DENSITY_SWEEP:
    s = dGd[dGd.p_edge == pe]
    t = paired(col(s, "QAOA p=2", "gap_unif"), col(s, "Random", "gap_unif"), f"S17 MaxCut edge probability {pe} QAOA vs Random", "S17 MaxCut per-density", "exploratory")
    viab(f"MaxCut n=12, edge probability {pe}", "MaxCut density", "Random", t)
    print(f"  edge probability {pe}: mean diff={t['mean_diff']:+.3f} [{t['md_lo']:+.3f},{t['md_hi']:+.3f}]  r={t['r']:+.3f}  P={fmtP(t['P'])}")
print("\nMean approximation ratio by size:"); print(dGs.groupby(["n", "arm"])["approx"].mean().unstack().round(4))
```

## 20. Coupling density

Standalone first subproblem at $n=12$; six sparse-coupling levels plus the dense generator.

```python
DENS_ARMS = {"QAOA p=2": ABL["COBYLA"], "QAOA p=2 transferred": ABL["Transferred p=2"],
             "Random": ABL["Random"], "SA": ABL["SA"]}
def _run_density():
    out = []
    for pl in PLINK_SWEEP + ["dense"]:
        mk = (lambda s: (*first_zupdate(N, s), None)) if pl == "dense" else (lambda s, pl=pl: (*first_zupdate(N, s, gen="sparse", p_link=pl), None))
        d = run_standalone(DENS_ARMS, range(N_DENS), mk, f"S18 {pl}"); d["level"] = str(pl); out.append(d)
    return {"seeds": pd.concat(out, ignore_index=True)}
I_ = cached("S18_density", _run_density); dI = I_["seeds"]
lev_rows = []
for lv, s in dI.groupby("level", sort=False):
    t = paired(col(s, "QAOA p=2", "gap_unif"), col(s, "Random", "gap_unif"), f"S18 ADMM level {lv} QAOA vs Random", "S18 ADMM per-density", "exploratory")
    viab(f"ADMM standalone, coupling density {s.density.mean():.2f}", "ADMM density", "Random", t)
    lev_rows.append({"family": "ADMM", "level": lv, "density": s.density.mean(), "delta": t["mean_diff"], "lo": t["md_lo"], "hi": t["md_hi"], "P": t["P"], "r": t["r"]})
for pe, s in dGd.groupby("p_edge"):
    t = paired(col(s, "QAOA p=2", "gap_unif"), col(s, "Random", "gap_unif"), record=False)
    lev_rows.append({"family": "MaxCut", "level": str(pe), "density": s.density.mean(), "delta": t["mean_diff"], "lo": t["md_lo"], "hi": t["md_hi"], "P": t["P"], "r": t["r"]})
dens_tab = pd.DataFrame(lev_rows); dens_tab.to_csv("S18_density_levels.csv", index=False)
print(dens_tab.round(3).to_string(index=False))
r_, P_, lo, hi = spearman_ci(dens_tab.density, dens_tab.delta)
register("S18 pooled density association", "S18 density", "exploratory", len(dens_tab), len(dens_tab), P_, r_)
print(f"\nPooled association: r_s={r_:+.3f} [{lo:+.2f},{hi:+.2f}] P={fmtP(P_)}")
```

## 22. Hamiltonian defect

QAOA's own gap with the defective versus the corrected converter, on identical instances, for both Hamiltonian scalings and both scoring rules, with coefficient-scale and rank-correlation diagnostics.

```python
DEF = {}
for prot, norm in (("unnormalized", False), ("normalized", True)):
    for conv, fn in (("corrected", qubo_to_ising_paulis), ("defective", qubo_to_ising_paulis_DEFECTIVE)):
        DEF[f"{conv}, {prot}"] = (lambda fn=fn, norm=norm: lambda Qp, h, st, s: run_qaoa(Qp, h, st["C"], reps=2, shots=SHOTS, maxiter=MAXITER, seed=s, converter=fn, normalize=norm))()
DEF["Random"] = ABL["Random"]
J_ = cached("S20_defect", lambda: {"seeds": run_standalone(DEF, range(N_BUG), admm_first, "S20")}); dJ = J_["seeds"]
print_table(arm_table(dJ, list(DEF), ["gap_unif", "gap_ro_unif"]), ["gap_unif", "gap_ro_unif"])
print("\n[S20 defective vs corrected (r>0: defective worse)]")
rows = []
for prot in ("unnormalized", "normalized"):
    for sc, lab in (("gap_unif", "best-sample"), ("gap_ro_unif", "read-out")):
        rows.append(paired(col(dJ, f"defective, {prot}", sc), col(dJ, f"corrected, {prot}", sc), f"S20 defective vs corrected, {prot}, {lab}", "S20 defect"))
show_tests(rows)
diag = []
for s in range(N_BUG):
    Qp, h = first_zupdate(N, s); Cv = cost_vector(Qp, h)
    Hc, _ = qubo_to_ising_paulis(Qp, h); Hd, offd = qubo_to_ising_paulis_DEFECTIVE(Qp, h)
    Ed = np.real(Hd.to_matrix(sparse=True).diagonal()) + offd
    diag.append({"seed": s, "scale_ratio": ising_scale(Hd) / ising_scale(Hc), "rank_corr": spearmanr(Ed, Cv)[0],
                 "same_ground_state": int(np.argmin(Ed) == np.argmin(Cv))})
diag = pd.DataFrame(diag); diag.to_csv("S20_defect_diagnostics.csv", index=False)
print(f"\nCoefficient scale, defective/corrected: median {diag.scale_ratio.median():.3f}")
print(f"Rank correlation with the true cost: median {diag.rank_corr.median():.3f}")
print(f"Same ground state on {diag.same_ground_state.mean():.0%} of instances")
```

## 23. Landscape mitigation

CVaR level selected on held-out seeds (from 2000); $\alpha=0.1$ is pre-specified. Multi-start with CVaR is compared with a matched multi-start run on the same subproblems. MaxCut warm start follows.

```python
sel_arms = {f"CVaR {a}": (lambda a=a: lambda Qp, h, st, s: run_qaoa(Qp, h, st["C"], reps=2, shots=SHOTS, maxiter=MAXITER, seed=s, objective="cvar", cvar_alpha=a))() for a in ALPHAS}
SEL = cached("S21_alpha_selection", lambda: {"seeds": run_standalone(sel_arms, range(SEL_OFFSET, SEL_OFFSET + N_SEL), admm_first, "S21 select")})
sel_means = SEL["seeds"].groupby("arm")["gap_unif"].mean()
ALPHA_STAR = float(sel_means.idxmin().split()[-1]); print("Held-out means:", sel_means.round(3).to_dict(), "-> selected alpha =", ALPHA_STAR)

cv = lambda a: dict(objective="cvar", cvar_alpha=a)
LAND = {
    "SS (mean)": ABL["COBYLA"],
    "SS CVaR 0.1 (pre-specified)": lambda Qp, h, st, s: run_qaoa(Qp, h, st["C"], reps=2, shots=SHOTS, maxiter=MAXITER, seed=s, **cv(0.1)),
    "SS CVaR selected level": lambda Qp, h, st, s: run_qaoa(Qp, h, st["C"], reps=2, shots=SHOTS, maxiter=MAXITER, seed=s, **cv(ALPHA_STAR)),
    "SS linear ramp": lambda Qp, h, st, s: run_qaoa(Qp, h, st["C"], reps=2, shots=SHOTS, maxiter=MAXITER, seed=s, init="linear_ramp"),
    "SS ramp + CVaR 0.1": lambda Qp, h, st, s: run_qaoa(Qp, h, st["C"], reps=2, shots=SHOTS, maxiter=MAXITER, seed=s, init="linear_ramp", **cv(0.1)),
    "MS (mean)": MECH["MS pooled, 4 x 16"],
    "MS + CVaR selected level": lambda Qp, h, st, s: run_qaoa_multistart(Qp, h, st["C"], reps=2, shots=16, budget=BUDGET, mode="pooled", readout_shots=16, seed=s, **cv(ALPHA_STAR)),
    "Transferred p=2": ABL["Transferred p=2"], "Random": ABL["Random"], "SA": ABL["SA"],
}
L_ = cached("S21_landscape", lambda: {"seeds": run_standalone(LAND, range(N_LAND), admm_first, "S21")}); dL = L_["seeds"]
print_table(arm_table(dL, list(LAND), ["gap_unif", "gap_ro_unif"]), ["gap_unif", "gap_ro_unif"])
print("\n[S21 confirmatory]")
show_tests([paired(col(dL, "SS CVaR 0.1 (pre-specified)", "gap_unif"), col(dL, "SS (mean)", "gap_unif"), "S21 CVaR 0.1 vs SS", "S21 confirmatory"),
            paired(col(dL, "MS + CVaR selected level", "gap_unif"), col(dL, "MS (mean)", "gap_unif"), "S21 MS + CVaR vs MS (matched)", "S21 confirmatory")])
print("\n[S21 exploratory]")
show_tests([paired(col(dL, a, "gap_unif"), col(dL, "SS (mean)", "gap_unif"), f"S21 {a} vs SS", "S21 exploratory", "exploratory")
            for a in ["SS CVaR selected level", "SS linear ramp", "SS ramp + CVaR 0.1", "Transferred p=2"]])
print("\n[S21 vs Random]")
tl = show_tests([paired(col(dL, a, "gap_unif"), col(dL, "Random", "gap_unif"), f"S21 {a} vs Random", "S21 vs Random")
                 for a in list(LAND)[:8]])
for _, t in tl.iterrows():
    viab("ADMM standalone: " + t["label"].split("S21 ")[1].split(" vs")[0], "ADMM mitigation", "Random", t)
```

## 24. Hardware validation at $n=8$

Five instances, random angles (unnormalized Hamiltonian) and transferred angles (normalized). References: uniform distribution, noiseless distribution, device-matched simulation of the transpiled circuit, and noisy simulation of the logical circuit. Hardware runs with and without dynamical decoupling and twirling. Saved counts are reused unless `RUN_HW_N8 = True`.

```python
HW_SHOTS = 4096; SIM_SHOTS = 8192; HW_N = 8

def find_fake(name):
    import qiskit_ibm_runtime.fake_provider as fpv
    target = name.replace("ibm_", "").replace("_", "").lower()
    for nm_ in dir(fpv):
        if nm_.startswith("Fake") and nm_[4:].replace("V2", "").lower() == target:
            return getattr(fpv, nm_)()
    return None

def ibm_backend():
    from qiskit_ibm_runtime import QiskitRuntimeService
    service = QiskitRuntimeService(channel="ibm_quantum_platform", token=os.environ.get("IBM_QUANTUM_TOKEN"))
    return service.backend(HW_BACKEND_NAME) if HW_BACKEND_NAME else service.least_busy(operational=True, simulator=False)

def hw_circuits():
    out = []
    for s in range(N_HW):
        Qp, h = first_zupdate(HW_N, s); st = exact_stats(Qp, h, keep=True)
        for angset in ("random", "transferred"):
            norm = angset == "transferred"
            ans, scale, off, _ = qaoa_ansatz(Qp, h, 2, normalize=norm)
            prm = TRANSFER[("admm", 2)] if norm else np.random.default_rng(s).uniform(-np.pi / 4, np.pi / 4, ans.num_parameters)
            out.append({"seed": s, "angles": angset, "st": st, "circ": ans.assign_parameters(prm),
                        "p_noiseless": qaoa_probs_fast((st["C"] - off) / scale, prm, 2, HW_N)})
    return out

HWC = hw_circuits(); fake = find_fake(HW_BACKEND_NAME)
if fake is None:
    print(f"No fake backend for {HW_BACKEND_NAME}; device simulations skipped.")
else:
    from qiskit.transpiler import generate_preset_pass_manager
    pm_fake = generate_preset_pass_manager(optimization_level=3, backend=fake, seed_transpiler=11)
    sim_dev = AerSimulator.from_backend(fake); nm_fake = NoiseModel.from_backend(fake); sim_log = AerSimulator(noise_model=nm_fake)
    for rec in HWC:
        tq = pm_fake.run(rec["circ"]); ops = tq.count_ops()
        rec["depth_t"], rec["cx_t"] = tq.depth(), sum(ops.get(g, 0) for g in ("cx", "cz", "ecr", "rzz"))
        rec["p_device"] = counts_to_probs(sim_dev.run(tq, shots=SIM_SHOTS, seed_simulator=rec["seed"]).result().get_counts(), HW_N)
        tl = transpile(rec["circ"], basis_gates=nm_fake.basis_gates, optimization_level=0)
        ops_l = tl.count_ops(); rec["cx_l"] = sum(ops_l.get(g, 0) for g in ("cx", "cz", "ecr", "rzz"))
        rec["p_logical"] = counts_to_probs(sim_log.run(tl, shots=SIM_SHOTS, seed_simulator=rec["seed"]).result().get_counts(), HW_N)

HW_FILE = "S22_hardware_counts.json"
if RUN_HW_N8:
    from qiskit_ibm_runtime import SamplerV2 as IBMSampler
    from qiskit.transpiler import generate_preset_pass_manager
    backend = ibm_backend(); pm_hw = generate_preset_pass_manager(optimization_level=3, backend=backend, seed_transpiler=11)
    tqs = [pm_hw.run(r["circ"]) for r in HWC]; saved = {"backend": backend.name, "runs": {}}
    for mode in ("suppressed", "unsuppressed"):
        smp = IBMSampler(mode=backend)
        smp.options.dynamical_decoupling.enable = mode == "suppressed"
        if mode == "suppressed": smp.options.dynamical_decoupling.sequence_type = "XY4"
        smp.options.twirling.enable_gates = mode == "suppressed"; smp.options.twirling.enable_measure = mode == "suppressed"
        job = smp.run([(t,) for t in tqs], shots=HW_SHOTS); print(f"{mode}: job {job.job_id()}"); res = job.result()
        saved["runs"][mode] = {f"{r['seed']}|{r['angles']}": res[i].data.meas.get_counts() for i, r in enumerate(HWC)}
    json.dump(saved, open(HW_FILE, "w"))
hw = json.load(open(HW_FILE)) if os.path.exists(HW_FILE) else None
print("Hardware counts:", f"loaded ({hw['backend']})" if hw else "not available; simulation-only analysis")

rows = []; unif = np.full(2 ** HW_N, 2.0 ** -HW_N)
for r in HWC:
    st = r["st"]; ug = st["Cmean"] - st["Cstar"]
    dists = {"uniform": unif, "noiseless": r["p_noiseless"]}
    if "p_device" in r: dists["device sim (transpiled)"] = r["p_device"]; dists["noisy sim (logical)"] = r["p_logical"]
    if hw:
        for mode, runs in hw["runs"].items():
            dists[f"hardware ({mode})"] = counts_to_probs(runs[f"{r['seed']}|{r['angles']}"], HW_N)
    for name, p in dists.items():
        row = {"seed": r["seed"], "angles": r["angles"], "condition": name, "mean_gap_rel_uniform": float((p @ st["C"] - st["Cstar"]) / ug),
               "p_opt": float(p[np.argmin(st["C"])]), "tvd_vs_uniform": tvd(p, unif), "tvd_vs_noiseless": tvd(p, r["p_noiseless"])}
        if "p_device" in r: row["tvd_vs_device"] = tvd(p, r["p_device"]); row["hellinger_vs_device"] = hellinger_fidelity(p, r["p_device"])
        rows.append(row)
    if "cx_t" in r: print(f"  seed {r['seed']} {r['angles']:<11s}: transpiled two-qubit gates {r['cx_t']}, logical {r['cx_l']}")
dH = pd.DataFrame(rows); dH.to_csv("S22_hardware_analysis.csv", index=False)
print("\nMean gap relative to uniform:")
print(dH.pivot_table(index="condition", columns="angles", values="mean_gap_rel_uniform").round(3))
if "tvd_vs_device" in dH:
    print("\nTotal variation distance to the device simulation:")
    print(dH.pivot_table(index="condition", columns="angles", values="tvd_vs_device").round(3))
```

## 26. Fixed absolute budget: scaling to $n=24$

Budget fixed at 768 evaluations while $n$ grows. For the subproblem at ADMM iteration 1 (after one exact update), the exact QAOA distribution gives the probability $q$ of sampling an optimal bitstring and the exact expected gap of the best of $\mathcal{B}$ samples; these are compared with uniform sampling. Sampled runs compare transferred and optimized QAOA with the classical solvers at the same budget. A second part runs ADMM at $n=16$ and $20$ with $K=8$, where the total evaluation count per run is far below $2^n$.

```python
def subproblem_k1(n, seed, rho=RHO):
    Q, c, B, b = generate_instance(n, seed); x = np.zeros(n); lam = np.zeros(n)
    Qp, h = build_qubo_subproblem(x, lam, Q, B, b, rho); C0 = cost_vector_big(Qp, h)
    z = z_from_index(int(np.argmin(C0)), n); del C0
    x = x_update(z, lam, c, B, b, rho); lam = lam + rho * (x + B @ z - b)
    Qp, h = build_qubo_subproblem(x, lam, Q, B, b, rho)
    return Qp, h, exact_stats_big(Qp, h)

def best_of_B_gap(pr, g_sorted, order, Bs):
    ps = pr[order]; ge = np.clip(np.cumsum(ps[::-1])[::-1], 0, 1); gt = np.clip(ge - ps, 0, 1)
    return {B: float(((ge ** B) - (gt ** B)) @ g_sorted) for B in Bs}

FIX_ANGLES = {"Transferred p=2": (TRANSFER[("admm", 2)], 2), "Transferred p=3": (TRANSFER[("admm", 3)], 3)}

def _run_fixed():
    arow, srow = [], []
    for n_ in FIX_SIZES:
        t0 = time.time()
        for s in range(FIX_SEEDS(n_)):
            Qp, h, st = subproblem_k1(n_, s); Cv = st["C"]
            du = st["Cmean"] - st["Cstar"]; opt = Cv - st["Cstar"] <= 1e-7 * max(1, abs(st["Cstar"]))
            order = np.argsort(Cv, kind="stable"); g_sorted = (Cv[order] - st["Cstar"]) / du
            dists = {"Uniform": None}
            d_sc = scaled_diagonal(Qp, h, Cv)
            for nm_, (ang, p_) in FIX_ANGLES.items():
                pr = qaoa_probs_fast(d_sc, ang, p_, n_); dists[nm_] = pr / pr.sum()
            for nm_, pr in dists.items():
                if pr is None:
                    qv = float(opt.mean()); gbar = 1.0
                    pr_u = np.full(len(Cv), 1.0 / len(Cv)); eg = best_of_B_gap(pr_u, g_sorted, order, FIX_B_GRID); del pr_u
                else:
                    qv = float(pr[opt].sum()); gbar = float((pr @ Cv - st["Cstar"]) / du)
                    eg = best_of_B_gap(pr, g_sorted, order, FIX_B_GRID)
                row = {"n": n_, "seed": s, "dist": nm_, "n_opt": int(opt.sum()), "q": qv, "gbar": gbar}
                for B_ in FIX_B_GRID:
                    row[f"success_B{B_}"] = 1 - (1 - qv) ** B_; row[f"Egap_B{B_}"] = eg[B_]
                arow.append(row)
            del dists, order, g_sorted
            solvers = {"Transferred p=2": lambda: run_qaoa_fixed(Qp, h, TRANSFER[("admm", 2)], Cv, 2, FIX_BUDGET, s, sim="fast"),
                       "SA": lambda: run_sa(Qp, h, FIX_BUDGET, s),
                       "SA tuned": lambda: run_sa_restarts(Qp, h, FIX_BUDGET, s, **SA_TUNED),
                       "Tabu": lambda: run_tabu(Qp, h, FIX_BUDGET, s),
                       "Local search": lambda: run_local_search(Qp, h, FIX_BUDGET, s),
                       "Random": lambda: run_random(Qp, h, FIX_BUDGET, s, Cv),
                       "SDP": lambda: run_sdp(Qp, h, FIX_BUDGET, s)}
            if n_ <= FIX_OPT_MAX_N:
                solvers["QAOA p=2"] = lambda: run_qaoa(Qp, h, Cv, reps=2, shots=SHOTS, maxiter=MAXITER, seed=s, sim="fast")
            for nm_, fn in solvers.items():
                t1 = time.perf_counter(); out = fn(); wt = time.perf_counter() - t1
                srow.append({"n": n_, "seed": s, "arm": nm_, **gaps(out["best"], st), "evals": out.get("evals", np.nan), "wall": wt})
            del Qp, h, st, Cv
        print(f"  [R26] n={n_} done ({time.time() - t0:.0f}s)")
    return {"analytic": pd.DataFrame(arow), "sampled": pd.DataFrame(srow)}

FX = cached("R26_fixed_budget", _run_fixed); dFa, dFs = FX["analytic"], FX["sampled"]
summ = dFa.groupby(["n", "dist"]).agg(q_geo=("q", lambda v: float(np.exp(np.mean(np.log(np.maximum(v, 1e-300)))))),
                                      success_768=("success_B768", "mean"), Egap_768=("Egap_B768", "mean"), gbar=("gbar", "mean")).unstack()
print(summ.round(5))
print("\nSampled best-of-768 success rate:"); print(dFs.groupby(["n", "arm"])["hit"].mean().unstack().round(3))
for nm_ in FIX_ANGLES:
    a_ = dFa[dFa.dist == nm_].sort_values(["n", "seed"]); u_ = dFa[dFa.dist == "Uniform"].sort_values(["n", "seed"])
    amp = np.log2(a_.q.values / u_.q.values); ns_ = a_.n.values
    slope = np.polyfit(ns_, amp, 1)[0]; rng_ = np.random.default_rng(0); bs = []
    for _ in range(2000):
        i = rng_.integers(0, len(ns_), len(ns_))
        if np.ptp(ns_[i]) > 0: bs.append(np.polyfit(ns_[i], amp[i], 1)[0])
    print(f"{nm_}: log2 amplification of the optimum grows by {slope:+.3f} per qubit [{np.percentile(bs, 2.5):+.3f}, {np.percentile(bs, 97.5):+.3f}]")
for n_ in FIX_SIZES:
    s_ = dFs[dFs.n == n_]
    for comp in ("SA tuned", "Tabu", "Random"):
        paired(col(s_, "Transferred p=2", "gap_unif"), col(s_, comp, "gap_unif"), f"R26 n={n_} transferred vs {comp}", "R26 fixed budget", "exploratory")
    viab(f"Fixed budget 768, n={n_}: transferred QAOA", "Fixed budget", "Random",
         paired(col(s_, "Transferred p=2", "gap_unif"), col(s_, "Random", "gap_unif"), record=False))
```

```python
arms_EMB = {"Exact": mk_exact(), "QAOA p=2 transferred": mk_fixed(TRANSFER[("admm", 2)], 2, FIX_BUDGET, sim="fast"),
            "QAOA p=2": mk_qaoa(reps=2, shots=SHOTS, maxiter=MAXITER, sim="fast"),
            "SA": mk_sa(FIX_BUDGET), "SA tuned": mk_sa_tuned(FIX_BUDGET), "Tabu": mk_tabu(FIX_BUDGET), "Random": mk_random(FIX_BUDGET)}
def _run_emb():
    its, sms = [], []
    for n_ in EMB_SIZES:
        out = run_arms_embedded(arms_EMB, list(range(N_EMB)), n_, K_EMB, f"R26 embedded n={n_}", big=True)
        for d in (out["iters"], out["seeds"]): d["n"] = n_
        its.append(out["iters"]); sms.append(out["seeds"])
    return {"iters": pd.concat(its, ignore_index=True), "seeds": pd.concat(sms, ignore_index=True)}
EMB = cached("R26_embedded", _run_emb); dEm = EMB["seeds"]
print(dEm.groupby(["n", "arm"])[["gap_unif", "hit", "mbo_opt_found", "mbo_gap_best_unif"]].mean().round(3))
for n_ in EMB_SIZES:
    s_ = dEm[dEm.n == n_]
    show_tests([paired(col(s_, "QAOA p=2 transferred", c_), col(s_, a, c_), f"R26 embedded n={n_} transferred vs {a} ({lab})", "R26 embedded", "exploratory")
                for a in ("SA tuned", "Tabu", "Random") for c_, lab in (("gap_unif", "subproblem"), ("mbo_gap_best_unif", "ADMM level"))])
```

## 27. Penalty sweep

ADMM-embedded, $n=12$, $K=12$, $\rho\in\{0.5,1,1.5,2,3,4\}$. Checks of Theorem 1 on the exact-solver runs: (a) at $\rho=1$ the global optimum is found by the second iteration; (b) once the iterate is fixed, the residual is divided by $1+\rho$ per iteration; (c) a final iterate held for the last three iterations satisfies the stalling inequality. The transferred angles are those trained at $\rho=2$.

```python
arms_P = {"Exact": mk_exact(), "SA": mk_sa(BUDGET), "SA tuned": mk_sa_tuned(BUDGET), "Random": mk_random(BUDGET),
          "QAOA p=2 transferred": mk_fixed(TRANSFER[("admm", 2)], 2, BUDGET, sim="fast"),
          "QAOA p=2": mk_qaoa(reps=2, shots=SHOTS, maxiter=MAXITER, sim="fast")}
def _run_rho():
    its, sms = [], []
    for r_ in RHOS:
        out = run_arms_embedded(arms_P, list(range(N_RHO)), N, K_MAIN, f"R27 rho={r_}", rho=r_)
        for d in (out["iters"], out["seeds"]): d["rho"] = r_
        its.append(out["iters"]); sms.append(out["seeds"])
    return {"iters": pd.concat(its, ignore_index=True), "seeds": pd.concat(sms, ignore_index=True)}
RH = cached("R27_penalty", _run_rho); dP, dPi = RH["seeds"], RH["iters"]
print(dP.pivot_table(index="rho", columns="arm", values="mbo_opt_found").round(3))

def theorem_checks():
    rows = []
    for r_ in RHOS:
        w = (r_ - 1) / 2; ex = dPi[(dPi.arm == "Exact") & (dPi.rho == r_)]
        for s, g in ex.groupby("seed"):
            g = g.sort_values("iter"); zi = g.zidx.values; rs = g.resid.values
            Q, c, B, b = generate_instance(N, int(s)); Fall = mbo_feasible_values(Q, c, B, b); Fstar = Fall.min()
            tolF = 1e-7 * max(1, abs(Fstar))
            first = next((k for k in range(len(zi)) if Fall[zi[k]] - Fstar <= tolF), -1)
            ratios = [rs[k - 1] / rs[k] for k in range(2, len(zi)) if zi[k] == zi[k - 1] == zi[k - 2] and rs[k] > 1e-13]
            held = len(zi) >= 3 and zi[-1] == zi[-2] == zi[-3]
            stall = np.nan
            if held:
                Z = all_bits(N, np.float64); zb = z_from_index(zi[-1], N)
                stall = float(np.all(Fall[zi[-1]] <= Fall + w * ((Z - zb) @ B.T) ** 2 @ np.ones(N) + 1e-9))
            rows.append({"rho": r_, "seed": s, "first_opt_iter": first, "global_by_iter2": float(0 <= first <= 1),
                         "ratio_dev": float(np.max(np.abs(np.array(ratios) - (1 + r_)))) if ratios else np.nan,
                         "held_last3": float(held), "stalling_ineq": stall,
                         "stalled_suboptimal": float(held and Fall[zi[-1]] - Fstar > tolF)})
    return pd.DataFrame(rows)
thm = theorem_checks(); thm.to_csv("R27_theorem_checks.csv", index=False)
print(thm.groupby("rho").agg(global_by_iter2=("global_by_iter2", "mean"), max_ratio_dev=("ratio_dev", "max"),
                             held_last3=("held_last3", "mean"), stalling_ineq_holds=("stalling_ineq", "mean"),
                             stalled_suboptimal=("stalled_suboptimal", "mean")).round(4))
print("\n[R27 rho=1 vs rho=2, best feasible gap]")
show_tests([paired(col(dP[dP.rho == 1.0], a, "mbo_gap_best_unif"), col(dP[dP.rho == 2.0], a, "mbo_gap_best_unif"),
                   f"R27 {a}: rho=1 vs rho=2", "R27 penalty") for a in arms_P])
```

## 28. Mixed-binary family that does not reduce to a QUBO

$\min \tfrac12 x^\top P x - c^\top x + z^\top Q z$ subject to $Ax+Bz=b$, $-1\le x\le 1$, with $x\in\mathbb{R}^{16}$, $z\in\{0,1\}^{12}$, 8 coupling constraints, and $P\succ 0$, $P\neq I$. The $x$-update is a box-constrained quadratic program; the problem optimum is obtained by solving the continuous problem for every $z$. Theorem 1 does not apply. A final cell verifies the generalized identity for $P\neq I$ without bounds: the binary step equals $F(z)+\tfrac12\|Bz-a^k\|^2_{\rho I-P}$ up to a constant.

```python
def generate_general(n, seed, m=GEN_M, r=GEN_R, box=GEN_BOX):
    rng = np.random.default_rng(10000 + seed)
    Q = rng.uniform(-1, 1, (n, n)); Q = (Q + Q.T) / 2; np.fill_diagonal(Q, rng.uniform(-1, 1, n))
    L = rng.normal(size=(m, m)) / np.sqrt(m); P = L @ L.T + 0.5 * np.eye(m)
    c = rng.uniform(-1, 1, m); A_ = rng.normal(size=(r, m)); B = rng.uniform(-0.7, 0.7, (r, n))
    x0 = rng.uniform(-0.5 * box, 0.5 * box, m); z0 = rng.integers(0, 2, n).astype(float)
    return {"Q": Q, "P": P, "c": c, "A": A_, "B": B, "b": A_ @ x0 + B @ z0, "box": box}

def general_values(inst):
    m = inst["P"].shape[0]; n = inst["B"].shape[1]
    x = cp.Variable(m); rhs = cp.Parameter(inst["b"].shape[0])
    prob = cp.Problem(cp.Minimize(0.5 * cp.quad_form(x, cp.psd_wrap(inst["P"])) - inst["c"] @ x),
                      [inst["A"] @ x == rhs, x >= -inst["box"], x <= inst["box"]])
    out = np.full(2 ** n, np.inf); Z = all_bits(n, np.float64); active = 0
    for k in range(2 ** n):
        rhs.value = inst["b"] - inst["B"] @ Z[k]
        try:
            prob.solve(solver=cp.CLARABEL)
        except Exception:
            continue
        if prob.status in ("optimal", "optimal_inaccurate"):
            out[k] = prob.value + Z[k] @ inst["Q"] @ Z[k]; active += bool(np.any(np.abs(x.value) > inst["box"] - 1e-6))
    return out, active / max(int(np.isfinite(out).sum()), 1)

def run_admm_general(inst, Fall, solver, K, rho, seed):
    Q, P, c, A_, B, b, box = (inst[k] for k in ("Q", "P", "c", "A", "B", "b", "box"))
    n = B.shape[1]; m = P.shape[0]; Lc = np.linalg.cholesky(P + rho * A_.T @ A_)
    x = np.zeros(m); lam = np.zeros(len(b)); ctx = {}; rows = []; Zs = []
    for k in range(K):
        Qp = Q + (rho / 2.0) * (B.T @ B); Qp = (Qp + Qp.T) / 2.0; h = B.T @ lam + rho * (B.T @ (A_ @ x - b))
        st = exact_stats(Qp, h, keep=True); out = solver(Qp, h, st, seed * 1000 + k, ctx); z = np.asarray(out["z"], float)
        g_ = -c + A_.T @ lam + rho * A_.T @ (B @ z - b)
        x = lsq_linear(Lc.T, -np.linalg.solve(Lc, g_), bounds=(-box, box), method="bvls", tol=1e-12).x
        res = A_ @ x + B @ z - b; lam = lam + rho * res
        rows.append({"iter": k, **gaps(out["best"], st), "resid": float(np.linalg.norm(res)), "zidx": z_to_index(z)})
        Zs.append(z_to_index(z))
    fin = np.isfinite(Fall); Fstar = float(Fall[fin].min()); Fmean = float(Fall[fin].mean()); Ff = Fall[Zs]
    tolF = 1e-6 * max(1, abs(Fstar)); fb = float(np.min(Ff))
    df = pd.DataFrame(rows); df["F_iter"] = Ff
    return df, {"resid_final": rows[-1]["resid"], "mbo_opt_found": float(fb - Fstar <= tolF),
                "mbo_gap_best_unif": float((fb - Fstar) / max(Fmean - Fstar, 1e-12)) if np.isfinite(fb) else np.nan,
                "feasible_fraction": float(fin.mean()), "iterates_feasible": float(np.isfinite(Ff).mean())}

arms_G = {"Exact": mk_exact(), "SA": mk_sa(BUDGET), "SA tuned": mk_sa_tuned(BUDGET), "Random": mk_random(BUDGET),
          "QAOA p=2 transferred": mk_fixed(TRANSFER[("admm", 2)], 2, BUDGET, sim="fast"),
          "QAOA p=2": mk_qaoa(reps=2, shots=SHOTS, maxiter=MAXITER, sim="fast")}
def _run_general():
    sms, its = [], []; t0 = time.time()
    for s in range(N_GEN):
        inst = generate_general(N, s); Fall, act = general_values(inst)
        for r_ in RHOS_GEN:
            for a, f in arms_G.items():
                df, sm = run_admm_general(inst, Fall, f, K_MAIN, r_, s)
                df["seed"], df["arm"], df["rho"] = s, a, r_; its.append(df)
                sms.append({"seed": s, "arm": a, "rho": r_, **df[["gap_unif", "hit"]].mean().to_dict(), **sm, "box_active_fraction": act})
        if (s + 1) % max(1, N_GEN // 5) == 0: print(f"  [R28] instance {s + 1}/{N_GEN} ({time.time() - t0:.0f}s)")
    return {"seeds": pd.DataFrame(sms), "iters": pd.concat(its, ignore_index=True)}
GEN = cached("R28_general", _run_general); dG = GEN["seeds"]
print(dG.pivot_table(index="arm", columns="rho", values=["mbo_opt_found", "mbo_gap_best_unif", "gap_unif"]).round(3))
print(f"Feasible fraction of z: median {dG.feasible_fraction.median():.3f}; box active at the continuous optimum: median {dG.box_active_fraction.median():.3f}")
for r_ in RHOS_GEN:
    s_ = dG[dG.rho == r_]
    show_tests([paired(col(s_, "QAOA p=2 transferred", "mbo_gap_best_unif"), col(s_, a, "mbo_gap_best_unif"),
                       f"R28 rho={r_:g} transferred vs {a} (problem level)", "R28 general family", "exploratory")
                for a in ("Exact", "SA", "SA tuned", "Random", "QAOA p=2")])
```

```python
def check_general_identity(n=8, seeds=range(5), rhos=(0.5, 1.0, 2.0), iters=4):
    worst = 0.0
    for s in seeds:
        rng = np.random.default_rng(20000 + s)
        Q = rng.uniform(-1, 1, (n, n)); Q = (Q + Q.T) / 2; c = rng.uniform(-1, 1, n)
        B = rng.uniform(-0.7, 0.7, (n, n)); b = rng.uniform(-1, 1, n)
        L = rng.normal(size=(n, n)) / np.sqrt(n); P = L @ L.T + 0.5 * np.eye(n)
        Z = all_bits(n, np.float64); X = b[None, :] - Z @ B.T
        Fz = 0.5 * np.einsum("ij,jk,ik->i", X, P, X) - X @ c + np.einsum("ij,jk,ik->i", Z, Q, Z)
        for r_ in rhos:
            x = np.zeros(n); lam = np.zeros(n); Mw = r_ * np.eye(n) - P
            for k in range(iters):
                Qp, h = build_qubo_subproblem(x, lam, Q, B, b, r_); Cq = cost_vector(Qp, h)
                if k >= 1:
                    D_ = Z @ B.T - (b - x)[None, :]
                    worst = max(worst, float(np.ptp(Cq - Fz - 0.5 * np.einsum("ij,jk,ik->i", D_, Mw, D_))))
                z = Z[int(np.argmin(Cq))]
                x = np.linalg.solve(P + r_ * np.eye(n), c - lam - r_ * (B @ z - b)); lam = lam + r_ * (x + B @ z - b)
    return worst
spread = check_general_identity()
print(f"Generalized proximal identity, max spread of the difference over all z: {spread:.2e}")
assert spread < 1e-8
```

## 29. Hardware at $n=12$

Ten $n=12$ first subproblems, $p=2$, transferred and random angles, both with the normalized Hamiltonian. Each circuit is sampled 4096 times; best-sample scoring uses consecutive blocks of 768 shots, compared with random search, SA and tuned SA at 768 evaluations and with noiseless and device-matched simulations. Readout errors are mitigated with a tensored assignment model calibrated on the measured physical qubits, followed by projection onto the nearest probability distribution. Retention $\eta=(1-\bar g_{\mathrm{noisy}})/(1-\bar g_{\mathrm{ideal}})$ is reported with shot-bootstrap intervals. With `RUN_HW_ADMM = True`, three instances run six ADMM iterations with the binary step executed on hardware.

```python
HW12_FILE = "R29_hardware_n12.json"; HW12_MODES = ("suppressed", "unsuppressed")
HW12_JOB_IDS = {"suppressed": "dauhi8qhcrkc73dvls90", "unsuppressed": "dauhj8ahcrkc73dvltu0"}  # submitted jobs, used to retrieve device usage

def job_quantum_seconds(job):
    for get in (lambda: job.usage(), lambda: job.metrics()["usage"]["quantum_seconds"], lambda: job.metrics()["usage"]["seconds"]):
        try:
            v = get()
            if v is not None: return float(v)
        except Exception:
            pass
    return None

def nearest_prob(qv):
    qv = np.asarray(qv, float); o = np.argsort(qv); qs = qv[o].copy(); n_ = len(qs); acc = 0.0; i = 0
    while i < n_ and qs[i] + acc / (n_ - i) < 0:
        acc += qs[i]; qs[i] = 0.0; i += 1
    if i < n_: qs[i:] += acc / (n_ - i)
    out = np.empty(n_); out[o] = qs; return out / out.sum()

def mitigate(p, e01, e10):
    n_ = len(e01); T_ = p.reshape([2] * n_)
    for i in range(n_):
        Ainv = np.linalg.inv(np.array([[1 - e01[i], e10[i]], [e01[i], 1 - e10[i]]])); ax = n_ - 1 - i
        T_ = np.moveaxis(np.tensordot(Ainv, T_, axes=([1], [ax])), 0, ax)
    return nearest_prob(T_.reshape(-1))

def bits_to_idx(bitstrings):
    return np.fromiter((int(b, 2) for b in bitstrings), dtype=np.int64, count=len(bitstrings))

def hw12_circuits():
    out = []
    for s in range(N_HW12):
        Qp, h = first_zupdate(HW12_N, s); st = exact_stats(Qp, h, keep=True)
        for angset in ("transferred", "random"):
            ans, scale, off, _ = qaoa_ansatz(Qp, h, 2)
            prm = TRANSFER[("admm", 2)] if angset == "transferred" else np.random.default_rng(100 + s).uniform(-np.pi / 4, np.pi / 4, 4)
            out.append({"seed": s, "angles": angset, "Qp": Qp, "h": h, "st": st, "prm": np.asarray(prm),
                        "circ": ans.assign_parameters(prm), "p_noiseless": qaoa_probs_fast((st["C"] - off) / scale, prm, 2, HW12_N)})
    return out
HW12 = hw12_circuits()

if RUN_HW_N12 and not (RESUME and os.path.exists(HW12_FILE)):
    from qiskit_ibm_runtime import SamplerV2 as IBMSampler
    from qiskit.transpiler import generate_preset_pass_manager
    backend = ibm_backend(); pm_hw = generate_preset_pass_manager(optimization_level=3, backend=backend, seed_transpiler=11)
    tqs = [pm_hw.run(r["circ"]) for r in HW12]
    layouts = [list(t.layout.final_index_layout()) for t in tqs]
    qubits = sorted({q_ for L_ in layouts for q_ in L_})
    saved = {"backend": backend.name, "layouts": layouts, "runs": {}, "usage": {}, "shots": {}, "jobs": {}}
    for mode in HW12_MODES:
        smp = IBMSampler(mode=backend)
        smp.options.dynamical_decoupling.enable = mode == "suppressed"
        if mode == "suppressed": smp.options.dynamical_decoupling.sequence_type = "XY4"
        smp.options.twirling.enable_gates = mode == "suppressed"; smp.options.twirling.enable_measure = mode == "suppressed"
        job = smp.run([(t,) for t in tqs], shots=HW12_SHOTS); print(f"{mode}: job {job.job_id()}"); res = job.result()
        saved["runs"][mode] = {f"{r['seed']}|{r['angles']}": res[i].data.meas.get_bitstrings() for i, r in enumerate(HW12)}
        saved["jobs"][mode] = job.job_id(); saved["usage"][mode] = job_quantum_seconds(job)
        saved["shots"][mode] = HW12_SHOTS * len(tqs)
    nq = backend.num_qubits; cal = []
    for flip in (0, 1):
        qc = QuantumCircuit(nq, len(qubits))
        if flip:
            for q_ in qubits: qc.x(q_)
        for j, q_ in enumerate(qubits): qc.measure(q_, j)
        cal.append(transpile(qc, backend, initial_layout=list(range(nq)), optimization_level=0))
    cres = IBMSampler(mode=backend).run([(c_,) for c_ in cal], shots=HW12_SHOTS).result()
    b0, b1 = cres[0].data.c.get_bitstrings(), cres[1].data.c.get_bitstrings()
    k_ = len(qubits)
    e01 = {str(qubits[j]): float(np.mean([bs[k_ - 1 - j] == "1" for bs in b0])) for j in range(k_)}
    e10 = {str(qubits[j]): float(np.mean([bs[k_ - 1 - j] == "0" for bs in b1])) for j in range(k_)}
    saved["readout"] = {"e01": e01, "e10": e10}
    json.dump(saved, open(HW12_FILE, "w"))
hw12 = json.load(open(HW12_FILE)) if os.path.exists(HW12_FILE) else None
print("n=12 hardware data:", f"loaded ({hw12['backend']})" if hw12 else "not available")
```

```python
def block_scores(idx, st, block=HW12_BLOCK):
    Cv = st["C"]; nb = len(idx) // block; hits, gs = [], []
    for j in range(nb):
        g = gaps(float(Cv[idx[j * block:(j + 1) * block]].min()), st); hits.append(g["hit"]); gs.append(g["gap_unif"])
    return float(np.mean(hits)), float(np.mean(gs))

def mean_gap_rel(p, st): return float((p @ st["C"] - st["Cstar"]) / (st["Cmean"] - st["Cstar"]))

def _analyze_hw12():
    rows, eta_rows = [], []
    dev = None
    if fake is not None:
        from qiskit.transpiler import generate_preset_pass_manager
        pmf = generate_preset_pass_manager(optimization_level=3, backend=fake, seed_transpiler=11); simd = AerSimulator.from_backend(fake, method="statevector")
    for ir, r in enumerate(HW12):
        st = r["st"]; key = f"{r['seed']}|{r['angles']}"; gi = mean_gap_rel(r["p_noiseless"], st)
        fs = FastSampler(scaled_diagonal(r["Qp"], r["h"], st["C"]), 2, 900 + r["seed"])
        idx = np.concatenate([np.repeat(*fs.sample(r["prm"], HW12_BLOCK)) for _ in range(HW12_SHOTS // HW12_BLOCK)])
        hr, gr = block_scores(idx, st); rows.append({"seed": r["seed"], "angles": r["angles"], "condition": "noiseless", "hit": hr, "gap_unif": gr, "mean_gap_rel": gi})
        if fake is not None:
            tq = pmf.run(r["circ"]); mem = simd.run(tq, shots=HW12_SHOTS, memory=True, seed_simulator=r["seed"]).result().get_memory()
            idx = bits_to_idx([m_.replace(" ", "") for m_ in mem]); hr, gr = block_scores(idx, st)
            p = np.bincount(idx, minlength=2 ** HW12_N) / len(idx)
            rows.append({"seed": r["seed"], "angles": r["angles"], "condition": "device simulation", "hit": hr, "gap_unif": gr, "mean_gap_rel": mean_gap_rel(p, st)})
            eta_rows.append({"seed": r["seed"], "angles": r["angles"], "condition": "device simulation", "abs_1_minus_gideal": abs(1 - gi),
                             "eta": (1 - mean_gap_rel(p, st)) / (1 - gi), "eta_lo": np.nan, "eta_hi": np.nan})
        if hw12:
            L_ = hw12["layouts"][ir]
            e01 = np.array([hw12["readout"]["e01"][str(q_)] for q_ in L_]); e10 = np.array([hw12["readout"]["e10"][str(q_)] for q_ in L_])
            for mode, runs in hw12["runs"].items():
                idx = bits_to_idx(runs[key]); hr, gr = block_scores(idx, st)
                p_raw = np.bincount(idx, minlength=2 ** HW12_N) / len(idx); p_mit = mitigate(p_raw, e01, e10)
                for lab, p in (("raw", p_raw), ("mitigated", p_mit)):
                    rows.append({"seed": r["seed"], "angles": r["angles"], "condition": f"hardware {mode}, {lab}",
                                 "hit": hr if lab == "raw" else np.nan, "gap_unif": gr if lab == "raw" else np.nan, "mean_gap_rel": mean_gap_rel(p, st)})
                    rng_ = np.random.default_rng(r["seed"]); bs = []
                    for _ in range(200 if QUICK_TEST else 1000):
                        pb = rng_.multinomial(len(idx), p_raw) / len(idx)
                        pb = pb if lab == "raw" else mitigate(pb, e01, e10)
                        bs.append((1 - mean_gap_rel(pb, st)) / (1 - gi))
                    eta_rows.append({"seed": r["seed"], "angles": r["angles"], "condition": f"hardware {mode}, {lab}",
                                     "abs_1_minus_gideal": abs(1 - gi), "eta": (1 - mean_gap_rel(p, st)) / (1 - gi),
                                     "eta_lo": float(np.percentile(bs, 2.5)), "eta_hi": float(np.percentile(bs, 97.5))})
        if r["angles"] == "transferred":
            for nm_, fn in (("Random", lambda s_: run_random(r["Qp"], r["h"], HW12_BLOCK, s_, st["C"])),
                            ("SA", lambda s_: run_sa(r["Qp"], r["h"], HW12_BLOCK, s_)),
                            ("SA tuned", lambda s_: run_sa_restarts(r["Qp"], r["h"], HW12_BLOCK, s_, **SA_TUNED))):
                gs = [gaps(fn(1000 * r["seed"] + j)["best"], st) for j in range(HW12_SHOTS // HW12_BLOCK)]
                rows.append({"seed": r["seed"], "angles": "classical", "condition": nm_, "hit": float(np.mean([g["hit"] for g in gs])),
                             "gap_unif": float(np.mean([g["gap_unif"] for g in gs])), "mean_gap_rel": np.nan})
    return {"scores": pd.DataFrame(rows), "retention": pd.DataFrame(eta_rows)}
H12 = cached("R29_hardware_n12" if hw12 else "R29_simulation_n12", _analyze_hw12) if (hw12 or fake is not None) else None
if H12 is not None:
    d12, e12 = H12["scores"], H12["retention"]
    print(d12.groupby(["angles", "condition"])[["hit", "gap_unif", "mean_gap_rel"]].mean().round(3))
    print(e12.groupby(["angles", "condition"])["eta"].agg(["mean", "min", "max"]).round(3))
    if hw12:
        hwc = [c_ for c_ in d12.condition.unique() if c_.startswith("hardware") and c_.endswith("raw")]
        for c_ in hwc:
            s_ = d12[(d12.angles == "transferred") & (d12.condition == c_)].sort_values("seed")
            for comp in ("Random", "SA", "SA tuned"):
                paired(s_.gap_unif.values, d12[d12.condition == comp].sort_values("seed").gap_unif.values,
                       f"R29 {c_} transferred vs {comp}", "R29 hardware n=12", "exploratory")
        if not any(hw12["usage"].values()) and os.environ.get("IBM_QUANTUM_TOKEN"):
            try:
                from qiskit_ibm_runtime import QiskitRuntimeService
                svc = QiskitRuntimeService(channel="ibm_quantum_platform", token=os.environ.get("IBM_QUANTUM_TOKEN"))
                for m, jid in hw12.get("jobs", HW12_JOB_IDS).items():
                    hw12["usage"][m] = job_quantum_seconds(svc.job(jid))
                json.dump(hw12, open(HW12_FILE, "w")); print("QPU usage retrieved:", hw12["usage"])
            except Exception as e:
                print("QPU usage not retrievable:", type(e).__name__)
        HW_SEC_PER_SHOT = np.nanmean([hw12["usage"][m] / hw12["shots"][m] for m in hw12["usage"] if hw12["usage"][m]]) if any(hw12["usage"].values()) else np.nan
    else:
        HW_SEC_PER_SHOT = np.nan
else:
    HW_SEC_PER_SHOT = np.nan
print("Measured QPU seconds per shot:", HW_SEC_PER_SHOT)
```

## 31. A planted family and landscape diagnostics

The main family turns out to be benign for local search. The second family takes $Q$ from the Wishart planted ensemble (Hamze et al., 2020): Ising couplings $J=-WW^\top/n$ with zero diagonal, where the $\alpha n$ Gaussian columns of $W$ are orthogonal to a planted configuration that a random gauge then hides; here $\alpha=0.5$. The couplings are mapped to a QUBO, scaled to unit maximum magnitude, and used as $Q$ in problem (1) with weak coupling to the continuous block ($B\sim U[-0.05,0.05]$, $c,b\sim U[-0.1,0.1]$), so that the subproblems inherit the ensemble's landscape. Test instances use seed offset 30000; held-out training and tuning use 35000 onward.

Diagnostics for the second ADMM subproblem and for the reduced problem $F$:
- the number of single-flip local minima;
- the basin of the optimum, the fraction of all starting points from which steepest descent reaches an optimal bitstring;
- the number of optimal bitstrings;
- the fraction of negative eigenvalues of $Q_p$;
- the ratio of the norm of the field to that of the couplings.

```python
from scipy.stats import binom

def wishart_Q(n, alpha, rng):
    M = max(1, int(round(alpha * n))); t = np.ones(n)
    W = rng.normal(size=(n, M)); W = W - np.outer(t, t @ W) / n
    J = -(W @ W.T) / n; np.fill_diagonal(J, 0.0)
    g = rng.choice([-1.0, 1.0], n); J = J * np.outer(g, g)
    Q = -2.0 * J + np.diag(2.0 * J.sum(1))
    return Q / np.abs(Q).max()

def generate_instance_wishart(n, seed, alpha=None, bscale=None):
    rng = np.random.default_rng(W_OFFSET + seed)
    Q = wishart_Q(n, W_ALPHA if alpha is None else alpha, rng)
    s_ = W_BSCALE if bscale is None else bscale
    c = rng.uniform(-1, 1, n) * W_CBSCALE; B = rng.uniform(-s_, s_, (n, n)); b = rng.uniform(-1, 1, n) * W_CBSCALE
    return Q, c, B, b

_instance_base = _instance
def _instance(n, seed, gen="dense", p_link=None):
    return generate_instance_wishart(n, seed) if gen == "wishart" else _instance_base(n, seed, gen, p_link)

FAMILIES = {"main": "dense", "Wishart": "wishart"}

def subproblem_k1_gen(n, seed, gen="dense", rho=RHO):
    Q, c, B, b = _instance(n, seed, gen); x = np.zeros(n); lam = np.zeros(n)
    Qp, h = build_qubo_subproblem(x, lam, Q, B, b, rho); C0 = cost_vector_big(Qp, h)
    z = z_from_index(int(np.argmin(C0)), n); del C0
    x = x_update(z, lam, c, B, b, rho); lam = lam + rho * (x + B @ z - b)
    Qp, h = build_qubo_subproblem(x, lam, Q, B, b, rho)
    return Qp, h, exact_stats_big(Qp, h)

def reduced_qubo(n, seed, gen="dense"):
    # F(z) = z^T (Q + B^T B / 2) z + (B^T (c - b))^T z + const
    Q, c, B, b = _instance(n, seed, gen)
    QF = Q + 0.5 * (B.T @ B); QF = (QF + QF.T) / 2
    return QF, B.T @ (c - b), (Q, c, B, b)

def landscape_stats(C, tol=1e-7):
    C = np.where(np.isfinite(C), C, np.nanmax(C[np.isfinite(C)]) + 1.0)
    N_ = len(C); n = int(round(np.log2(N_))); idx = np.arange(N_, dtype=np.int64)
    best_nb = idx ^ 1; best_c = C[best_nb]
    for k in range(1, n):
        nb = idx ^ (1 << k); cn = C[nb]; m = cn < best_c
        best_nb = np.where(m, nb, best_nb); best_c = np.where(m, cn, best_c)
    loc = best_c >= C
    nxt = np.where(best_c < C, best_nb, idx)
    for _ in range(64):
        n2 = nxt[nxt]
        if np.array_equal(n2, nxt): break
        nxt = n2
    Cs = C.min(); opt = C - Cs <= tol * max(1.0, abs(Cs))
    return {"local_minima": int(loc.sum()), "basin_opt": float(opt[nxt].mean()), "n_opt": int(opt.sum())}

def spectral_stats(Qp, h):
    Qs = (Qp + Qp.T) / 2; ev = np.linalg.eigvalsh(Qs); off = Qs - np.diag(np.diag(Qs))
    return {"neg_eig_frac": float((ev < -1e-12).mean()), "field_coupling_ratio": float(np.linalg.norm(h) / max(np.linalg.norm(off), 1e-12))}

# consistency of the reduced QUBO with the enumerated problem values
for gen_ in FAMILIES.values():
    QF, hF, (Q_, c_, B_, b_) = reduced_qubo(8, 0, gen_)
    dlt = cost_vector(QF, hF) - mbo_feasible_values(Q_, c_, B_, b_)
    assert np.ptp(dlt) < 1e-9, gen_
print("Planted family and diagnostics ready; reduced-QUBO identity verified.")
```

```python
def _run_diag():
    rows = []
    for fam, gen in FAMILIES.items():
        for n_ in DIAG_SIZES:
            for s in range(N_DIAG(n_)):
                Qp, h, st = subproblem_k1_gen(n_, s, gen)
                QF, hF, _ = reduced_qubo(n_, s, gen); CF = cost_vector_big(QF, hF)
                rows.append({"family": fam, "n": n_, "seed": s, "object": "subproblem", **landscape_stats(st["C"]), **spectral_stats(Qp, h)})
                rows.append({"family": fam, "n": n_, "seed": s, "object": "reduced problem", **landscape_stats(CF), **spectral_stats(QF, hF)})
                del st, CF
    for s in range(N_DIAG_GEN):
        inst = generate_general(N, s); Fall, _ = general_values(inst)
        rows.append({"family": "general", "n": N, "seed": s, "object": "reduced problem", **landscape_stats(Fall),
                     "neg_eig_frac": np.nan, "field_coupling_ratio": np.nan})
    return {"stats": pd.DataFrame(rows)}
HD = cached("H31_landscape", _run_diag); dHd = HD["stats"]
print(dHd.groupby(["family", "n", "object"])[["local_minima", "basin_opt", "n_opt", "neg_eig_frac", "field_coupling_ratio"]]
      .median().round(3).to_string())
```

## 32. Transferred angles for the planted family; extended SA tuning

Angles are trained as for the main family, on the first, third and fifth subproblems of 12 held-out exact trajectories. SA is tuned on the same 36 held-out subproblems of each family over an extended grid of 125 configurations:
- initial temperature in {0.25, 0.5, 1, 2, 5};
- final temperature in {0.005, 0.02, 0.1, 0.3, 1};
- 1, 2, 4, 8 or 16 restarts.

The output reports whether the selected configuration lies on the edge of the grid.

```python
def training_set_gen(gen, n=N):
    out = []
    for s in range(TRAIN_OFFSET, TRAIN_OFFSET + N_TRAIN):
        for (Qp, h, st) in exact_trajectory(n, s, K=6, gen=gen)[::2]:
            out.append((Qp, h, st))
    return out

def train_angles_gen(train_raw, p):
    train = [_prep(Qp, h, st) for Qp, h, st in train_raw]; rng = np.random.default_rng(p); best = (np.inf, None)
    starts = [np.concatenate([np.linspace(0.5, 0.5 / p, p), np.linspace(0.5 / p, 0.5, p)])]
    starts += [rng.uniform(-np.pi / 4, np.pi / 4, 2 * p) for _ in range(TRAIN_RESTARTS - 1)]
    for x0 in starts:
        r = minimize(transfer_objective, x0, args=(p, train), method="Nelder-Mead",
                     options={"maxfev": TRAIN_MAXFEV, "xatol": 1e-4, "fatol": 1e-6})
        if r.fun < best[0]: best = (r.fun, r.x)
    return best[1], best[0]

TRAIN_W = training_set_gen("wishart")
if RESUME and os.path.exists("H32_wishart_angles.json"):
    TRANSFER_W = {int(k): np.array(v) for k, v in json.load(open("H32_wishart_angles.json")).items()}; print("[resume] loaded Wishart angles")
else:
    TRANSFER_W, rows_ = {}, []
    for p_ in (2, 3):
        x_, f_ = train_angles_gen(TRAIN_W, p_); TRANSFER_W[p_] = x_; rows_.append({"p": p_, "train_Egap_unif": f_, "angles": np.round(x_, 4).tolist()})
    json.dump({str(k): v.tolist() for k, v in TRANSFER_W.items()}, open("H32_wishart_angles.json", "w"))
    pd.DataFrame(rows_).to_csv("H32_wishart_training.csv", index=False)
for p_ in (2, 3):
    print(f"Planted-family angles p={p_}: {np.round(TRANSFER_W[p_], 3)}; training mean expected g_unif "
          f"{transfer_objective(TRANSFER_W[p_], p_, [_prep(*t) for t in TRAIN_W]):.3f}")
print(f"Main-family p=2 angles on the planted training set: {transfer_objective(TRANSFER[('admm', 2)], 2, [_prep(*t) for t in TRAIN_W]):.3f}")

def tune_sa_ext(train_raw, tag):
    path = f"H37_sa_tuning_{tag}.csv"
    if RESUME and os.path.exists(path):
        d = pd.read_csv(path); print(f"[resume] loaded SA tuning ({tag})")
    else:
        rows = []
        for (T0, Tm, r) in SA_GRID_EXT:
            g = [gaps(run_sa_restarts(Qp, h, BUDGET, 7000 + 10 * j + rep, r, T0, Tm)["best"], st)["gap_unif"]
                 for j, (Qp, h, st) in enumerate(train_raw) for rep in range(SA_TUNE_EXT_REPS)]
            rows.append({"T0": T0, "Tmin": Tm, "restarts": r, "gap_unif": float(np.mean(g))})
        d = pd.DataFrame(rows); d.to_csv(path, index=False)
    b = d.loc[d.gap_unif.idxmin()]; cfg = {"T0": float(b.T0), "Tmin": float(b.Tmin), "restarts": int(b.restarts)}
    edge = [k for k, vals in (("T0", (0.25, 5.0)), ("Tmin", (0.005, 1.0)), ("restarts", (1, 16))) if cfg[k] in vals]
    print(f"SA extended tuning ({tag}): {cfg}; mean g_unif {b.gap_unif:.5f}; on grid edge: {edge if edge else 'no'}")
    return cfg

TRAIN_MAIN = training_set_gen("dense")
SA_EXT = {"main": tune_sa_ext(TRAIN_MAIN, "main"), "Wishart": tune_sa_ext(TRAIN_W, "wishart")}
def mk_sa_cfg(B, cfg): return lambda Qp, h, st, s, ctx: run_sa_restarts(Qp, h, B, s, **cfg)
```

```python
def arms_family(fam):
    ang = TRANSFER_W if fam == "Wishart" else {2: TRANSFER[("admm", 2)], 3: TRANSFER[("admm", 3)]}
    arms = {"Exact": mk_exact(), "SA": mk_sa(BUDGET), "SA tuned (ext)": mk_sa_cfg(BUDGET, SA_EXT[fam]),
            "Tabu": mk_tabu(BUDGET), "Local search": mk_ls(BUDGET), "SDP": mk_sdp(BUDGET), "Random": mk_random(BUDGET),
            "QAOA p=2": mk_qaoa(reps=2, shots=SHOTS, maxiter=MAXITER, sim="fast"),
            "QAOA p=2 transferred": mk_fixed(ang[2], 2, BUDGET, sim="fast"),
            "QAOA p=3 transferred": mk_fixed(ang[3], 3, BUDGET, sim="fast")}
    if fam == "Wishart":
        arms["QAOA p=2 transferred (main-family angles)"] = mk_fixed(TRANSFER[("admm", 2)], 2, BUDGET, sim="fast")
    return arms

def _run_fam(fam):
    arms = arms_family(fam)
    if fam == "main": arms.pop("QAOA p=2")  # reported for the main family in Sections 11 and 27
    its, sms = [], []
    for r_ in RHOS_W:
        out = run_arms_embedded(arms, list(range(N_W)), N, K_MAIN, f"H33 {fam} rho={r_}", gen=FAMILIES[fam], rho=r_)
        for d in (out["iters"], out["seeds"]): d["rho"] = r_; d["family"] = fam
        its.append(out["iters"]); sms.append(out["seeds"])
    return {"iters": pd.concat(its, ignore_index=True), "seeds": pd.concat(sms, ignore_index=True)}
HW_ = cached("H33_wishart_embedded", lambda: _run_fam("Wishart")); dW8 = HW_["seeds"]; dW8["family"] = "Wishart"
HM_ = cached("H33_main_embedded", lambda: _run_fam("main")); dM8 = HM_["seeds"]; dM8["family"] = "main"
dFAM = pd.concat([dM8, dW8], ignore_index=True)
print(dFAM.pivot_table(index=["family", "arm"], columns="rho", values=["hit", "mbo_opt_found"]).round(3).to_string())
_ci = [{"family": f_, "rho": r_, "arm": a_, **dict(zip(("hit", "hit_lo", "hit_hi"), bootstrap_ci(s_.hit))), "gap_unif_x1e3": 1e3 * s_.gap_unif.mean(),
        "mbo_opt_found": s_.mbo_opt_found.mean()} for (f_, r_, a_), s_ in dFAM.groupby(["family", "rho", "arm"])]
dFAMCI = pd.DataFrame(_ci); dFAMCI.to_csv("H33_summary_ci.csv", index=False)
print("\nFraction of subproblems solved exactly, mean [95% bootstrap CI]; mean g_unif x 1e3; problem optimum found")
print(dFAMCI.round(3).to_string(index=False))
for r_ in RHOS_W:
    s_ = dW8[dW8.rho == r_]
    show_tests([paired(col(s_, "QAOA p=2 transferred", "gap_unif"), col(s_, a, "gap_unif"), f"H33 rho={r_:g} transferred vs {a}", f"H33 Wishart rho={r_:g}", "confirmatory")
                for a in ("SA tuned (ext)", "Tabu", "Local search", "SDP", "Random", "QAOA p=2")])
    show_tests([paired(col(s_, "QAOA p=2 transferred", "mbo_gap_best_unif"), col(s_, a, "mbo_gap_best_unif"), f"H33 rho={r_:g} transferred vs {a} (ADMM level)", f"H33 Wishart ADMM level rho={r_:g}", "exploratory")
                for a in ("SA tuned (ext)", "Tabu", "Random")])
```

## 34. Direct solution of the reduced QUBO

Problem (1) reduces to the QUBO $F$. Each solver is applied to $F$ directly with the total budget of one ADMM run, $K\times768$ evaluations, as the reference for the ADMM-level results at $\rho=2$ and $\rho=1$. The McNemar counts compare the two settings instance by instance and are descriptive.

```python
def mcnemar_exact(a, b):
    a, b = np.asarray(a, bool), np.asarray(b, bool); n01, n10 = int((a & ~b).sum()), int((~a & b).sum()); m = n01 + n10
    return {"a_only": n01, "b_only": n10, "P": float(min(1.0, 2 * binom.cdf(min(n01, n10), m, 0.5))) if m else 1.0}

def _run_dq():
    rows = []
    for fam, gen in FAMILIES.items():
        ang = TRANSFER_W if fam == "Wishart" else {2: TRANSFER[("admm", 2)], 3: TRANSFER[("admm", 3)]}
        for s in range(N_DQ):
            QF, hF, _ = reduced_qubo(N, s, gen); st = exact_stats(QF, hF, keep=True); Cv = st["C"]
            solvers = {"SA tuned (ext)": lambda: run_sa_restarts(QF, hF, DQ_BUDGET, s, **SA_EXT[fam]),
                       "Tabu": lambda: run_tabu(QF, hF, DQ_BUDGET, s), "Local search": lambda: run_local_search(QF, hF, DQ_BUDGET, s),
                       "Random": lambda: run_random(QF, hF, DQ_BUDGET, s, Cv), "SDP": lambda: run_sdp(QF, hF, DQ_BUDGET, s),
                       "QAOA p=2 transferred": lambda: run_qaoa_fixed(QF, hF, ang[2], Cv, 2, DQ_BUDGET, s, sim="fast"),
                       "QAOA p=3 transferred": lambda: run_qaoa_fixed(QF, hF, ang[3], Cv, 3, DQ_BUDGET, s, sim="fast")}
            for a, fn in solvers.items():
                out = fn(); g = gaps(out["best"], st)
                rows.append({"family": fam, "seed": s, "arm": a, "mbo_opt_found": g["hit"], "mbo_gap_best_unif": g["gap_unif"]})
    return {"seeds": pd.DataFrame(rows)}
DQ = cached("H34_direct_qubo", _run_dq); dDQ = DQ["seeds"]
ref = []
for (fam, r_, a), v in dFAM.groupby(["family", "rho", "arm"])["mbo_opt_found"].mean().items():
    ref.append({"family": fam, "setting": f"ADMM rho={r_:g}", "arm": a, "mbo_opt_found": v})
for _, t in dDQ.groupby(["family", "arm"])["mbo_opt_found"].mean().reset_index().iterrows():
    ref.append({"family": t.family, "setting": "direct QUBO", "arm": t.arm, "mbo_opt_found": t.mbo_opt_found})
dREF = pd.DataFrame(ref); dREF.to_csv("H34_problem_level_summary.csv", index=False)
print(dREF.pivot_table(index=["family", "arm"], columns="setting", values="mbo_opt_found").round(3).to_string())
for fam in FAMILIES:
    s_ = dDQ[dDQ.family == fam]
    show_tests([paired(col(s_, "QAOA p=2 transferred", "mbo_gap_best_unif"), col(s_, a, "mbo_gap_best_unif"),
                       f"H34 {fam} direct: transferred vs {a}", f"H34 direct QUBO {fam}", "exploratory")
                for a in ("SA tuned (ext)", "Tabu", "Local search", "SDP", "Random")])
mc_rows = []
for fam in FAMILIES:
    dq_ = dDQ[dDQ.family == fam]
    for a in sorted(set(dq_.arm) & set(dFAM[dFAM.family == fam].arm)):
        x_ = dq_[dq_.arm == a].sort_values("seed").set_index("seed").mbo_opt_found
        for r_ in RHOS_W:
            y_ = dFAM[(dFAM.family == fam) & (dFAM.rho == r_) & (dFAM.arm == a)].sort_values("seed").set_index("seed").mbo_opt_found
            ix = x_.index.intersection(y_.index)
            if len(ix): mc_rows.append({"family": fam, "arm": a, "admm_rho": r_, "n": len(ix), "direct": float(x_[ix].mean()), "admm": float(y_[ix].mean()),
                                        **mcnemar_exact(x_[ix].values, y_[ix].values)})
dMC34 = pd.DataFrame(mc_rows); dMC34.to_csv("H34_direct_vs_admm_mcnemar.csv", index=False); print(dMC34.round(4).to_string(index=False))
```

## 35. Shared trajectory with all baselines

Every solver is scored on the identical subproblems of the exact-solver trajectory ($n=12$, $K=12$, 80 instances per family), which removes differences between trajectories. Reported here: signed-rank tests on the per-instance normalized gap, pooled McNemar counts (descriptive), and two one-sided tests on the per-instance fraction solved exactly with a margin of 2 percentage points. Instance-level tests on the fraction solved, including the stronger solvers, are in Section 37.

```python
def tost_hit(d, margin, n_perm=20000, seed=0):
    d = np.asarray(d, float); rng = np.random.default_rng(seed); sg = rng.choice([-1.0, 1.0], (n_perm, len(d)))
    lo, hi = d + margin, d - margin
    p_lo = float(((sg * lo).mean(1) >= lo.mean()).mean()); p_hi = float(((sg * hi).mean(1) <= hi.mean()).mean())
    m = d[rng.integers(0, len(d), (5000, len(d)))].mean(1); c_lo, c_hi = np.percentile(m, 5), np.percentile(m, 95)
    return {"mean_diff": float(d.mean()), "ci90_lo": float(c_lo), "ci90_hi": float(c_hi), "P_tost": max(p_lo, p_hi),
            "equivalent_ci": bool(c_lo > -margin and c_hi < margin), "equivalent_perm": bool(max(p_lo, p_hi) < 0.05)}

def _run_shared():
    rows = []; t0 = time.time()
    for fam, gen in FAMILIES.items():
        arms = {k: v for k, v in arms_family(fam).items() if k in ("QAOA p=2 transferred", "SDP", "SA tuned (ext)", "Tabu", "Local search", "Random", "QAOA p=2")}
        for s in range(N_SH):
            traj = exact_trajectory(N, s, K_MAIN, gen=gen)
            for a, f in arms.items():
                df = score_on_trajectory(traj, f, s); df["family"], df["seed"], df["arm"] = fam, s, a; rows.append(df)
        print(f"  [H35] {fam} done ({time.time() - t0:.0f}s)")
    return {"iters": pd.concat(rows, ignore_index=True)}
SH = cached("H35_shared_all", _run_shared); dSHi = SH["iters"]
dSH = dSHi.groupby(["family", "seed", "arm"])[["gap_unif", "hit"]].mean().reset_index()
print(dSH.groupby(["family", "arm"])[["gap_unif", "hit"]].mean().round(4).unstack(0).to_string())
EQH = []
for fam in FAMILIES:
    s_ = dSH[dSH.family == fam]; si = dSHi[dSHi.family == fam]
    ref_ = si[si.arm == "QAOA p=2 transferred"].sort_values(["seed", "iter"]).hit.values.astype(bool)
    for a in ("SDP", "SA tuned (ext)", "Tabu", "Local search", "Random", "QAOA p=2"):
        oth = si[si.arm == a].sort_values(["seed", "iter"]).hit.values.astype(bool); mc = mcnemar_exact(ref_, oth)
        t = tost_hit(col(s_, "QAOA p=2 transferred", "hit") - col(s_, a, "hit"), TOST_MARGIN_HIT)
        EQH.append({"family": fam, "comparison": f"transferred vs {a}", "transferred_only": mc["a_only"], "other_only": mc["b_only"],
                    "P_mcnemar": mc["P"], **t})
    show_tests([paired(col(s_, "QAOA p=2 transferred", "gap_unif"), col(s_, a, "gap_unif"), f"H35 {fam} transferred vs {a}", f"H35 shared {fam}", "confirmatory")
                for a in ("SDP", "SA tuned (ext)", "Tabu", "Local search", "Random", "QAOA p=2")])
for a in ("SA tuned", "Tabu", "Local search", "SDP"):
    t = tost_hit(col(dR, "QAOA p=2 transferred", "hit") - col(dR, a, "hit"), TOST_MARGIN_HIT)
    EQH.append({"family": "main (embedded, R25)", "comparison": f"transferred vs {a}", "transferred_only": np.nan, "other_only": np.nan, "P_mcnemar": np.nan, **t})
for a, lab in (("QAOA p=2 transferred from n=6", "n=6 angles"), ("QAOA p=2 transferred from n=8", "n=8 angles")):
    if a in set(dAb.arm):
        dd_ = pd.concat([dAb, dR[dR.arm == "SA tuned"]], ignore_index=True)
        EQ_ = equivalence(col(dd_, a, "gap_unif"), col(dd_, "SA tuned", "gap_unif"), EQ_MARGIN)
        print(f"Equivalence in g_unif, {lab} vs tuned SA (margin {EQ_MARGIN}): {EQ_}")
dEQH = pd.DataFrame(EQH); dEQH.to_csv("H35_success_tests.csv", index=False)
print("\n", dEQH.round(4).to_string(index=False))
```

## 36. Stronger classical solvers

Three solvers that are standard for rugged QUBO landscapes, under the same evaluation accounting as the other solvers:
- **Tabu search with restarts.** The tabu search of Section 6, restarted from a uniformly random bitstring when its best value has not improved for a given number of iterations.
- **Parallel tempering (PT).** $M$ replicas at geometrically spaced temperatures run single-flip Metropolis moves, with neighbour exchanges after every sweep. Each proposal counts as one evaluation; exchanges reuse known energies and cost nothing.
- **SDP rounding with local search.** Each hyperplane rounding of the SDP relaxation (one evaluation) is followed by steepest descent (each scan of the $n$ neighbours counts as $n$ evaluations), until the budget is spent. As for the SDP, the relaxation is solved once per subproblem and its time is reported, not budgeted.

Tabu tenure and restart length, and the number of PT replicas and its temperature range, are tuned for each family on the same 36 held-out subproblems as SA and the angles (three repetitions per configuration); ties go to the first configuration in grid order. Because the extended SA grid of Section 32 selected an edge configuration on the planted family, SA is also tuned there on a wider grid (initial temperature 0.1 to 1, final temperature 0.0005 to 0.02, 8 to 128 restarts); if it selects a different configuration, that configuration is added as the arm *SA tuned (wide)*. The output reports how many configurations tie at the minimum and whether a uniquely selected value lies on the edge of its grid.

```python
from scipy.stats import norm

def run_tabu_restarts(Qp, h, budget, seed=0, tenure=3, stall=24):
    n = len(h); rng = np.random.default_rng(seed); dQ = np.diag(Qp).copy()
    z = rng.integers(0, 2, size=n).astype(float); cur = qubo_cost(z, Qp, h); used = 1
    bz, bc = z.copy(), cur; rb = cur; idle = 0; tabu = np.zeros(n, dtype=int); it = 0
    while used + n <= budget:
        if idle >= stall:
            z = rng.integers(0, 2, size=n).astype(float); cur = qubo_cost(z, Qp, h); used += 1
            rb = cur; idle = 0; tabu[:] = 0
            if cur < bc - 1e-12: bz, bc = z.copy(), cur
            if used + n > budget: break
        d = _flip_deltas(Qp, h, z, dQ); used += n; it += 1
        ok = (tabu <= it) | (cur + d < bc - 1e-12)
        if not ok.any(): ok[:] = True
        i = int(np.argmin(np.where(ok, d, np.inf)))
        z[i] = 1.0 - z[i]; cur += d[i]; tabu[i] = it + tenure
        if cur < rb - 1e-12: rb, idle = cur, 0
        else: idle += 1
        if cur < bc - 1e-12: bz, bc = z.copy(), cur
    return {"best": qubo_cost(bz, Qp, h), "z": bz, "evals": used}

def _pt_temps(M, Tmin, Tmax): return Tmin * (Tmax / Tmin) ** (np.arange(M) / max(M - 1, 1))

def run_pt(Qp, h, budget, seed=0, M=8, Tmin=0.05, Tmax=2.0):
    n = len(h); rng = np.random.default_rng(seed); Qs = (Qp + Qp.T) / 2; dQ = np.diag(Qs).copy()
    T = _pt_temps(M, Tmin, Tmax); ar = np.arange(M)
    Z = rng.integers(0, 2, (M, n)).astype(float); G = Z @ Qs; E = np.einsum("ij,ij->i", G, Z) + Z @ h; used = M
    k = int(np.argmin(E)); bz, bc = Z[k].copy(), float(E[k]); step = 0
    while used + M <= budget:
        i = rng.integers(0, n, M); sgn = 1.0 - 2.0 * Z[ar, i]; d = sgn * (2.0 * G[ar, i] + h[i]) + dQ[i]; used += M; step += 1
        acc = (d <= 0) | (rng.random(M) < np.exp(-np.clip(d, 0, None) / T))
        if acc.any():
            r_, i_ = ar[acc], i[acc]; Z[r_, i_] = 1.0 - Z[r_, i_]; G[r_] += sgn[acc][:, None] * Qs[i_]; E[acc] += d[acc]
            k = int(np.argmin(E))
            if E[k] < bc - 1e-12: bz, bc = Z[k].copy(), float(E[k])
        if step % n == 0:
            for j in range(step // n % 2, M - 1, 2):
                if rng.random() < np.exp(min(0.0, (1 / T[j] - 1 / T[j + 1]) * (E[j] - E[j + 1]))):
                    Z[[j, j + 1]] = Z[[j + 1, j]]; G[[j, j + 1]] = G[[j + 1, j]]; E[[j, j + 1]] = E[[j + 1, j]]
    return {"best": qubo_cost(bz, Qp, h), "z": bz, "evals": used}

def sdp_factor(Qp, h):
    # vectors of the general-QUBO SDP relaxation (auxiliary spin first); hyperplane roundings are sign(V r)
    n = len(h); cs, css, off = qubo_to_ising_coeffs(Qp, h); m = n + 1
    J = np.zeros((m, m)); J[0, 1:] = cs; J[1:, 0] = cs; J[1:, 1:] = css + css.T
    t0 = time.perf_counter(); Y = cp.Variable((m, m), PSD=True)
    cp.Problem(cp.Maximize(cp.sum(cp.multiply(np.triu(J, 1), (1 - Y))) / 2), [cp.diag(Y) == 1]).solve(solver=cp.SCS, verbose=False)
    Ym = (Y.value + Y.value.T) / 2; w, V = np.linalg.eigh(Ym)
    return V @ np.diag(np.sqrt(np.clip(w, 0, None))), time.perf_counter() - t0

def sdp_roundings(V, k, rng):
    S = np.sign(V @ rng.normal(size=(V.shape[1], int(k)))); S[S == 0] = 1.0
    S = S * np.where(S[0] < 0, -1.0, 1.0)[None, :]
    return ((1 - S[1:]) / 2).T

def run_sdp_ls(Qp, h, budget, seed=0, V=None):
    n = len(h); rng = np.random.default_rng(seed); dQ = np.diag(Qp).copy(); t_ = 0.0
    if V is None: V, t_ = sdp_factor(Qp, h)
    bz, bc, used = None, np.inf, 0
    while used + 1 <= budget:
        z = sdp_roundings(V, 1, rng)[0]; cur = qubo_cost(z, Qp, h); used += 1
        if cur < bc - 1e-12: bz, bc = z.copy(), cur
        while used + n <= budget:
            d = _flip_deltas(Qp, h, z, dQ); used += n; i = int(np.argmin(d))
            if d[i] >= -1e-12: break
            z[i] = 1.0 - z[i]; cur += d[i]
            if cur < bc - 1e-12: bz, bc = z.copy(), cur
    return {"best": qubo_cost(bz, Qp, h), "z": bz, "evals": used, "sdp_time": t_}

# a single-subproblem check against exhaustive enumeration
_Qp, _h = first_zupdate(10, 0); _st = exact_stats(_Qp, _h, keep=True)
for _f in (lambda: run_tabu_restarts(_Qp, _h, 768, 1), lambda: run_pt(_Qp, _h, 768, 1), lambda: run_sdp_ls(_Qp, _h, 768, 1)):
    _o = _f(); assert _o["evals"] <= 768 and abs(qubo_cost(_o["z"], _Qp, _h) - _o["best"]) < 1e-9 and _o["best"] >= _st["Cstar"] - 1e-9
print("Tabu with restarts, parallel tempering, and SDP rounding with local search ready.")
```

```python
def _tune(train_raw, grid, runner, tag, keys, edges):
    # configurations ranked by held-out mean g_unif; ties go to the first configuration in grid order
    global RETUNED
    path = f"E36_tuning_{tag}.csv"
    d = pd.read_csv(path) if RESUME and os.path.exists(path) else None
    if d is not None and len(d) == len(grid) and all(((d[list(g)] == pd.Series(g)).all(axis=1)).any() for g in grid):
        print(f"[resume] loaded {path}")
    else:
        rows = []
        for cfg in grid:
            g = [gaps(runner(Qp, h, BUDGET, 7000 + 10 * j + rep, **cfg)["best"], st)["gap_unif"]
                 for j, (Qp, h, st) in enumerate(train_raw) for rep in range(TUNE_REPS)]
            rows.append({**cfg, "gap_unif": float(np.mean(g))})
        d = pd.DataFrame(rows); d.to_csv(path, index=False); RETUNED = True
    b = d.loc[d.gap_unif.idxmin()]; cfg = {k: (int(b[k]) if k in ("tenure", "stall", "M", "restarts") else float(b[k])) for k in keys}
    edge = [k for k in keys if cfg[k] in edges[k]]; ties = int((d.gap_unif <= b.gap_unif + 1e-12).sum())
    print(f"{tag}: {cfg}; held-out mean g_unif {b.gap_unif:.5f}; configurations tied at the minimum: {ties}/{len(d)}; "
          f"on grid edge: {'n/a (tied)' if ties > 1 else (edge or 'no')}")
    return cfg

TABU_GRID = [{"tenure": t, "stall": s} for t in TABU_TENURES for s in TABU_STALLS]
PT_GRID = [{"M": m, "Tmin": a, "Tmax": b} for m in PT_REPLICAS for a in PT_TMIN for b in PT_TMAX if a < b]
TABU_R, PT_CFG, RETUNED = {}, {}, False
for fam, tr in (("main", TRAIN_MAIN), ("Wishart", TRAIN_W)):
    TABU_R[fam] = _tune(tr, TABU_GRID, run_tabu_restarts, f"tabu_{fam}", ("tenure", "stall"),
                        {"tenure": (min(TABU_TENURES), max(TABU_TENURES)), "stall": (min(TABU_STALLS), max(TABU_STALLS))})
    PT_CFG[fam] = _tune(tr, PT_GRID, run_pt, f"pt_{fam}", ("M", "Tmin", "Tmax"),
                        {"M": (min(PT_REPLICAS), max(PT_REPLICAS)), "Tmin": (min(PT_TMIN), max(PT_TMIN)), "Tmax": (min(PT_TMAX), max(PT_TMAX))})

# SA on a wider grid for the planted family: more restarts and lower final temperatures than the extended grid of Section 32
SAW_GRID = [{"T0": a, "Tmin": b, "restarts": r} for a in SAW_T0 for b in SAW_TMIN for r in SAW_RESTARTS if b < a]
SA_WIDE = _tune(TRAIN_W, SAW_GRID, run_sa_restarts, "sa_wide_Wishart", ("T0", "Tmin", "restarts"),
                {"T0": (min(SAW_T0), max(SAW_T0)), "Tmin": (min(SAW_TMIN), max(SAW_TMIN)), "restarts": (min(SAW_RESTARTS), max(SAW_RESTARTS))})
print("Wide-grid SA configuration " + ("equals the extended-grid configuration of Section 32; the planted-family SA arm is unchanged."
      if SA_WIDE == SA_EXT["Wishart"] else f"differs from the extended-grid configuration {SA_EXT['Wishart']}; it is added as 'SA tuned (wide)'."))

# cached results that depend on the selected configurations are recomputed when the selection changes
CFG_FILE, cur_cfg = "E36_selected_configs.json", {"tabu": TABU_R, "pt": PT_CFG, "sa_wide": SA_WIDE, "tts_budgets": list(TTSX_BUDGETS)}
prev_cfg = json.load(open(CFG_FILE)) if os.path.exists(CFG_FILE) else None
if (prev_cfg is not None and prev_cfg != cur_cfg) or (prev_cfg is None and RETUNED):
    for name in ("E37_embedded", "E37_shared", "E37_direct", "E39_tts_selection", "E39_tts", "E40_hardware_q", "E41_hardware", "E43_general",
                 "E43_general_planted"):
        if os.path.exists(f"{name}__keys.json"): os.remove(f"{name}__keys.json")
    print("Selected configurations changed; dependent results will be recomputed.")
json.dump(cur_cfg, open(CFG_FILE, "w"))

def mk_tabu_r(B, fam): return lambda Qp, h, st, s, ctx: run_tabu_restarts(Qp, h, B, s, **TABU_R[fam])
def mk_pt(B, fam):     return lambda Qp, h, st, s, ctx: run_pt(Qp, h, B, s, **PT_CFG[fam])
def mk_sdp_ls(B):      return lambda Qp, h, st, s, ctx: run_sdp_ls(Qp, h, B, s)
STRONG = ("Tabu (restarts)", "PT", "SDP + LS", "SA tuned (wide)")
def strong_arms(fam, B=BUDGET):
    arms = {"Tabu (restarts)": mk_tabu_r(B, fam), "PT": mk_pt(B, fam), "SDP + LS": mk_sdp_ls(B)}
    if fam == "Wishart" and SA_WIDE != SA_EXT["Wishart"]: arms["SA tuned (wide)"] = mk_sa_cfg(B, SA_WIDE)
    return arms
```

## 37. Stronger solvers in ADMM, on the shared trajectory, and on the reduced problem

The three solvers are added to the comparisons of Sections 33–35 on the same instances: embedded in ADMM ($n=12$, $K=12$, $\rho\in\{2,1\}$, 80 instances per family), on the identical subproblems of the exact-solver trajectory, and applied directly to the reduced QUBO $F$ with the budget $K\times768$.

The subproblems of one instance are dependent, because consecutive ADMM subproblems differ only through the anchor of the proximal term. All inference on the fraction of subproblems solved exactly is therefore at the instance level:
- the exact signed-rank test on per-instance fractions;
- the cluster-adjusted McNemar statistic of Obuchowski (1998), which treats instances as clusters, with its $P$-value from the sign-flip distribution over instances (exact when at most 20 instances are discordant);
- two one-sided tests (bootstrap 90% interval and sign-flip randomization) with margins of 2 and 1 percentage points.

The pooled McNemar counts are reported for description only.

```python
def _run_strong_embedded():
    its, sms = [], []
    for fam, gen in FAMILIES.items():
        for r_ in RHOS_W:
            out = run_arms_embedded(strong_arms(fam), list(range(N_W)), N, K_MAIN, f"E37 {fam} rho={r_}", gen=gen, rho=r_)
            for d in (out["iters"], out["seeds"]): d["rho"] = r_; d["family"] = fam
            its.append(out["iters"]); sms.append(out["seeds"])
    return {"iters": pd.concat(its, ignore_index=True), "seeds": pd.concat(sms, ignore_index=True)}
E37E = cached("E37_embedded", _run_strong_embedded)
dFAMX = pd.concat([dFAM, E37E["seeds"]], ignore_index=True)

def _run_strong_shared():
    rows = []
    for fam, gen in FAMILIES.items():
        for s in range(N_SH):
            traj = exact_trajectory(N, s, K_MAIN, gen=gen)
            for a, f in strong_arms(fam).items():
                df = score_on_trajectory(traj, f, s); df["family"], df["seed"], df["arm"] = fam, s, a; rows.append(df)
    return {"iters": pd.concat(rows, ignore_index=True)}
E37S = cached("E37_shared", _run_strong_shared)
dSHiX = pd.concat([dSHi, E37S["iters"]], ignore_index=True)
dSHX = dSHiX.groupby(["family", "seed", "arm"])[["gap_unif", "hit"]].mean().reset_index()

def _run_strong_direct():
    rows = []
    for fam, gen in FAMILIES.items():
        for s in range(N_DQ):
            QF, hF, _ = reduced_qubo(N, s, gen); st = exact_stats(QF, hF, keep=True)
            for a, f in strong_arms(fam, DQ_BUDGET).items():
                g = gaps(f(QF, hF, st, s, {})["best"], st)
                rows.append({"family": fam, "seed": s, "arm": a, "mbo_opt_found": g["hit"], "mbo_gap_best_unif": g["gap_unif"]})
    return {"seeds": pd.DataFrame(rows)}
E37D = cached("E37_direct", _run_strong_direct)
dDQX = pd.concat([dDQ, E37D["seeds"]], ignore_index=True)

def col2(ds, a, b, metric):
    # values of two arms on their common instances
    x = ds[ds.arm == a].set_index("seed")[metric]; y = ds[ds.arm == b].set_index("seed")[metric]; ix = x.index.intersection(y.index).sort_values()
    return x[ix].values, y[ix].values

def mcnemar_clustered(a, b, cluster, n_mc=200000, seed=0):
    # instances are clusters, subproblems are units. Obuchowski (1998) z statistic, with its P-value from the
    # sign-flip distribution of the per-instance discordance (exact when at most 20 instances are discordant,
    # otherwise Monte Carlo with the add-one estimate)
    a, b, cl = np.asarray(a, float), np.asarray(b, float), np.asarray(cluster)
    ids = np.unique(cl); K_ = len(ids); N_ = len(a); d = np.array([a[cl == i].sum() - b[cl == i].sum() for i in ids])
    m = np.array([(cl == i).sum() for i in ids]); diff = d.sum() / N_
    var = K_ / (K_ - 1) / N_ ** 2 * float(np.sum((d - m * diff) ** 2)); zst = diff / np.sqrt(var) if var > 0 else 0.0
    nz = d[d != 0]
    if len(nz) == 0: return {"diff": float(diff), "z": 0.0, "P": 1.0, "discordant_instances": 0}
    if len(nz) <= 20:
        signs = 1 - 2 * ((np.arange(2 ** len(nz))[:, None] >> np.arange(len(nz))) & 1)
    else:
        signs = np.random.default_rng(seed).choice([-1, 1], (n_mc, len(nz)))
    ext = int(np.sum(np.abs(signs @ nz) >= abs(nz.sum()) - 1e-12))
    P = ext / len(signs) if len(nz) <= 20 else (ext + 1) / (n_mc + 1)
    return {"diff": float(diff), "z": float(zst), "P": P, "discordant_instances": int(len(nz))}

REF_ARM = "QAOA p=2 transferred"
SUCC = []
for fam in FAMILIES:
    piv = dSHiX[dSHiX.family == fam].pivot_table(index=["seed", "iter"], columns="arm", values="hit")
    for a in [x for x in ("SDP", "SDP + LS", "PT", "Tabu (restarts)", "SA tuned (wide)", "SA tuned (ext)", "Local search", "Tabu", "QAOA p=2", "Random") if x in piv]:
        sub = piv[[REF_ARM, a]].dropna(); seeds_ = sub.index.get_level_values("seed").values
        pooled = mcnemar_exact(sub[REF_ARM].values.astype(bool), sub[a].values.astype(bool))
        mc = mcnemar_clustered(sub[REF_ARM].values, sub[a].values, seeds_)
        register(f"E37 {fam} clustered McNemar transferred vs {a}", f"E37 success {fam}", "confirmatory", len(sub), pooled["a_only"] + pooled["b_only"], mc["P"],
                 float(np.sign(-mc["diff"])))
        inst = sub.groupby(level="seed").mean()
        t = paired(1 - inst[REF_ARM].values, 1 - inst[a].values, f"E37 {fam} per-instance Pr[opt] transferred vs {a}", f"E37 success {fam}", "confirmatory")
        dh = inst[REF_ARM].values - inst[a].values
        SUCC.append({"family": fam, "comparison": f"transferred vs {a}", "pr_opt_transferred": float(inst[REF_ARM].mean()),
                     "pr_opt_other": float(inst[a].mean()), "P_signed_rank": t["P"], "nontied_instances": t["n_nonzero"],
                     "P_clustered_mcnemar": mc["P"], "discordant_instances": mc["discordant_instances"], "pooled_transferred_only": pooled["a_only"], "pooled_other_only": pooled["b_only"],
                     **{f"{k}_2pp": v for k, v in tost_hit(dh, 0.02).items()}, **{f"{k}_1pp": v for k, v in tost_hit(dh, 0.01).items() if k.startswith("equivalent")}})
dSUCC = pd.DataFrame(SUCC); dSUCC.to_csv("E37_success_tests.csv", index=False)
print("Fraction of subproblems solved exactly on the shared trajectory (instance means):")
print(dSHX.groupby(["family", "arm"]).hit.mean().unstack(0).round(4).to_string())
print("\n", dSUCC[["family", "comparison", "pr_opt_transferred", "pr_opt_other", "P_signed_rank", "nontied_instances", "P_clustered_mcnemar", "discordant_instances",
                 "pooled_transferred_only", "pooled_other_only", "mean_diff_2pp", "ci90_lo_2pp", "ci90_hi_2pp", "equivalent_ci_2pp", "equivalent_perm_2pp",
                 "equivalent_ci_1pp", "equivalent_perm_1pp"]].round(4).to_string(index=False))
conc = dSHiX[(dSHiX.family == "Wishart")].pivot_table(index=["seed", "iter"], columns="arm", values="hit")
won = conc[(conc[REF_ARM] == 1) & (conc["SDP"] == 0)].reset_index().groupby("seed").size().sort_values(ascending=False)
print(f"\nPlanted family, subproblems solved by transferred QAOA but not the SDP: {int(won.sum())}, from {len(won)} instances; "
      f"the two largest instances contribute {int(won.head(2).sum())}.")

_ciX = [{"family": f_, "rho": r_, "arm": a_, **dict(zip(("hit", "hit_lo", "hit_hi"), bootstrap_ci(s_.hit))), "gap_unif_x1e3": 1e3 * s_.gap_unif.mean(),
         "mbo_opt_found": s_.mbo_opt_found.mean()} for (f_, r_, a_), s_ in dFAMX.groupby(["family", "rho", "arm"])]
dFAMCIX = pd.DataFrame(_ciX); dFAMCIX.to_csv("E37_embedded_summary.csv", index=False)
print("\nADMM-embedded: fraction of subproblems solved exactly [95% CI], mean g_unif x 1e3, problem optimum found")
print(dFAMCIX[dFAMCIX.arm.isin(STRONG + (REF_ARM, "SDP", "SA tuned (ext)", "Tabu", "Local search"))].round(3).to_string(index=False))
for fam in FAMILIES:
    for r_ in RHOS_W:
        s_ = dFAMX[(dFAMX.family == fam) & (dFAMX.rho == r_)]
        show_tests([paired(*col2(s_, REF_ARM, a, "gap_unif"), f"E37 {fam} rho={r_:g} transferred vs {a}", f"E37 embedded {fam} rho={r_:g}", "confirmatory")
                    for a in strong_arms(fam)])
        show_tests([paired(*col2(s_, REF_ARM, a, "mbo_gap_best_unif"), f"E37 {fam} rho={r_:g} transferred vs {a} (ADMM level)",
                           f"E37 ADMM level {fam} rho={r_:g}", "exploratory") for a in strong_arms(fam)])

refX = []
for (fam, r_, a), v in dFAMX.groupby(["family", "rho", "arm"])["mbo_opt_found"].mean().items():
    refX.append({"family": fam, "setting": f"ADMM rho={r_:g}", "arm": a, "mbo_opt_found": v})
for (fam, a), v in dDQX.groupby(["family", "arm"])["mbo_opt_found"].mean().items():
    refX.append({"family": fam, "setting": "direct QUBO", "arm": a, "mbo_opt_found": v})
dREFX = pd.DataFrame(refX); dREFX.to_csv("E37_problem_level_summary.csv", index=False)
print("\nProblem optimum found: ADMM at rho = 1 and 2, and direct solution of F with K x 768 evaluations")
print(dREFX.pivot_table(index=["family", "arm"], columns="setting", values="mbo_opt_found").round(3).to_string())
for fam in FAMILIES:
    s_ = dDQX[dDQX.family == fam]
    show_tests([paired(*col2(s_, REF_ARM, a, "mbo_gap_best_unif"), f"E37 {fam} direct: transferred vs {a}",
                       f"E37 direct {fam}", "exploratory") for a in strong_arms(fam)])
```

## 38. Angles trained at $n=16$

To test whether the time-to-solution exponent of transferred QAOA reflects angles trained at a single size, $p=2$ angles are retrained for each family on held-out instances at $n=16$, starting from the $n=12$ angles and two random points.

```python
def train_angles_at(gen, n_, x_init, p=2):
    train = []
    for s in range(TRAIN_OFFSET, TRAIN_OFFSET + N_TRAIN16):
        for (Qp, h, st) in exact_trajectory(n_, s, K=6, gen=gen)[::2]:
            train.append(_prep(Qp, h, st))
    rng = np.random.default_rng(16); best = (np.inf, None)
    for x0 in [np.asarray(x_init)] + [rng.uniform(-np.pi / 4, np.pi / 4, 2 * p) for _ in range(TRAIN16_RESTARTS - 1)]:
        r = minimize(transfer_objective, x0, args=(p, train), method="Nelder-Mead", options={"maxfev": TRAIN16_MAXFEV, "xatol": 1e-4, "fatol": 1e-6})
        if r.fun < best[0]: best = (r.fun, r.x)
    return best[1], best[0], transfer_objective(np.asarray(x_init), p, train)

ANG16_FILE = "E38_angles_n16.json"
if RESUME and os.path.exists(ANG16_FILE):
    A16 = json.load(open(ANG16_FILE)); print("[resume] loaded n=16 angles")
else:
    A16 = {}
    for fam, gen in FAMILIES.items():
        a12 = TRANSFER_W[2] if fam == "Wishart" else TRANSFER[("admm", 2)]
        x_, f_, f12 = train_angles_at(gen, TRAIN16_N, a12); A16[fam] = {"angles": x_.tolist(), "train_gap": f_, "train_gap_n12_angles": f12}
    json.dump(A16, open(ANG16_FILE, "w"))
ANG16 = {fam: np.array(v["angles"]) for fam, v in A16.items()}
for fam, v in A16.items():
    print(f"{fam}: n=16 angles {np.round(v['angles'], 3)}; held-out mean expected g_unif {v['train_gap']:.3f} (n=12 angles: {v['train_gap_n12_angles']:.3f})")
```

## 39. Time to solution

$\mathrm{TTS}_{99}$ is the number of evaluations needed to find the optimum with probability 0.99. It is measured on the second ADMM subproblem, for $n$ from 12 to 20, on 20 test instances per size and family.
- **Samplers with a fixed per-sample probability $q$:** $\mathrm{TTS}_{99}=\ln 0.01/\ln(1-q)$. Transferred QAOA (angles trained at $n=12$ and at $n=16$) and uniform sampling use the exact $q$; SDP hyperplane rounding uses $q$ estimated from 4096 roundings of one relaxation, whose solve is not counted.
- **Solvers with a run length:** SA (one anneal with the tuned temperatures), tabu search with restarts, local search with restarts, PT, and SDP rounding with local search. For each solver and size, one budget $b^\star$ per run is chosen on 20 *held-out* instances (256 runs at each budget of the grid), as the budget that minimizes the median $\mathrm{TTS}_{99}$. Each test instance is then scored at $b^\star$ from 1024 independent runs: $\mathrm{TTS}_{99}=b^\star\ln 0.01/\ln(1-\hat p)$.
- **Censoring:** an instance with no successful run (or rounding) has $\mathrm{TTS}_{99}=\infty$ and is censored; no probability is imputed. Sizes are summarized by the median over instances.
- **Scaling:** the exponent is the least-squares slope of $\log_2$ of the median against $n$, with a 95% bootstrap interval over instances; a resampled median that is censored gives an infinite slope.
- **Paired tests** on $\log_2\mathrm{TTS}_{99}$ are exact signed-rank tests; censored values are ranked above every finite value, and pairs in which both are censored are dropped as ties.

```python
def _sym(Qp): return (Qp + Qp.T) / 2

def sa_batch(Qp, h, budget, R, rng, T0=2.0, Tmin=0.01):
    n = len(h); Qs = _sym(Qp); dQ = np.diag(Qs).copy(); ar = np.arange(R)
    Z = rng.integers(0, 2, (R, n)).astype(float); G = Z @ Qs; cur = np.einsum("ij,ij->i", G, Z) + Z @ h; best = cur.copy()
    steps = max(int(budget), 1)
    for t in range(steps):
        T = T0 * ((Tmin / T0) ** (t / max(steps - 1, 1))); i = rng.integers(0, n, R)
        sgn = 1.0 - 2.0 * Z[ar, i]; d = sgn * (2.0 * G[ar, i] + h[i]) + dQ[i]
        acc = (d <= 0) | (rng.random(R) < np.exp(-np.clip(d, 0, None) / max(T, 1e-9)))
        if acc.any():
            r_, i_ = ar[acc], i[acc]; Z[r_, i_] = 1.0 - Z[r_, i_]; G[r_] += sgn[acc][:, None] * Qs[i_]; cur[acc] += d[acc]; best = np.minimum(best, cur)
    return best

def tabu_r_batch(Qp, h, budget, R, rng, tenure=3, stall=24):
    n = len(h); Qs = _sym(Qp); dQ = np.diag(Qs).copy(); ar = np.arange(R)
    Z = rng.integers(0, 2, (R, n)).astype(float); cur = np.einsum("ij,jk,ik->i", Z, Qs, Z) + Z @ h; best = cur.copy(); rb = cur.copy()
    used = np.ones(R); idle = np.zeros(R, dtype=int); tabu = np.zeros((R, n), dtype=int); it = 0
    while True:
        rs = (idle >= stall) & (used + 1 + n <= budget)
        if rs.any():
            k = int(rs.sum()); Z[rs] = rng.integers(0, 2, (k, n)).astype(float)
            cur[rs] = np.einsum("ij,jk,ik->i", Z[rs], Qs, Z[rs]) + Z[rs] @ h; used[rs] += 1; rb[rs] = cur[rs]; idle[rs] = 0; tabu[rs] = 0
            best = np.minimum(best, cur)
        act = used + n <= budget
        if not act.any(): break
        it += 1; D = (1.0 - 2.0 * Z) * (2.0 * (Z @ Qs) + h) + dQ
        ok = (tabu <= it) | (cur[:, None] + D < best[:, None] - 1e-12); ok[~ok.any(1)] = True
        i = np.argmin(np.where(ok, D, np.inf), axis=1); a_ = ar[act]; i_ = i[act]
        Z[a_, i_] = 1.0 - Z[a_, i_]; cur[a_] += D[a_, i_]; tabu[a_, i_] = it + tenure; used[act] += n
        imp = act & (cur < rb - 1e-12); rb[imp] = cur[imp]; idle[imp] = 0; idle[act & ~imp] += 1
        best = np.minimum(best, cur)
    return best

def ls_batch(Qp, h, budget, R, rng, init=None):
    n = len(h); Qs = _sym(Qp); dQ = np.diag(Qs).copy(); ar = np.arange(R)
    draw = init or (lambda k: rng.integers(0, 2, (k, n)).astype(float))
    Z = draw(R); cur = np.einsum("ij,jk,ik->i", Z, Qs, Z) + Z @ h; best = cur.copy(); used = np.ones(R)
    while True:
        act = used + n <= budget
        if not act.any(): break
        D = (1.0 - 2.0 * Z) * (2.0 * (Z @ Qs) + h) + dQ; used[act] += n
        i = np.argmin(D, axis=1); dm = D[ar, i]; imp = act & (dm < -1e-12)
        Z[ar[imp], i[imp]] = 1.0 - Z[ar[imp], i[imp]]; cur[imp] += dm[imp]
        rs = act & ~imp & (used + 1 <= budget)
        if rs.any():
            Z[rs] = draw(int(rs.sum())); cur[rs] = np.einsum("ij,jk,ik->i", Z[rs], Qs, Z[rs]) + Z[rs] @ h; used[rs] += 1
        best = np.minimum(best, cur)
    return best

def pt_batch(Qp, h, budget, R, rng, M=8, Tmin=0.05, Tmax=2.0):
    n = len(h); Qs = _sym(Qp); dQ = np.diag(Qs).copy(); T = _pt_temps(M, Tmin, Tmax)
    Z = rng.integers(0, 2, (R, M, n)).astype(float); G = Z @ Qs; E = np.einsum("rmi,rmi->rm", G, Z) + Z @ h
    best = E.min(1); steps = (int(budget) - M) // M; rr, mm = np.meshgrid(np.arange(R), np.arange(M), indexing="ij")
    for t in range(1, steps + 1):
        i = rng.integers(0, n, (R, M)); zi = Z[rr, mm, i]; sgn = 1.0 - 2.0 * zi; d = sgn * (2.0 * G[rr, mm, i] + h[i]) + dQ[i]
        acc = (d <= 0) | (rng.random((R, M)) < np.exp(-np.clip(d, 0, None) / T[None, :]))
        if acc.any():
            r_, m_, i_ = rr[acc], mm[acc], i[acc]; Z[r_, m_, i_] = 1.0 - Z[r_, m_, i_]; G[r_, m_] += sgn[acc][:, None] * Qs[i_]; E[acc] += d[acc]
            best = np.minimum(best, E.min(1))
        if t % n == 0:
            for j in range(t // n % 2, M - 1, 2):
                sw = rng.random(R) < np.exp(np.minimum(0.0, (1 / T[j] - 1 / T[j + 1]) * (E[:, j] - E[:, j + 1])))
                if sw.any():
                    Z[sw, j], Z[sw, j + 1] = Z[sw, j + 1].copy(), Z[sw, j].copy(); G[sw, j], G[sw, j + 1] = G[sw, j + 1].copy(), G[sw, j].copy()
                    E[sw, j], E[sw, j + 1] = E[sw, j + 1].copy(), E[sw, j].copy()
    return best

LN01 = np.log(0.01)
def _tag(name): return int(hashlib.md5(name.encode()).hexdigest()[:8], 16)
def tts_q(q): return np.inf if q <= 0 else (1.0 if q >= 1 else max(1.0, LN01 / np.log1p(-q)))
def tts_b(p, b): return np.inf if p <= 0 else (float(b) if p >= 1 else b * LN01 / np.log1p(-p))

def tts_runners(fam, Qp, h, V):
    n = len(h); cfg = SA_WIDE if fam == "Wishart" else SA_EXT[fam]
    return {"SA (tuned temperatures)": lambda b, R, rng: sa_batch(Qp, h, b, R, rng, cfg["T0"], cfg["Tmin"]),
            "Tabu (restarts)": lambda b, R, rng: tabu_r_batch(Qp, h, b, R, rng, **TABU_R[fam]),
            "Local search": lambda b, R, rng: ls_batch(Qp, h, b, R, rng),
            "PT": lambda b, R, rng: pt_batch(Qp, h, b, R, rng, **PT_CFG[fam]),
            "SDP + LS": lambda b, R, rng: ls_batch(Qp, h, b, R, rng, init=lambda k: sdp_roundings(V, k, rng))}

def _run_tts_select():
    rows = []; t0 = time.time()
    for fam, gen in FAMILIES.items():
        for n_ in TTS_SIZES:
            for s in TTS_SEL_SEEDS:
                Qp, h, st = subproblem_k1_gen(n_, s, gen); Cs = st["Cstar"]; tol = 1e-7 * max(1.0, abs(Cs)); V, _ = sdp_factor(Qp, h)
                for a, fn in tts_runners(fam, Qp, h, V).items():
                    rng = np.random.default_rng([s, n_, _tag(a)])
                    for b in TTSX_BUDGETS:
                        if b < n_ + 1 and a != "SA (tuned temperatures)" and a != "PT": continue
                        rows.append({"family": fam, "n": n_, "seed": s, "arm": a, "budget": b, "p": float((fn(b, TTS_R_SEL, rng) - Cs <= tol).mean())})
                del st
            print(f"  [E39 selection] {fam} n={n_} ({time.time() - t0:.0f}s)")
    return {"runs": pd.DataFrame(rows)}
TTS_SEL = cached("E39_tts_selection", _run_tts_select)["runs"]
BSTARX = {}
for (fam, a, n_), s_ in TTS_SEL.groupby(["family", "arm", "n"]):
    med = s_.assign(t=[tts_b(p, b) for p, b in zip(s_.p, s_.budget)]).groupby("budget").t.median()
    BSTARX[(fam, a, n_)] = int(med.idxmin()) if np.isfinite(med.min()) else int(max(TTSX_BUDGETS))
pd.Series({f"{k[0]}|{k[1]}|{k[2]}": v for k, v in BSTARX.items()}).to_csv("E39_tts_budget_choice.csv")

def _run_tts():
    rows = []; t0 = time.time()
    for fam, gen in FAMILIES.items():
        ang = {"QAOA p=2 transferred": TRANSFER_W[2] if fam == "Wishart" else TRANSFER[("admm", 2)], "QAOA p=2 transferred (n=16 angles)": ANG16[fam]}
        for n_ in TTS_SIZES:
            for s in range(N_TTS):
                Qp, h, st = subproblem_k1_gen(n_, s, gen); Cv = st["C"]; Cs = st["Cstar"]; tol = 1e-7 * max(1.0, abs(Cs)); opt = Cv - Cs <= tol
                d_ = scaled_diagonal(Qp, h, Cv)
                for a, prm in ang.items():
                    pr = qaoa_probs_fast(d_, prm, 2, n_); q_ = float(pr[opt].sum() / pr.sum())
                    rows.append({"family": fam, "n": n_, "seed": s, "arm": a, "q": q_, "budget": 1, "p_hat": q_, "tts99": tts_q(q_)})
                q_ = float(opt.mean()); rows.append({"family": fam, "n": n_, "seed": s, "arm": "Random", "q": q_, "budget": 1, "p_hat": q_, "tts99": tts_q(q_)})
                V, _ = sdp_factor(Qp, h); rng = np.random.default_rng([s, n_, 99])
                Z = sdp_roundings(V, TTS_NROUND, rng); rc = np.einsum("ij,jk,ik->i", Z, Qp, Z) + Z @ h; q_ = float((rc - Cs <= tol).mean())
                rows.append({"family": fam, "n": n_, "seed": s, "arm": "SDP rounding", "q": q_, "budget": 1, "p_hat": q_, "tts99": tts_q(q_)})
                for a, fn in tts_runners(fam, Qp, h, V).items():
                    b = BSTARX[(fam, a, n_)]; rng = np.random.default_rng([s, n_, _tag(a), 7])
                    p_ = float((fn(b, TTS_R_TEST, rng) - Cs <= tol).mean())
                    rows.append({"family": fam, "n": n_, "seed": s, "arm": a, "q": np.nan, "budget": b, "p_hat": p_, "tts99": tts_b(p_, b)})
                del st, Cv, d_
            print(f"  [E39] {fam} n={n_} ({time.time() - t0:.0f}s)")
    return {"tts": pd.DataFrame(rows)}
dTTX = cached("E39_tts", _run_tts)["tts"]; dTTX["censored"] = ~np.isfinite(dTTX.tts99)

print("Median TTS99 over test instances (evaluations):")
print(dTTX.groupby(["family", "arm", "n"]).tts99.median().unstack().round(1).to_string())
print("\nCensored fraction:"); print(dTTX.groupby(["family", "arm", "n"]).censored.mean().unstack().round(2).to_string())
print("\nBudget per run chosen on held-out instances:")
print(pd.DataFrame([{"family": k[0], "arm": k[1], "n": k[2], "b": v} for k, v in BSTARX.items()]).pivot_table(index=["family", "arm"], columns="n", values="b").to_string())

SLX = []
for (fam, a), s_ in dTTX.groupby(["family", "arm"]):
    ns_ = np.array(sorted(s_.n.unique()), float); by_n = [s_[s_.n == n_].tts99.values for n_ in ns_]
    med = np.array([np.median(v) for v in by_n]); sl = np.polyfit(ns_, np.log2(med), 1)[0] if np.all(np.isfinite(med)) else np.inf
    rng_ = np.random.default_rng(0); bs = []
    for _ in range(2000):
        mb = np.array([np.median(v[rng_.integers(0, len(v), len(v))]) for v in by_n])
        bs.append(np.polyfit(ns_, np.log2(mb), 1)[0] if np.all(np.isfinite(mb)) else np.inf)
    bs = np.array(bs)
    SLX.append({"family": fam, "arm": a, "slope": sl, "lo": float(np.quantile(bs, 0.025, method="lower")), "hi": float(np.quantile(bs, 0.975, method="higher")),
                "median_n12": med[0], "median_n20": med[-1], "censored_fraction": float(s_.censored.mean())})
dSLX = pd.DataFrame(SLX); dSLX.to_csv("E39_tts_slopes.csv", index=False)
print("\nExponent s in TTS99 ~ 2^(s n), with 95% bootstrap interval:"); print(dSLX.round(3).to_string(index=False))

def _l2(v): v = np.asarray(v, float); return np.where(np.isfinite(v), np.log2(np.where(np.isfinite(v), v, 1.0)), 1e3)
for fam in FAMILIES:
    for n_ in TTS_SIZES:
        s_ = dTTX[(dTTX.family == fam) & (dTTX.n == n_)]
        for a in ("Tabu (restarts)", "PT", "Local search", "SA (tuned temperatures)", "SDP + LS", "SDP rounding"):
            paired(_l2(col(s_, REF_ARM, "tts99")), _l2(col(s_, a, "tts99")), f"E39 {fam} n={n_} log2 TTS transferred vs {a}", f"E39 TTS {fam}", "exploratory")
```

**Resolution, exponent differences, and wall-clock cost.**
- A run that succeeds every time has $\mathrm{TTS}_{99}=b^\star$, so its estimate is bounded above by the budget grid; the fraction of such instances is reported.
- Exponents are compared with a paired bootstrap: the same test instances are resampled for transferred QAOA and for each classical solver, and the difference of the fitted slopes is recorded. The two-sided $P$-value is twice the smaller tail fraction.
- Wall-clock cost: the time per evaluation of each classical solver is measured on this machine with its serial reference implementation (budget 4096, three test instances per size); the SDP relaxation time is measured separately and charged once per subproblem. For QAOA the time per shot is the device usage reported by the runtime service for the second $n=12$ submission (Section 41); circuits are deeper at larger $n$, so this is a lower bound there.

```python
ns_all = np.array(TTS_SIZES, float)
fl = dTTX[dTTX.budget > 1].assign(floored=lambda d: np.isclose(d.tts99, d.budget))
print("Fraction of test instances on which every run succeeded (TTS99 = budget per run):")
print(fl.groupby(["family", "arm", "n"]).floored.mean().unstack().round(2).to_string())

def _slope(meds): return np.polyfit(ns_all, np.log2(meds), 1)[0] if np.all(np.isfinite(meds)) else np.inf
SDX = []
for fam in FAMILIES:
    tab = {a: np.array([dTTX[(dTTX.family == fam) & (dTTX.arm == a) & (dTTX.n == n_)].sort_values("seed").tts99.values for n_ in TTS_SIZES])
           for a in dTTX[dTTX.family == fam].arm.unique()}
    for ref in ("QAOA p=2 transferred", "QAOA p=2 transferred (n=16 angles)"):
        for a in ("Tabu (restarts)", "PT", "Local search", "SA (tuned temperatures)", "SDP + LS"):
            X, Y = tab[ref], tab[a]; rng_ = np.random.default_rng(3); diffs = []
            for _ in range(4000):
                ix = rng_.integers(0, X.shape[1], X.shape)
                diffs.append(_slope(np.median(np.take_along_axis(X, ix, 1), 1)) - _slope(np.median(np.take_along_axis(Y, ix, 1), 1)))
            diffs = np.array(diffs); diffs = diffs[np.isfinite(diffs)]
            obs = _slope(np.median(X, 1)) - _slope(np.median(Y, 1))
            P_ = float(min(1.0, 2 * min(np.mean(diffs <= 0), np.mean(diffs >= 0)))) if len(diffs) else np.nan
            lab = f"E39 {fam} slope difference {'n=16 angles' if 'n=16' in ref else 'transferred'} minus {a}"
            register(lab, f"E39 slopes {fam}", "exploratory", X.shape[1], X.shape[1], max(P_, 1 / 4001), float(np.sign(obs)))
            SDX.append({"family": fam, "qaoa": ref, "classical": a, "difference": obs, "lo": float(np.quantile(diffs, 0.025)) if len(diffs) else np.nan,
                        "hi": float(np.quantile(diffs, 0.975)) if len(diffs) else np.nan, "P_bootstrap": P_, "finite_resamples": len(diffs)})
dSDX = pd.DataFrame(SDX); dSDX.to_csv("E39_slope_differences.csv", index=False)
print("\nExponent of transferred QAOA minus exponent of the classical solver (paired bootstrap over test instances):")
print(dSDX.round(3).to_string(index=False))
```

```python
def _per_eval(fn):
    t0 = time.perf_counter(); out = fn(); return (time.perf_counter() - t0) / max(out["evals"], 1)

TIME_RUNS = []
_time_cached = RESUME and os.path.exists("E39_time_per_evaluation.csv")
for fam, gen in ({} if _time_cached else FAMILIES).items():
    for n_ in TTS_SIZES:
        for s in range(3):
            Qp, h, st = subproblem_k1_gen(n_, s, gen); del st
            V, t_sdp = sdp_factor(Qp, h); cfg = SA_WIDE if fam == "Wishart" else SA_EXT[fam]; rng_ = np.random.default_rng(s)
            def _round():
                t0 = time.perf_counter(); Z = sdp_roundings(V, 4096, rng_); _ = np.einsum("ij,jk,ik->i", Z, Qp, Z) + Z @ h
                return {"evals": 4096, "t": time.perf_counter() - t0}
            r_ = _round()
            per = {"Tabu (restarts)": _per_eval(lambda: run_tabu_restarts(Qp, h, 4096, s, **TABU_R[fam])),
                   "PT": _per_eval(lambda: run_pt(Qp, h, 4096, s, **PT_CFG[fam])),
                   "Local search": _per_eval(lambda: run_local_search(Qp, h, 4096, s)),
                   "SA (tuned temperatures)": _per_eval(lambda: run_sa(Qp, h, 4096, s, cfg["T0"], cfg["Tmin"])),
                   "SDP + LS": _per_eval(lambda: run_sdp_ls(Qp, h, 4096, s, V=V)), "SDP rounding": r_["t"] / r_["evals"]}
            for a, t_ in per.items():
                TIME_RUNS.append({"family": fam, "n": n_, "seed": s, "arm": a, "sec_per_eval": t_, "sdp_sec": t_sdp if a.startswith("SDP") else 0.0})
if _time_cached:
    dTIME = pd.read_csv("E39_time_per_evaluation.csv"); print("[resume] loaded E39_time_per_evaluation.csv (measured timings)")
else:
    dTIME = pd.DataFrame(TIME_RUNS).groupby(["family", "arm", "n"])[["sec_per_eval", "sdp_sec"]].median().reset_index()
    dTIME.to_csv("E39_time_per_evaluation.csv", index=False)
print("Median seconds per evaluation (serial reference implementations on this machine):")
print(dTIME.pivot_table(index=["family", "arm"], columns="n", values="sec_per_eval").map(lambda v: f"{v:.2e}").to_string())
print("\nSDP relaxation, median seconds per subproblem:")
print(dTIME[dTIME.arm == "SDP + LS"].pivot_table(index="family", columns="n", values="sdp_sec").round(4).to_string())

_hwj = json.load(open("E41_hardware.json")) if os.path.exists("E41_hardware.json") else None
if _hwj:
    _u = {k: v for k, v in _hwj["usage"].items() if v}; SHOT_SEC = sum(_u.values()) / sum(_hwj["shots"][k] for k in _u)
else:
    SHOT_SEC = np.nan
print(f"\nDevice time per shot (second n=12 submission): {SHOT_SEC:.2e} s")

rows = []
for _, r in dTTX.iterrows():
    if r.arm.startswith("QAOA"): sec = r.tts99 * SHOT_SEC
    elif r.arm == "Random": continue
    else:
        t_ = dTIME[(dTIME.family == r.family) & (dTIME.arm == r.arm) & (dTIME.n == r.n)].iloc[0]; sec = r.tts99 * t_.sec_per_eval + t_.sdp_sec
    rows.append({"family": r.family, "arm": r.arm, "n": r.n, "seed": r.seed, "tts99_seconds": sec})
dTTSEC = pd.DataFrame(rows); dTTSEC.to_csv("E39_tts_seconds.csv", index=False)
print("\nMedian TTS99 in seconds (QAOA: device time per shot; classical: serial time on this machine; SDP: relaxation included):")
print(dTTSEC.groupby(["family", "arm", "n"]).tts99_seconds.median().unstack().map(lambda v: f"{v:.2e}").to_string())

TPS = []
for s in ([] if RESUME and os.path.exists("E39_time_per_subproblem.csv") else range(20)):
    Qp, h = first_zupdate(N, s); st = exact_stats(Qp, h, keep=True)
    arms_t = {"QAOA p=2 transferred (simulated)": mk_fixed(TRANSFER[("admm", 2)], 2, BUDGET, sim="fast"),
              "QAOA p=2 (simulated)": mk_qaoa(reps=2, shots=SHOTS, maxiter=MAXITER, sim="fast"),
              "SA": mk_sa(BUDGET), "SA restarts": mk_sa_restarts(BUDGET), "SA tuned": mk_sa_tuned(BUDGET), "SA tuned (ext)": mk_sa_cfg(BUDGET, SA_EXT["main"]),
              "Tabu": mk_tabu(BUDGET), "Tabu (restarts)": mk_tabu_r(BUDGET, "main"), "PT": mk_pt(BUDGET, "main"), "Local search": mk_ls(BUDGET),
              "SDP": mk_sdp(BUDGET), "SDP + LS": mk_sdp_ls(BUDGET), "Random": mk_random(BUDGET)}
    for a, f in arms_t.items():
        t0 = time.perf_counter(); f(Qp, h, st, s, {}); TPS.append({"seed": s, "arm": a, "seconds": time.perf_counter() - t0})
if TPS:
    dTPS = pd.DataFrame(TPS).groupby("arm").seconds.median().sort_values(); dTPS.to_csv("E39_time_per_subproblem.csv")
else:
    dTPS = pd.read_csv("E39_time_per_subproblem.csv", index_col=0).iloc[:, 0]; print("[resume] loaded E39_time_per_subproblem.csv (measured timings)")
print(f"\nMedian wall-clock seconds per n=12 subproblem at {BUDGET} evaluations (one machine; QAOA by classical simulation):")
print(dTPS.map(lambda v: f"{v:.4f}").to_string())
print(f"Device time for {BUDGET} shots: {BUDGET * SHOT_SEC:.3f} s")
```

## 40. Hardware at $n=12$: probability of the optimum and time to solution

The mean-gap retention $\eta$ of Section 29 averages over the whole output distribution. Time to solution depends instead on the probability $q$ of the optimum. From the recorded counts, for each instance and mode:
- $q$ on hardware (raw counts, and after readout mitigation);
- $q$ in noiseless simulation;
- their ratio, the retention of the probability of the optimum;
- the implied $\mathrm{TTS}_{99}$.

The classical solvers are scored on the same first subproblems at the budgets chosen for the main family at $n=12$ in Section 39, from 1024 runs.

```python
def _hw_q_rows(runs_by_mode, circuits, label_fmt, readout=None, layouts=None):
    rows = []
    for ir, r in enumerate(circuits):
        st = r["st"]; opt = st["C"] - st["Cstar"] <= 1e-7 * max(1.0, abs(st["Cstar"])); q0 = float(r["p_noiseless"][opt].sum() / r["p_noiseless"].sum())
        rows.append({"seed": r["seed"], "family": r.get("family", "main"), "condition": "noiseless", "q": q0, "q_retention": 1.0, "tts99": tts_q(q0)})
        for mode, runs in runs_by_mode.items():
            key = label_fmt(r)
            if key not in runs: continue
            idx = bits_to_idx(runs[key]); p_raw = np.bincount(idx, minlength=2 ** HW12_N) / len(idx); qh = float(p_raw[opt].sum())
            rows.append({"seed": r["seed"], "family": r.get("family", "main"), "condition": f"hardware {mode}", "q": qh, "q_retention": qh / q0, "tts99": tts_q(qh),
                         "hit_768": block_scores(idx, st)[0]})
            if readout is not None and layouts is not None:
                L_ = layouts[ir]; e01 = np.array([readout["e01"][str(q_)] for q_ in L_]); e10 = np.array([readout["e10"][str(q_)] for q_ in L_])
                qm = float(mitigate(p_raw, e01, e10)[opt].sum())
                rows.append({"seed": r["seed"], "family": r.get("family", "main"), "condition": f"hardware {mode}, mitigated", "q": qm, "q_retention": qm / q0, "tts99": tts_q(qm)})
    return rows

def _classical_tts_on(circuits, fam):
    rows = []
    for r in circuits:
        st = r["st"]; Cs = st["Cstar"]; tol = 1e-7 * max(1.0, abs(Cs)); V, _ = sdp_factor(r["Qp"], r["h"])
        for a, fn in tts_runners(fam, r["Qp"], r["h"], V).items():
            b = BSTARX[(fam, a, 12)] if (fam, a, 12) in BSTARX else 768
            p_ = float((fn(b, TTS_R_TEST, np.random.default_rng([r["seed"], 12, _tag(a), 11])) - Cs <= tol).mean())
            rows.append({"seed": r["seed"], "family": fam, "condition": a, "q": np.nan, "q_retention": np.nan, "tts99": tts_b(p_, b)})
    return rows

if hw12:
    HW12T = [dict(r, family="main") for r in HW12 if r["angles"] == "transferred"]
    def _run_hwq():
        rows = _hw_q_rows(hw12["runs"], HW12T, lambda r: f"{r['seed']}|transferred", hw12.get("readout"), [hw12["layouts"][i] for i, r in enumerate(HW12) if r["angles"] == "transferred"])
        return {"q": pd.DataFrame(rows + _classical_tts_on(HW12T, "main"))}
    dHQ = cached("E40_hardware_q", _run_hwq)["q"]
    print(dHQ.groupby("condition")[["q", "q_retention", "tts99", "hit_768"]].median().round(4).to_string())
    for c_ in [c for c in dHQ.condition.unique() if c.startswith("hardware") and "mitigated" not in c]:
        s_ = dHQ[dHQ.condition == c_].sort_values("seed")
        for comp in ("Tabu (restarts)", "PT", "SA (tuned temperatures)"):
            paired(_l2(s_.tts99.values), _l2(dHQ[dHQ.condition == comp].sort_values("seed").tts99.values),
                   f"E40 {c_} log2 TTS vs {comp}", "E40 hardware TTS", "exploratory")
else:
    print("n=12 hardware counts not available.")
```

## 41. Hardware: suppression arms, a second submission, and the planted family

One submission to the device, with the circuits transpiled once at optimization level 3 and recorded as submitted (two-qubit gate count, depth, layout and OpenQASM 3), together with the device's calibration timestamp:
- **Main family:** the ten $n=12$ transferred-angle circuits of Section 29 in four modes: no suppression, dynamical decoupling only (XY4), Pauli twirling of gates and measurements only, and both. The modes without suppression and with both repeat Section 29 as a second submission.
- **Planted family:** ten $n=12$ first subproblems with the angles trained on held-out planted instances, without suppression and with both techniques.
- **Readout calibration:** all-zeros and all-ones preparations on every measured qubit.

All circuits use 4096 shots and are scored as in Section 29.

```python
def planted_hw_circuits():
    out = []
    for s in range(N_HW12):
        Q_, c_, B_, b_ = _instance(HW12_N, s, "wishart"); Qp, h = build_qubo_subproblem(np.zeros(HW12_N), np.zeros(HW12_N), Q_, B_, b_, RHO)
        st = exact_stats(Qp, h, keep=True); ans, scale, off, _ = qaoa_ansatz(Qp, h, 2); prm = np.asarray(TRANSFER_W[2])
        out.append({"seed": s, "family": "Wishart", "Qp": Qp, "h": h, "st": st, "prm": prm, "circ": ans.assign_parameters(prm),
                    "p_noiseless": qaoa_probs_fast((st["C"] - off) / scale, prm, 2, HW12_N)})
    return out
HWP = planted_hw_circuits(); HWM = [dict(r, family="main") for r in HW12 if r["angles"] == "transferred"]
HWX_FILE = "E41_hardware.json"
HWX_MODES = {"main": ("none", "dd", "twirl", "dd_twirl"), "Wishart": ("none", "dd_twirl")}

def configure(smp, mode):
    smp.options.dynamical_decoupling.enable = mode in ("dd", "dd_twirl")
    if mode in ("dd", "dd_twirl"): smp.options.dynamical_decoupling.sequence_type = "XY4"
    smp.options.twirling.enable_gates = mode in ("twirl", "dd_twirl"); smp.options.twirling.enable_measure = mode in ("twirl", "dd_twirl")
    return smp

def hw_target():
    if os.environ.get("IBM_QUANTUM_TOKEN"): return ibm_backend()
    if QUICK_TEST and fake is not None:
        print("No token: local test of the submission code on a simulator of the device."); return AerSimulator.from_backend(fake, method="statevector")
    return None

if RUN_HW_ARMS and not (RESUME and os.path.exists(HWX_FILE)):
    from qiskit_ibm_runtime import SamplerV2 as IBMSampler
    from qiskit.transpiler import generate_preset_pass_manager
    backend = hw_target()
    if backend is None:
        print("No IBM Quantum token available; hardware runs skipped.")
    else:
        pm_hw = generate_preset_pass_manager(optimization_level=3, backend=backend, seed_transpiler=11)
        sets = {"main": HWM, "Wishart": HWP}; tq = {f: [pm_hw.run(r["circ"]) for r in rs] for f, rs in sets.items()}
        try:
            from qiskit import qasm3
            qasm = {f: [qasm3.dumps(t) for t in ts] for f, ts in tq.items()}
        except Exception:
            qasm = {}
        try:
            props = backend.properties(); cal_time = str(props.last_update_date) if props is not None else None
        except Exception:
            cal_time = None
        saved = {"backend": getattr(backend, "name", str(backend)), "calibration_time": cal_time, "submitted_utc": pd.Timestamp.utcnow().isoformat(),
                 "circuits": {f: [{"seed": r["seed"], "layout": list(t.layout.final_index_layout()), "depth": t.depth(),
                                   "two_qubit": sum(t.count_ops().get(g, 0) for g in ("cx", "cz", "ecr", "rzz"))} for r, t in zip(sets[f], ts)] for f, ts in tq.items()},
                 "qasm": qasm, "runs": {}, "jobs": {}, "usage": {}, "shots": {}}
        for f, modes in HWX_MODES.items():
            for mode in modes:
                smp = configure(IBMSampler(mode=backend), mode); job = smp.run([(t,) for t in tq[f]], shots=HWX_SHOTS)
                print(f"{f} / {mode}: job {job.job_id()}"); res = job.result()
                saved["runs"][f"{f}|{mode}"] = {str(r["seed"]): res[i].data.meas.get_bitstrings() for i, r in enumerate(sets[f])}
                saved["jobs"][f"{f}|{mode}"] = job.job_id(); saved["usage"][f"{f}|{mode}"] = job_quantum_seconds(job)
                saved["shots"][f"{f}|{mode}"] = HWX_SHOTS * len(tq[f])
        qubits = sorted({q_ for f in saved["circuits"] for c_ in saved["circuits"][f] for q_ in c_["layout"]}); nq = backend.num_qubits; cal = []
        for flip in (0, 1):
            qc = QuantumCircuit(nq, len(qubits))
            if flip:
                for q_ in qubits: qc.x(q_)
            for j, q_ in enumerate(qubits): qc.measure(q_, j)
            cal.append(transpile(qc, backend, initial_layout=list(range(nq)), optimization_level=0))
        cres = IBMSampler(mode=backend).run([(c_,) for c_ in cal], shots=HWX_SHOTS).result()
        b0, b1 = cres[0].data.c.get_bitstrings(), cres[1].data.c.get_bitstrings(); k_ = len(qubits)
        saved["readout"] = {"e01": {str(qubits[j]): float(np.mean([bs[k_ - 1 - j] == "1" for bs in b0])) for j in range(k_)},
                            "e10": {str(qubits[j]): float(np.mean([bs[k_ - 1 - j] == "0" for bs in b1])) for j in range(k_)}}
        json.dump(saved, open(HWX_FILE, "w"))

hwx = json.load(open(HWX_FILE)) if os.path.exists(HWX_FILE) else None
if hwx:
    print(f"Backend {hwx['backend']}; calibration {hwx['calibration_time']}; submitted {hwx['submitted_utc']}")
    for f, cs in hwx["circuits"].items():
        print(f"  {f}: two-qubit gates {min(c['two_qubit'] for c in cs)}-{max(c['two_qubit'] for c in cs)}, depth {min(c['depth'] for c in cs)}-{max(c['depth'] for c in cs)}")
    print("  jobs:", hwx["jobs"])
    print("  device usage (s):", {k: (v if v else "not reported") for k, v in hwx["usage"].items()})
    def _run_hwx():
        rows, eta = [], []
        for f, circs in (("main", HWM), ("Wishart", HWP)):
            runs = {k.split("|")[1]: {f"{s}": v for s, v in d.items()} for k, d in hwx["runs"].items() if k.startswith(f + "|")}
            lay = [c["layout"] for c in hwx["circuits"][f]]
            rows += _hw_q_rows(runs, circs, lambda r: f"{r['seed']}", hwx["readout"], lay)
            for r in circs:
                st = r["st"]; gi = mean_gap_rel(r["p_noiseless"], st)
                for mode, rr in runs.items():
                    idx = bits_to_idx(rr[str(r["seed"])]); p = np.bincount(idx, minlength=2 ** HW12_N) / len(idx)
                    eta.append({"family": f, "seed": r["seed"], "mode": mode, "eta": (1 - mean_gap_rel(p, st)) / (1 - gi), "hit_768": block_scores(idx, st)[0]})
        for r in HWP:
            st = r["st"]
            for a, fn in {**{k: v for k, v in strong_arms("Wishart", HW12_BLOCK).items()}, "SA tuned (ext)": mk_sa_cfg(HW12_BLOCK, SA_EXT["Wishart"]),
                          "Local search": mk_ls(HW12_BLOCK), "Random": mk_random(HW12_BLOCK)}.items():
                hits = [gaps(fn(r["Qp"], r["h"], st, 1000 * r["seed"] + j, {})["best"], st)["hit"] for j in range(HWX_SHOTS // HW12_BLOCK)]
                eta.append({"family": "Wishart", "seed": r["seed"], "mode": f"classical: {a}", "eta": np.nan, "hit_768": float(np.mean(hits))})
        return {"q": pd.DataFrame(rows), "modes": pd.DataFrame(eta)}
    HX = cached("E41_hardware", _run_hwx); dHXq, dHXm = HX["q"], HX["modes"]
    print(dHXm.groupby(["family", "mode"])[["eta", "hit_768"]].agg(["mean", "min", "max"]).round(3).to_string())
    print("\nFraction of 768-shot blocks (or 768-evaluation runs) containing the optimum, mean [95% CI over instances]:")
    for (f, m_), s_ in dHXm.groupby(["family", "mode"]):
        _, lo, hi = bootstrap_ci(s_.hit_768.values); print(f"  {f:8s} {m_:32s} {s_.hit_768.mean():.2f} [{lo:.2f}, {hi:.2f}]")
    print("\nMedian over instances:")
    print(dHXq.groupby(["family", "condition"])[["q", "q_retention", "tts99"]].median().round(4).to_string())
    for f, base in (("main", "none"), ("Wishart", "none")):
        b_ = dHXm[(dHXm.family == f) & (dHXm["mode"] == base)].sort_values("seed")
        show_tests([paired(dHXm[(dHXm.family == f) & (dHXm["mode"] == mode)].sort_values("seed").eta.values, b_.eta.values,
                           f"E41 {f} retention {mode} vs none", "E41 suppression", "exploratory") for mode in HWX_MODES[f] if mode != base])
    if hw12:
        e1 = e12[(e12.angles == "transferred") & (e12.condition == "hardware unsuppressed, raw")].set_index("seed").eta
        e2 = dHXm[(dHXm.family == "main") & (dHXm["mode"] == "none")].set_index("seed").eta; ix = e1.index.intersection(e2.index).sort_values()
        t_ = paired(e2[ix].values, e1[ix].values, "E41 main retention, second vs first submission (no suppression)", "E41 drift", "exploratory")
        print(f"Retention without suppression, first vs second submission: {e1[ix].mean():.3f} vs {e2[ix].mean():.3f} (P={fmtP(t_['P'])})")
else:
    print("Hardware runs of this section not available.")
```

**Uncertainty of the probability of the optimum, and a per-instance test of Proposition 3(d).** The hardware probability of the optimum rests on few counts, so each instance's count $k$ out of $N$ shots gets a Clopper–Pearson interval. Retention is corrected for the uniform floor, $(q_{\mathrm{hw}}-2^{-n})/(q_{\mathrm{ideal}}-2^{-n})$, as Proposition 3(d) states it. The median over instances gets a 95% interval from a bootstrap over instances with binomial resampling of the counts. On each instance it is compared with the mean-gap retention $\eta$ of the same run (exact signed-rank test). The Runtime options used for suppression are printed for the record.

```python
from scipy.stats import beta as _beta
from qiskit_ibm_runtime.options import SamplerOptions
_o = SamplerOptions()
print("Dynamical decoupling options (defaults, with enable and sequence_type set per mode):", _o.dynamical_decoupling)
print("Twirling options (defaults, with enable_gates and enable_measure set per mode):", _o.twirling)

U12 = 2.0 ** -HW12_N
def _cp(k, N_, a=0.05):
    return (0.0 if k == 0 else _beta.ppf(a / 2, k, N_ - k + 1), 1.0 if k == N_ else _beta.ppf(1 - a / 2, k + 1, N_ - k))
RET = []
if hw12 and hwx is not None:
    src = []
    for mode, eta_c in (("suppressed", "hardware suppressed, raw"), ("unsuppressed", "hardware unsuppressed, raw")):
        q_ = dHQ[dHQ.condition == f"hardware {mode}"].set_index("seed").q; qi = dHQ[dHQ.condition == "noiseless"].set_index("seed").q
        et = e12[(e12.angles == "transferred") & (e12.condition == eta_c)].set_index("seed").eta
        N_ = len(hw12["runs"][mode][f"{q_.index[0]}|transferred"])
        src.append(("first", "main", "dd_twirl" if mode == "suppressed" else "none", q_, qi, et, N_))
    for f in ("main", "Wishart"):
        qi = dHXq[(dHXq.family == f) & (dHXq.condition == "noiseless")].set_index("seed").q
        for m in HWX_MODES[f]:
            q_ = dHXq[(dHXq.family == f) & (dHXq.condition == f"hardware {m}")].set_index("seed").q
            et = dHXm[(dHXm.family == f) & (dHXm["mode"] == m)].set_index("seed").eta
            src.append(("second", f, m, q_, qi, et, len(hwx["runs"][f"{f}|{m}"][str(q_.index[0])])))
    for sub, f, m, q_, qi, et, N_ in src:
        for sd in q_.index:
            k = int(round(q_[sd] * N_)); lo, hi = _cp(k, N_); den = qi[sd] - U12
            RET.append({"submission": sub, "family": f, "mode": m, "seed": sd, "shots": N_, "count": k, "q_hw": k / N_, "q_lo": lo, "q_hi": hi,
                        "q_ideal": qi[sd], "q_ret": (k / N_ - U12) / den, "q_ret_lo": (lo - U12) / den, "q_ret_hi": (hi - U12) / den, "eta": et[sd]})
dRET = pd.DataFrame(RET); dRET.to_csv("E41_retention_comparison.csv", index=False)
if len(dRET):
    SUMR = []
    for (sub, f, m), g in dRET.groupby(["submission", "family", "mode"], sort=False):
        rng_ = np.random.default_rng(5); meds = []
        for _ in range(4000):
            ix = rng_.integers(0, len(g), len(g)); gg = g.iloc[ix]
            meds.append(np.median((rng_.binomial(gg.shots.values, gg.q_hw.values) / gg.shots.values - U12) / (gg.q_ideal.values - U12)))
        t_ = paired(g.q_ret.values, g.eta.values, f"E41 {sub} {f} {m}: probability-of-optimum retention vs eta", "E41 retention", "exploratory")
        SUMR.append({"submission": sub, "family": f, "mode": m, "median_count": g["count"].median(), "min_count": g["count"].min(),
                     "q_ret_median": g.q_ret.median(), "q_ret_lo": np.quantile(meds, 0.025), "q_ret_hi": np.quantile(meds, 0.975),
                     "eta_median": g.eta.median(), "median_ratio_eta_over_qret": float(np.median(g.eta / g.q_ret)),
                     "ratio_of_medians": float(g.eta.median() / g.q_ret.median()), "instances_qret_below_eta": int((g.q_ret < g.eta).sum()), "instances": len(g), "P_signed_rank": t_["P"]})
    dSUMR = pd.DataFrame(SUMR); dSUMR.to_csv("E41_retention_summary.csv", index=False)
    print("Floor-corrected retention of the probability of the optimum versus mean-gap retention, per submission, family, and mode:")
    print(dSUMR.round(3).to_string(index=False))
```

## 42. ADMM with the binary step on hardware, on moving trajectories

In Section 29 the hardware-driven ADMM runs held one bitstring from the first iteration on, so the hardware only had to resample one optimum. Here the instances are planted-family instances at $\rho=2$ whose noiseless transferred-QAOA trajectory changes its bitstring at least twice in six iterations, selected in simulation before any hardware run. Three instances run six iterations with the binary step on the device (768 shots per iteration, no suppression), compared with the same runs under noiseless sampling and the device-matched simulation.

```python
def moving_seeds(k=3, K=6, pool=80):
    found, changes = [], {}
    for s in range(pool):
        df, _ = run_admm(HW12_N, s, mk_fixed(TRANSFER_W[2], 2, FIX_BUDGET, sim="fast"), K, gen="wishart")
        changes[s] = int((np.diff(df.zidx.values) != 0).sum())
        if changes[s] >= min(2, K - 1): found.append(s)
        if len(found) == k: return found
    return found + [s for s in sorted(changes, key=changes.get, reverse=True) if s not in found][:k - len(found)]
HW_MOVE_SEEDS = moving_seeds(N_HW_ADMM, K_HW_ADMM); print("Selected planted instances:", HW_MOVE_SEEDS)


def hw_admm_solver_p(sampler, pm, prm, shots=FIX_BUDGET):
    def f(Qp, h, st, s, ctx):
        ans, _, _, _ = qaoa_ansatz(Qp, h, 2); tq = pm.run(ans.assign_parameters(prm))
        idx = bits_to_idx(sampler.run([(tq,)], shots=shots).result()[0].data.meas.get_bitstrings()); k = int(idx[np.argmin(st["C"][idx])])
        return {"best": float(st["C"][k]), "z": z_from_index(k, len(h)), "evals": shots}
    return f

HW_MOVE_FILE = "E42_hw_admm_seeds.csv"
if RUN_HW_ADMM_MOVING and not (RESUME and os.path.exists(HW_MOVE_FILE)):
    from qiskit_ibm_runtime import SamplerV2 as IBMSampler
    from qiskit.transpiler import generate_preset_pass_manager
    backend = hw_target()
    if backend is None:
        print("No IBM Quantum token available; hardware ADMM skipped.")
    else:
        pm_hw = generate_preset_pass_manager(optimization_level=3, backend=backend, seed_transpiler=11)
        def _runs(smp):
            its, sms = [], []
            for s in HW_MOVE_SEEDS:
                df, sm = run_admm(HW12_N, s, hw_admm_solver_p(configure(smp, "none"), pm_hw, TRANSFER_W[2]), K_HW_ADMM, gen="wishart")
                df["seed"], df["arm"] = s, "hardware"; its.append(df); sms.append({"seed": s, "arm": "hardware", **sm}); print(f"  instance {s} done")
            return its, sms
        try:
            from qiskit_ibm_runtime import Session
            with Session(backend=backend) as session: its, sms = _runs(IBMSampler(mode=session))
        except Exception as e:
            print("Session unavailable, job mode:", type(e).__name__); its, sms = _runs(IBMSampler(mode=backend))
        pd.concat(its).to_csv("E42_hw_admm_iters.csv", index=False); pd.DataFrame(sms).to_csv(HW_MOVE_FILE, index=False)

if os.path.exists(HW_MOVE_FILE):
    def device_sim_solver_p(prm, shots=FIX_BUDGET):
        from qiskit.transpiler import generate_preset_pass_manager
        pmf = generate_preset_pass_manager(optimization_level=3, backend=fake, seed_transpiler=11); simd = AerSimulator.from_backend(fake, method="statevector")
        def f(Qp, h, st, s, ctx):
            ans, _, _, _ = qaoa_ansatz(Qp, h, 2); tq = pmf.run(ans.assign_parameters(prm))
            idx = bits_to_idx([m_.replace(" ", "") for m_ in simd.run(tq, shots=shots, memory=True, seed_simulator=s).result().get_memory()])
            k = int(idx[np.argmin(st["C"][idx])]); return {"best": float(st["C"][k]), "z": z_from_index(k, len(h)), "evals": shots}
        return f
    def _refs():
        arms = {"noiseless": mk_fixed(TRANSFER_W[2], 2, FIX_BUDGET, sim="fast"), "exact": mk_exact()}
        if fake is not None: arms["device simulation"] = device_sim_solver_p(TRANSFER_W[2])
        out = run_arms_embedded(arms, HW_MOVE_SEEDS, HW12_N, K_HW_ADMM, "E42 references", gen="wishart")
        return {"seeds": out["seeds"], "iters": out["iters"]}
    R42 = cached("E42_references", _refs)
    it_hw = pd.read_csv("E42_hw_admm_iters.csv"); it_all = pd.concat([it_hw, R42["iters"]], ignore_index=True)
    print("Bitstring changes per instance:", it_all.groupby(["arm", "seed"]).zidx.apply(lambda v: int((np.diff(v.values) != 0).sum())).unstack().to_string())
    print("Subproblems solved exactly:", it_all.groupby("arm").hit.mean().round(3).to_dict())
    dHAX = pd.concat([pd.read_csv(HW_MOVE_FILE), R42["seeds"]], ignore_index=True); dHAX.to_csv("E42_hw_admm_summary.csv", index=False)
    print(dHAX.groupby("arm")[["resid_final", "mbo_gap_best_unif", "mbo_opt_found"]].mean().round(4).to_string())
else:
    print("Hardware ADMM of this section not available.")
```

## 43. Mixed-binary family without the QUBO reduction, extended

The general family of Section 28 is extended to 80 instances and penalties $\rho\in\{0.5,1,2,4\}$, with tabu search with restarts and two references:
- **ADMM with restarts:** four runs of three iterations each from random continuous starting points, with exact subproblem solutions and the same total of 12 subproblems. This is the classical restart heuristic for nonconvex ADMM.
- **Problem optimum:** obtained by solving one box-constrained quadratic program for every $z$.

```python
def run_admm_general_restarts(inst, Fall, solver, K, rho, seed, R=4):
    Q, P, c, A_, B, b, box = (inst[k] for k in ("Q", "P", "c", "A", "B", "b", "box"))
    n = B.shape[1]; m = P.shape[0]; Lc = np.linalg.cholesky(P + rho * A_.T @ A_); rng = np.random.default_rng(seed); Zs = []
    per = max(1, K // R)
    for r in range(max(1, K // per)):
        x = rng.uniform(-box, box, m); lam = np.zeros(len(b))
        for k in range(per):
            Qp = Q + (rho / 2.0) * (B.T @ B); Qp = (Qp + Qp.T) / 2.0; h = B.T @ lam + rho * (B.T @ (A_ @ x - b))
            st = exact_stats(Qp, h, keep=True); z = np.asarray(solver(Qp, h, st, seed * 1000 + 10 * r + k, {})["z"], float)
            g_ = -c + A_.T @ lam + rho * A_.T @ (B @ z - b)
            x = lsq_linear(Lc.T, -np.linalg.solve(Lc, g_), bounds=(-box, box), method="bvls", tol=1e-12).x
            lam = lam + rho * (A_ @ x + B @ z - b); Zs.append(z_to_index(z))
    fin = np.isfinite(Fall); Fstar = float(Fall[fin].min()); Fmean = float(Fall[fin].mean()); fb = float(np.min(Fall[Zs])); tolF = 1e-6 * max(1, abs(Fstar))
    return {"mbo_opt_found": float(fb - Fstar <= tolF), "mbo_gap_best_unif": float((fb - Fstar) / max(Fmean - Fstar, 1e-12)) if np.isfinite(fb) else np.nan}

arms_GX = {"Exact": mk_exact(), "SA tuned (ext)": mk_sa_cfg(BUDGET, SA_EXT["main"]), "Tabu (restarts)": mk_tabu_r(BUDGET, "main"),
           "Random": mk_random(BUDGET), "QAOA p=2 transferred": mk_fixed(TRANSFER[("admm", 2)], 2, BUDGET, sim="fast")}
def _run_generalX():
    sms = []; t0 = time.time()
    for s in range(N_GENX):
        inst = generate_general(N, s); Fall, _ = general_values(inst)
        for r_ in RHOS_GENX:
            for a, f in arms_GX.items():
                _, sm = run_admm_general(inst, Fall, f, K_MAIN, r_, s); sms.append({"seed": s, "arm": a, "rho": r_, "mbo_opt_found": sm["mbo_opt_found"], "mbo_gap_best_unif": sm["mbo_gap_best_unif"]})
            sms.append({"seed": s, "arm": "Exact, ADMM with restarts", "rho": r_, **run_admm_general_restarts(inst, Fall, mk_exact(), K_MAIN, r_, s)})
        if (s + 1) % max(1, N_GENX // 5) == 0: print(f"  [E43] instance {s + 1}/{N_GENX} ({time.time() - t0:.0f}s)")
    return {"seeds": pd.DataFrame(sms)}
dGX = cached("E43_general", _run_generalX)["seeds"]
print(dGX.pivot_table(index="arm", columns="rho", values="mbo_opt_found").round(3).to_string())
for r_ in RHOS_GENX:
    s_ = dGX[dGX.rho == r_]
    show_tests([paired(col(s_, "QAOA p=2 transferred", "mbo_gap_best_unif"), col(s_, a, "mbo_gap_best_unif"), f"E43 rho={r_:g} transferred vs {a}",
                       f"E43 general rho={r_:g}", "exploratory") for a in ("Exact", "SA tuned (ext)", "Tabu (restarts)", "Random", "Exact, ADMM with restarts")])
```

**General family with planted couplings.** The one family in which the decomposition is not a reformulation of a single QUBO is the only place where a better subproblem solver could change the ADMM outcome. Here it is given a rugged binary part: $Q$ is drawn from the Wishart planted ensemble ($\alpha=0.5$, scaled to unit maximum magnitude), with weaker coupling to the continuous block ($B\sim U[-0.1,0.1]$) so that the subproblems inherit the planted landscape; $H$, $A$, $c$, and the box are as above (seed offset 40000). Every solver of the planted-family comparison is run inside ADMM at the same penalties, with transferred angles trained on the planted family.

```python
def generate_general_planted(n, seed, m=GEN_M, r=GEN_R, box=GEN_BOX):
    rng = np.random.default_rng(GENP_OFFSET + seed)
    Q = wishart_Q(n, W_ALPHA, rng)
    L = rng.normal(size=(m, m)) / np.sqrt(m); P = L @ L.T + 0.5 * np.eye(m)
    c = rng.uniform(-1, 1, m); A_ = rng.normal(size=(r, m)); B = rng.uniform(-GENP_BSCALE, GENP_BSCALE, (r, n))
    x0 = rng.uniform(-0.5 * box, 0.5 * box, m); z0 = rng.integers(0, 2, n).astype(float)
    return {"Q": Q, "P": P, "c": c, "A": A_, "B": B, "b": A_ @ x0 + B @ z0, "box": box}

arms_GP = {"Exact": mk_exact(), "QAOA p=2 transferred": mk_fixed(TRANSFER_W[2], 2, BUDGET, sim="fast"), "SDP + LS": mk_sdp_ls(BUDGET),
           "Tabu (restarts)": mk_tabu_r(BUDGET, "Wishart"), "PT": mk_pt(BUDGET, "Wishart"),
           ("SA tuned (ext)" if SA_WIDE == SA_EXT["Wishart"] else "SA tuned (wide)"): mk_sa_cfg(BUDGET, SA_WIDE),
           "Local search": mk_ls(BUDGET), "Random": mk_random(BUDGET)}
def _run_general_planted():
    sms = []; t0 = time.time()
    for s in range(N_GENX):
        inst = generate_general_planted(N, s); Fall, _ = general_values(inst)
        for r_ in RHOS_GENX:
            for a, f in arms_GP.items():
                df, sm = run_admm_general(inst, Fall, f, K_MAIN, r_, s)
                sms.append({"seed": s, "arm": a, "rho": r_, "hit": float(df.hit.mean()), "mbo_opt_found": sm["mbo_opt_found"], "mbo_gap_best_unif": sm["mbo_gap_best_unif"]})
            sms.append({"seed": s, "arm": "Exact, ADMM with restarts", "rho": r_, "hit": 1.0, **run_admm_general_restarts(inst, Fall, mk_exact(), K_MAIN, r_, s)})
        if (s + 1) % max(1, N_GENX // 5) == 0: print(f"  [E43 planted] instance {s + 1}/{N_GENX} ({time.time() - t0:.0f}s)")
    return {"seeds": pd.DataFrame(sms)}
dGP = cached("E43_general_planted", _run_general_planted)["seeds"]
print("Subproblems solved exactly (instance mean):"); print(dGP.pivot_table(index="arm", columns="rho", values="hit").round(3).to_string())
print("\nProblem optimum found:"); print(dGP.pivot_table(index="arm", columns="rho", values="mbo_opt_found").round(3).to_string())
for r_ in RHOS_GENX:
    s_ = dGP[dGP.rho == r_]
    show_tests([paired(col(s_, "QAOA p=2 transferred", "mbo_gap_best_unif"), col(s_, a, "mbo_gap_best_unif"), f"E43 planted rho={r_:g} transferred vs {a}",
                       f"E43 general planted rho={r_:g}", "exploratory")
                for a in ("Exact", "SDP + LS", "Tabu (restarts)", "PT", next(k for k in arms_GP if k.startswith("SA")), "Local search", "Random", "Exact, ADMM with restarts")])
```

## 44. Sensitivity: minimum detectable effect

Power of the primary QAOA ($p=2$) versus random-search comparison, by injecting location shifts into the centred paired differences; the minimum detectable effect is interpolated at 80% power.

```python
def power_curve(diffs, n_target, grid, n_boot, alpha=0.05, seed=0):
    d = np.asarray(diffs) - np.median(diffs); rng = np.random.default_rng(seed); out = []
    for sh in grid:
        rej = 0
        for _ in range(n_boot):
            smp = rng.choice(d, n_target) + sh
            try:
                rej += wilcoxon(smp).pvalue < alpha
            except ValueError:
                pass
        out.append(rej / n_boot)
    return np.array(out)

def interp_mde(grid, pw, target=0.8):
    idx = np.where(pw >= target)[0]
    if len(idx) == 0: return np.nan
    i = idx[0]
    if i == 0: return float(grid[0])
    return float(grid[i - 1] + (target - pw[i - 1]) * (grid[i] - grid[i - 1]) / (pw[i] - pw[i - 1]))

mde_diffs = col(dA, "QAOA p=2", "gap_unif") - col(dA, "Random", "gap_unif")
MDE_GRID = np.linspace(0, 2.5 * max(np.std(mde_diffs), 1e-6), 41)
POWER = {nt: power_curve(mde_diffs, nt, MDE_GRID, MDE_BOOT, seed=nt) for nt in (40, 80)}
for nt, pw in POWER.items():
    print(f"  {nt} instances: MDE (80% power) = {interp_mde(MDE_GRID, pw):.4f}; power at zero shift = {pw[0]:.3f}")
print(f"Observed mean difference (QAOA p=2 minus Random): {np.mean(mde_diffs):+.4f}")
pd.DataFrame({"shift": MDE_GRID, **{f"power_{k}": v for k, v in POWER.items()}}).to_csv("S23_power.csv", index=False)
```

## 46. Test registry

Every test with its family, label, effective sample size, exact $P$, Holm-adjusted $P$ within family, and the global Bonferroni decision. Tests on identical data appear once, with the other labels listed. *Confirmatory* marks a comparison whose hypothesis was fixed when its experiment was designed; the planted family and the stronger solvers were designed after the main-family results, so their tests are confirmatory only in that sense. *Exploratory* marks everything else, including all time-to-solution and hardware tests.

```python
reg = pd.DataFrame(REGISTRY).drop_duplicates("label", keep="last").reset_index(drop=True)
if "also_reported_as" not in reg: reg["also_reported_as"] = ""
reg["P_holm_family"] = reg.groupby("family")["P"].transform(lambda p: holm(p.values))
thr = 0.05 / len(reg); reg["survives_global_bonferroni"] = reg.P < thr
reg.to_csv("S25_test_registry.csv", index=False)
conf = reg[reg.kind == "confirmatory"]
print(f"{len(reg)} tests ({len(conf)} confirmatory, {len(reg) - len(conf)} exploratory); global threshold P < {thr:.2e}")
print(f"  survive global threshold: {int(reg.survives_global_bonferroni.sum())}")
print(f"  significant after Holm within family: {int((reg.P_holm_family < 0.05).sum())}")
print(f"  significant at 0.05 but not globally: {int(((reg.P < 0.05) & ~reg.survives_global_bonferroni).sum())}")
print(f"  tests reported under more than one label: {int((reg.also_reported_as.fillna('') != '').sum())}")
```

```python
# names used in the manuscript, applied to tick labels and legends when a figure is saved
DISPLAY = {"QAOA p=2 transferred": "Transferred QAOA", "QAOA p=3 transferred": "Transferred QAOA, p=3",
           "QAOA p=2 transferred (n=16 angles)": "Transferred QAOA, n=16 angles", "QAOA p=2 transferred (main-family angles)": "Transferred, main-family angles",
           "QAOA p=2": "Optimized QAOA", "QAOA p=1": "Optimized QAOA, p=1", "QAOA p=3": "Optimized QAOA, p=3", "QAOA multi-start": "Optimized QAOA, multi-start",
           "QAOA p=2 warm-start": "Optimized QAOA, warm-start", "QAOA p=2 unnormalized": "Optimized QAOA, unnormalized",
           "QAOA budget control": "Optimized QAOA, budget control", "QAOA p=2 from transferred angles": "Optimized from transferred angles",
           "SA tuned": "SA, tuned (small grid)", "SA tuned (ext)": "SA, tuned", "SA tuned (wide)": "SA, tuned (wide grid)", "SA (tuned temperatures)": "SA, tuned",
           "SA restarts": "SA, restarts", "Tabu (restarts)": "Tabu, restarts", "SDP + LS": "SDP+LS", "Random": "Random search",
           "Exact, ADMM with restarts": "Exact, ADMM restarts"}
def disp(t):
    if t in DISPLAY: return DISPLAY[t]
    if ": " in t:
        a, b = t.split(": ", 1); return f"{a}: {DISPLAY.get(b, b)}"
    return t

def _relabel(fig):
    fig.canvas.draw()
    for ax in fig.axes:
        for axis in (ax.xaxis, ax.yaxis):
            labs = [t.get_text() for t in axis.get_ticklabels()]
            new = [disp(t) for t in labs]
            if new != labs:
                axis.set_ticks(axis.get_ticklocs()); axis.set_ticklabels(new)
        if ax.get_legend() is not None:
            for t in ax.get_legend().get_texts(): t.set_text(disp(t.get_text()))
    for lg in fig.legends:
        for t in lg.get_texts(): t.set_text(disp(t.get_text()))

def save_figure(fig, name):
    _relabel(fig)
    fig.savefig(os.path.join(FIG_DIR, f"{name}.png"), dpi=600, bbox_inches="tight", pad_inches=0.02)
    fig.savefig(os.path.join(FIG_DIR, f"{name}.pdf"), bbox_inches="tight", pad_inches=0.02)
    plt.show(); plt.close(fig)
    print(f"Saved {FIG_DIR}/{name}.png and .pdf")
```

```python
def colr(a):
    if a in C: return C[a]
    if "SA tuned" in a: return C["SA-T"]
    if "SA restarts" in a: return C["SA-R"]
    if a == "Tabu (restarts)": return C["Tabu-R"]
    if a == "PT": return C["PT"]
    if a in ("SDP + LS", "SDP rounding"): return C["SDP-LS"] if a == "SDP + LS" else C["SDP"]
    if "Tabu" in a: return C["Tabu"]
    if "Local search" in a: return C["LS"]
    if "transferred" in a.lower(): return C["QAOA-T"]
    if "multi" in a.lower() or a.startswith("MS"): return C["QAOA-MS"]
    if "warm" in a.lower(): return C["QAOA-W"]
    if "unnormalized" in a or "no top-up" in a: return C["QAOA-raw"]
    if a.startswith(("QAOA", "SS", "COBYLA", "Nelder", "SPSA", "Standard")): return C["QAOA"]
    if a.startswith("SA"): return C["SA"]
    return "#444444"

def ci_rows(ax, d, arms, metric, ref_uniform=False, xlabel=r"Normalized gap $g_{\mathrm{unif}}$", xlim=None):
    y = np.arange(len(arms))[::-1]; top = 0.0
    for yi, a in zip(y, arms):
        m, lo, hi = bootstrap_ci(d[d.arm == a][metric]); top = max(top, hi if np.isfinite(hi) else 0.0)
        ax.errorbar(m, yi, xerr=[[m - lo], [hi - m]], fmt="o", color=colr(a), ms=3.5, capsize=1.5, elinewidth=0.9)
    ax.set_yticks(y); ax.set_yticklabels(arms); ax.set_xlabel(xlabel)
    top = top if top > 0 else 0.01
    ax.set_xlim(*(xlim or (-0.04 * top, 1.12 * top)))
    if ref_uniform and top > 0.8: ax.axvline(1.0, color="0.4", ls="--", lw=0.7)
    ax.axvline(0.0, color="0.7", lw=0.5); ax.grid(axis="y", alpha=0)

def line_ci(ax, x, groups, label, color, marker="o", ls="-", hollow_at=None):
    ms, los, his = [], [], []
    for g in groups:
        m, lo, hi = bootstrap_ci(g); ms.append(m); los.append(lo); his.append(hi)
    ms, los, his = map(np.array, (ms, los, his))
    ax.fill_between(x, los, his, color=color, alpha=0.15, lw=0)
    ax.plot(x, ms, ls=ls, color=color, label=label)
    for xi, mi in zip(x, ms):
        hollow = hollow_at is not None and xi in hollow_at
        ax.plot(xi, mi, marker, color=color, mfc="white" if hollow else color, ms=3.5)
def line_m(ax, x, groups, label, color, marker="o", ls="-", mfc=None, ms=3.6, band=True):
    # mean line with markers drawn in the legend; optional bootstrap band
    ms_, lo_, hi_ = [], [], []
    for g in groups:
        m, lo, hi = bootstrap_ci(g); ms_.append(m); lo_.append(lo); hi_.append(hi)
    if band: ax.fill_between(x, lo_, hi_, color=color, alpha=0.12, lw=0)
    ax.plot(x, ms_, ls=ls, marker=marker, color=color, mfc=mfc or color, ms=ms, label=label)
print("Figure helpers ready.")
```

## 50. Revision analyses of the cached results

Analyses added in revision. They read the cached results of the earlier sections and run no new experiment, except the optional re-execution of the device-matched simulation in Section 50.3. None of their tests enters the test log of Section 46.

- **50.1** Scoring rules with independent draws: the penalty sweep repeats the main optimized-QAOA arm on the same instances and seeds with the exact-distribution sampler.
- **50.2** Planted family: re-scoring of the shared-trajectory runs under criteria that also credit near-degenerate competitors.
- **50.3** Hardware at $n=12$: retention resolved by threshold and by Hamming distance; probability of the optimum under the device-matched model.
- **50.4** Circuit sampler: check of the random stream shared across calls.
- **50.5** Planted-family angles: distance under the symmetries of the QAOA output distribution.
- **50.6** Derived quantities quoted in the manuscript.

### 50.1 Scoring rules with independent draws

Optimized QAOA ($p=2$) against random search under best-sample and read-out scoring: the main comparison (circuit sampler, Section 11), the penalty sweep (exact-distribution sampler, Section 27) at every penalty, and the ADMM-embedded runs at $n=16$ and $20$ (Section 26). At $\rho=2$ the classical arms of Sections 11 and 27 are identical, so the two experiments differ only in how QAOA's shots were drawn.

```python
def scoring_compare(df, label):
    q = df[df.arm == "QAOA p=2"].sort_values("seed"); rd = df[df.arm == "Random"].sort_values("seed")
    assert (q.seed.values == rd.seed.values).all()
    bs = paired(q.gap_unif.values, rd.gap_unif.values, record=False)
    ro = paired(q.gap_ro_unif.values, rd.gap_unif.values, record=False)
    d, lo, hi = bootstrap_ci(q.hit.values - rd.hit.values)
    return {"setting": label, "instances": len(q), "hit_best": q.hit.mean(), "hit_readout": q.hit_ro.mean(), "hit_random": rd.hit.mean(),
            "gap_best": q.gap_unif.mean(), "gap_readout": q.gap_ro_unif.mean(), "gap_random": rd.gap_unif.mean(),
            "r_readout": ro["r"], "P_readout": ro["P"], "nontied_readout": ro["n_nonzero"],
            "r_best": bs["r"], "P_best": bs["P"], "nontied_best": bs["n_nonzero"],
            "hit_diff_pp": 100 * d, "hit_diff_lo": 100 * lo, "hit_diff_hi": 100 * hi, "problem_opt": q.mbo_opt_found.mean()}

rv_s10 = pd.read_csv("S10_main_seeds.csv"); rv_r27 = pd.read_csv("R27_penalty_seeds.csv"); rv_r26 = pd.read_csv("R26_embedded_seeds.csv")
RV1 = [scoring_compare(rv_s10, "n=12, rho=2, circuit sampler")]
RV1 += [scoring_compare(rv_r27[rv_r27.rho == r_], f"n=12, rho={r_:g}, independent draws") for r_ in sorted(rv_r27.rho.unique())]
RV1 += [scoring_compare(rv_r26[rv_r26.n == n_], f"n={n_}, rho=2, independent draws") for n_ in (16, 20)]
dRV1 = pd.DataFrame(RV1); dRV1.to_csv("RV1_scoring_replicate.csv", index=False)
print(dRV1.round(4).to_string(index=False))

a_ = rv_s10[rv_s10.arm == "QAOA p=2"].sort_values("seed"); b_ = rv_r27[(rv_r27.arm == "QAOA p=2") & (rv_r27.rho == 2.0)].sort_values("seed")
for arm in ("Exact", "SA", "Random"):
    x_ = rv_s10[rv_s10.arm == arm].sort_values("seed"); y_ = rv_r27[(rv_r27.arm == arm) & (rv_r27.rho == 2.0)].sort_values("seed")
    print(f"{arm}: identical in Sections 11 and 27 at rho=2: {bool(np.allclose(x_.gap_unif.values, y_.gap_unif.values))}")
print("\nOptimized QAOA at rho=2, circuit sampler minus independent draws:")
for col in ("gap_unif", "gap_ro_unif", "hit", "hit_ro", "mbo_opt_found"):
    t_ = paired(a_[col].values, b_[col].values, record=False)
    print(f"  {col:14s} {a_[col].mean():.4f} vs {b_[col].mean():.4f}   r={t_['r']:+.3f}  P={fmtP(t_['P'])}  non-tied {t_['n_nonzero']}")
```

### 50.2 Planted family: near-degenerate competitors

The exact-solver trajectories of the 80 planted test instances are recomputed by exhaustive enumeration (the subproblems of Sections 35 and 37). Each solver's logged best cost is classified as the optimum, the bitwise complement of the optimum, or another bitstring, and ranked among all $2^n$ bitstrings. No solver is re-run.

```python
rv_rows = []
for s in range(N_SH):
    for k, (Qp, h, st) in enumerate(exact_trajectory(N, s, K_MAIN, gen="wishart")):
        C_ = st["C"]; kopt = int(np.argmin(C_)); kc = kopt ^ (2 ** N - 1); order = np.argsort(C_)
        rv_rows.append({"seed": s, "iter": k, "Cstar": float(C_[kopt]), "gap_comp": float(C_[kc] - C_[kopt]),
                        "gap_unif_comp": float((C_[kc] - C_[kopt]) / (st["Cmean"] - C_[kopt])), "rank_comp": int(np.where(order == kc)[0][0]),
                        "top_gaps": [float(C_[order[j]] - C_[kopt]) for j in range(16)]})
rvT = pd.DataFrame(rv_rows)
rvL = pd.concat([pd.read_csv("H35_shared_all_iters.csv"), pd.read_csv("E37_shared_iters.csv")], ignore_index=True)
rvL = rvL[rvL.family == "Wishart"].merge(rvT, on=["seed", "iter"])
tol_ = 1e-9 * np.maximum(1.0, np.abs(rvL.Cstar))
rvL["is_opt"] = rvL.gap <= tol_
rvL["is_comp"] = (~rvL.is_opt) & (np.abs(rvL.gap - rvL.gap_comp) <= tol_)
rvL["rank"] = [next((j for j, g in enumerate(tg) if abs(g - g0) <= t0), 16) for tg, g0, t0 in zip(rvL.top_gaps, rvL.gap, tol_)]
rvL["opt_or_comp"] = rvL.is_opt | rvL.is_comp; rvL["best2"] = rvL["rank"] <= 1; rvL["best4"] = rvL["rank"] <= 3; rvL["g010"] = rvL.gap_unif <= 0.10
rvL.drop(columns="top_gaps").to_csv("RV2_planted_complement_iters.csv", index=False)
crit = ["is_opt", "opt_or_comp", "best2", "best4", "g010"]
rvI = rvL.groupby(["arm", "seed"])[crit + ["is_comp"]].mean()
rvS = rvI.groupby("arm").mean()
rvS["misses_at_complement"] = rvL[~rvL.is_opt].groupby("arm").is_comp.mean()
rvS.to_csv("RV2_planted_complement_summary.csv")
print("Fraction of subproblems solved (instance-level means):"); print(rvS.round(4).to_string())
print("\nComplement of the optimum, 960 subproblems: rank (1 = second best):", rvT.rank_comp.value_counts().sort_index().head(4).to_dict(),
      f"; among the four best on {int((rvT.rank_comp <= 3).sum())}")
print(f"normalized gap of the complement over all 960: median {rvT.gap_unif_comp.median():.4f}, "
      f"10th-90th percentile {rvT.gap_unif_comp.quantile(.1):.4f}-{rvT.gap_unif_comp.quantile(.9):.4f}")
qI = rvI.loc["QAOA p=2 transferred"]; rvP = []
for a in ("Tabu (restarts)", "Local search", "PT", "SA tuned (ext)", "SDP + LS"):
    for c in crit:
        t_ = paired(qI[c].values, rvI.loc[a][c].values, record=False)
        rvP.append({"comparator": a, "criterion": c, "P": t_["P"], "r": t_["r"], "non_tied": t_["n_nonzero"]})
dRVP = pd.DataFrame(rvP); dRVP.to_csv("RV2_planted_complement_tests.csv", index=False)
print("\nTransferred QAOA versus each solver, instance-level exact signed-rank test (r>0: QAOA solves more):")
print(dRVP.pivot(index="comparator", columns="criterion", values="P").to_string(float_format=lambda v: f"{v:.2g}"))
loc_ = ["Tabu (restarts)", "Local search", "PT", "SA tuned (ext)"]
print("largest P against the four tuned local heuristics:", f"{dRVP[dRVP.comparator.isin(loc_)].P.max():.2g}")
```

### 50.3 Hardware at $n=12$: threshold- and Hamming-resolved retention

For every instance and run, the floor-corrected retention of a set $\Omega$ of bitstrings, $(\pi_{\mathrm{hw}}(\Omega)-u)/(\pi_{\mathrm{ideal}}(\Omega)-u)$ with $u=|\Omega|/2^n$, which global depolarization predicts to equal $\eta$ for every $\Omega$ (Proposition 3(d)). Sets: $\{g\le t\}$, the probability deficit of $\{g>t\}$, and Hamming shells around the optimum. With `RV_DEVICE_SIM = True`, the device-matched simulation of the first submission is re-executed with the transpiler and simulator seeds of Section 29 (several minutes per circuit) to obtain its probability of the optimum; set it to `False` to skip.

```python
RV_DEVICE_SIM = True
_U = 2.0 ** -HW12_N; _NB = 2 ** HW12_N; _pc = np.array([bin(i).count("1") for i in range(_NB)])
RV_TH = [0.0, 0.02, 0.05, 0.1, 0.2, 0.3, 0.5]; RV_TH_BAD = [1.0, 1.5, 2.0]
def _ret(ph, pi, m):
    u = m.mean(); return (ph[m].sum() - u) / (pi[m].sum() - u)
RV3 = []
def rv_resolve(source, sub, fam, mode, r, idx):
    st = r["st"]; g = (st["C"] - st["Cstar"]) / (st["Cmean"] - st["Cstar"]); pi = r["p_noiseless"]; ph = np.bincount(idx, minlength=_NB) / len(idx)
    d = _pc[np.arange(_NB) ^ int(np.argmin(st["C"]))]
    o = {"source": source, "submission": sub, "family": fam, "mode": mode, "seed": r["seed"], "count_opt": int(round(ph[d == 0].sum() * len(idx))),
         "q": float(ph[d == 0].sum()), "q_ideal": float(pi[d == 0].sum()), "eta": (1 - mean_gap_rel(ph, st)) / (1 - mean_gap_rel(pi, st))}
    for t in RV_TH: o[f"ret_g<={t}"] = _ret(ph, pi, g <= t + 1e-12); o[f"share_g<={t}"] = float((g <= t + 1e-12).mean())
    for t in RV_TH_BAD: o[f"ret_g>{t}"] = _ret(ph, pi, g > t); o[f"share_g>{t}"] = float((g > t).mean())
    o["ret_d=1"] = _ret(ph, pi, d == 1); o["ret_d<=1"] = _ret(ph, pi, d <= 1)
    RV3.append(o)
_h12 = json.load(open(HW12_FILE)); _hx = json.load(open(HWX_FILE))
for mode, lab in (("suppressed", "dd_twirl"), ("unsuppressed", "none")):
    for r in HWM: rv_resolve("hardware", "first", "main", lab, r, bits_to_idx(_h12["runs"][mode][f"{r['seed']}|transferred"]))
for fam, circs in (("main", HWM), ("Wishart", HWP)):
    for key, runs in _hx["runs"].items():
        if key.startswith(fam + "|"):
            for r in circs: rv_resolve("hardware", "second", fam, key.split("|")[1], r, bits_to_idx(runs[str(r["seed"])]))
if RV_DEVICE_SIM and fake is not None:
    from qiskit.transpiler import generate_preset_pass_manager
    _pmf = generate_preset_pass_manager(optimization_level=3, backend=fake, seed_transpiler=11); _simd = AerSimulator.from_backend(fake, method="statevector")
    for r in HWM:
        _mem = _simd.run(_pmf.run(r["circ"]), shots=HW12_SHOTS, memory=True, seed_simulator=r["seed"]).result().get_memory()
        rv_resolve("device simulation", "first", "main", "model", r, bits_to_idx([m_.replace(" ", "") for m_ in _mem]))
dRV3 = pd.DataFrame(RV3); dRV3.to_csv("RV3_hw_threshold_retention.csv", index=False)
cols_ = ["eta"] + [f"ret_g<={t}" for t in RV_TH] + [f"ret_g>{t}" for t in RV_TH_BAD] + ["ret_d=1", "ret_d<=1"]
print("Medians over instances:"); print(dRV3.groupby(["source", "submission", "family", "mode"], sort=False)[cols_].median().round(3).to_string())
hw_ = dRV3[dRV3.source == "hardware"]
print("\nInstances (of 10) with retention below eta, per threshold; and with the {g>1} deficit retained above eta:")
for k_, g_ in hw_.groupby(["submission", "family", "mode"], sort=False):
    print(" ", k_, {t: int((g_[f"ret_g<={t}"] < g_.eta).sum()) for t in RV_TH}, "| {g>1}:", int((g_["ret_g>1.0"] > g_.eta).sum()),
          "| d<=1:", int((g_["ret_d<=1"] < g_.eta).sum()))
mf_ = hw_[hw_.family == "main"]
print("median share of bitstrings: g<=0.1", round(mf_["share_g<=0.1"].median(), 4), " g<=0.5", round(mf_["share_g<=0.5"].median(), 4), " g>1", round(mf_["share_g>1.0"].median(), 4))
if (dRV3.source == "device simulation").any():
    M_ = dRV3[dRV3.source == "device simulation"].sort_values("seed")
    rec_ = pd.read_csv("R29_hardware_n12_retention.csv"); rec_ = rec_[(rec_.angles == "transferred") & (rec_.condition == "device simulation")].set_index("seed").eta
    t_ = paired(M_["ret_g<=0.0"].values, M_.eta.values, record=False)
    _rng = np.random.default_rng(5); _meds = []
    for _ in range(4000):
        ix = _rng.integers(0, len(M_), len(M_))
        _meds.append(np.median((_rng.binomial(HW12_SHOTS, M_.q.values[ix]) / HW12_SHOTS - _U) / (M_.q_ideal.values[ix] - _U)))
    print(f"\nDevice-matched model (re-executed): largest |eta - recorded eta| = {np.max(np.abs(M_.eta.values - rec_[M_.seed].values)):.4f}; "
          f"mean eta {M_.eta.mean():.3f} (recorded {rec_.mean():.3f}); median eta {M_.eta.median():.3f}")
    print(f"  median q {M_.q.median():.4f}; median eta_q {M_['ret_g<=0.0'].median():.3f} [{np.quantile(_meds, .025):.3f}, {np.quantile(_meds, .975):.3f}]; "
          f"median TTS99 {np.median(np.log(0.01) / np.log(1 - M_.q.values)):.0f}; counts median {M_.count_opt.median()} (min {M_.count_opt.min()})")
    print(f"  eta_q below eta on {int((M_['ret_g<=0.0'] < M_.eta).sum())} of {len(M_)}; signed-rank P={t_['P']:.3f}; median eta/eta_q {np.median(M_.eta / M_['ret_g<=0.0']):.2f}")
    f1_ = hw_[(hw_.submission == "first") & (hw_.family == "main")]
    print(f"  hardware (first submission) eta_q below the model's by a factor {M_['ret_g<=0.0'].median() / f1_.groupby('mode')['ret_g<=0.0'].median().max():.1f} "
          f"to {M_['ret_g<=0.0'].median() / f1_.groupby('mode')['ret_g<=0.0'].median().min():.1f}; "
          f"mean eta below the recorded model mean by {rec_.mean() / f1_.groupby('mode').eta.mean().max():.1f} to {rec_.mean() / f1_.groupby('mode').eta.mean().min():.1f}")
```

### 50.4 Circuit sampler: random stream shared across calls

Optimized QAOA ($p=2$, 23 evaluations of 32 shots and a 32-shot read-out, circuit sampler) on the first ADMM subproblem of test instances 0 to 4, with every sampler call recorded: whether the read-out at the returned angles repeats an earlier evaluation shot for shot, and how many distinct bitstrings the 768 shots contain.

```python
RV4 = []; _orig_run = SamplerV2.run
for s in range(5):
    Qp, h = first_zupdate(N, s); st = exact_stats(Qp, h, keep=True); _calls = []
    def _rec(self, pubs, shots=None, _o=_orig_run):
        res = _o(self, pubs, shots=shots); _calls.append((np.array(pubs[0][1]).copy(), shots, res.result()[0].data.meas.get_bitstrings())); return res
    SamplerV2.run = _rec
    try:
        out = run_qaoa(Qp, h, st["C"], reps=2, shots=SHOTS, maxiter=MAXITER, seed=s * 1000)
    finally:
        SamplerV2.run = _orig_run
    at_x = [i for i, c in enumerate(_calls) if np.allclose(c[0], out["x"]) and c[1] == SHOTS]
    RV4.append({"instance": s, "calls": len(_calls), "readout_repeats_evaluation": any(_calls[i][2] == _calls[at_x[-1]][2] for i in at_x[:-1]),
                "distinct_bitstrings": len({b for c in _calls for b in c[2]}), "shots": sum(c[1] for c in _calls)})
dRV4 = pd.DataFrame(RV4); dRV4.to_csv("RV4_readout_replay.csv", index=False); print(dRV4.to_string(index=False))
```

### 50.5 Planted-family angles under the symmetries of the QAOA output distribution

Time reversal $(\boldsymbol\beta,\boldsymbol\gamma)\mapsto(-\boldsymbol\beta,-\boldsymbol\gamma)$ leaves the output distribution unchanged; a shift $\beta_j\mapsto\beta_j+\pi/2$ maps the state of a complement-symmetric cost to its complement. Smallest componentwise distance between the $n=12$ and $n=16$ planted-family angles over all such combinations.

```python
a12 = np.asarray(TRANSFER_W[2]); a16 = np.asarray(json.load(open("E38_angles_n16.json"))["Wishart"]["angles"])
best_ = None
for sg in (1.0, -1.0):
    for k1 in range(-2, 3):
        for k2 in range(-2, 3):
            v = sg * a12 + np.array([k1 * np.pi / 2, k2 * np.pi / 2, 0.0, 0.0]); d = np.abs(v - a16)
            if best_ is None or d.max() < best_[0]: best_ = (d.max(), sg, k1, k2, d)
print(f"n=12 angles {np.round(a12, 3)}; n=16 angles {np.round(a16, 3)}")
print(f"closest image: sign {best_[1]:+.0f}, beta shifts {best_[2]}, {best_[3]} x pi/2; componentwise distance {np.round(best_[4], 3)}; largest {best_[0]:.3f} rad")
m12 = np.asarray(TRANSFER[("admm", 2)]); m16 = np.asarray(json.load(open("E38_angles_n16.json"))["main"]["angles"])
print(f"main family: largest component difference between n=12 and n=16 angles {np.abs(m12 - m16).max():.4f} rad")
pd.DataFrame([{"family": "Wishart", "largest_distance": best_[0], **{f"d{j}": best_[4][j] for j in range(4)}}]).to_csv("RV5_angle_symmetry.csv", index=False)
```
